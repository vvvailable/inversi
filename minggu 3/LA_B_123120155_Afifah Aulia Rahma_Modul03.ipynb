"><b>1.1 Latar Belakang</b></h2>
Metode inversi merupakan salah satu pendekatan penting dalam geofisika untuk memperoleh model bawah permukaan berdasarkan data observasi lapangan. Dalam praktiknya, proses inversi dilakukan dengan mencari parameter model yang menghasilkan respons sintetik mendekati data hasil pengukuran. Namun, permasalahan inversi geofisika umumnya bersifat tidak unik, sehingga satu set data observasi dapat dijelaskan oleh lebih dari satu model bawah permukaan. Ketidakunikan solusi menjadi semakin kompleks pada kasus under-determined, yaitu ketika jumlah parameter model lebih banyak dibandingkan jumlah data observasi. Kondisi tersebut menyebabkan ruang solusi menjadi sangat luas dan solusi inversi menjadi tidak stabil. Selain itu, keberadaan noise data dan keterbatasan observasi juga dapat memengaruhi kualitas model hasil inversi.

Untuk mengatasi permasalahan tersebut, diperlukan informasi tambahan atau a priori information yang berfungsi sebagai constraint dalam proses inversi. Informasi a priori dapat berupa model awal, batas parameter, maupun asumsi karakter model tertentu. Selain itu, penerapan regularisasi melalui parameter damping digunakan untuk menstabilkan solusi sehingga model yang diperoleh tidak hanya sesuai terhadap data, tetapi juga tetap realistis secara matematis dan fisis. Dalam proses inversi, penggunaan model referensi juga berperan penting sebagai acuan awal untuk mengarahkan iterasi solusi. Di samping itu, constraint flatness dan smoothness digunakan untuk mengontrol karakter model hasil inversi. Constraint flatness menghasilkan model yang relatif seragam antarparameter, sedangkan smoothness menghasilkan perubahan model yang lebih gradual dan halus. Oleh karena itu, pada praktikum Modul 3 ini dilakukan percobaan mengenai penerapan informasi a priori, regularisasi, model referensi, flatness, dan smoothness dalam penyelesaian masalah inversi linear untuk memahami pengaruh constraint terhadap kestabilan solusi dan kualitas model hasil inversi pada berbagai kondisi permasalahan geofisika.

<h3 style="text-align:left; font-size:13pt"><b>1.3.1 Konsep Dasar Inversi dalam Geofisika</b></h3>

<p>
Geofisika merupakan ilmu yang mempelajari kondisi bawah permukaan bumi berdasarkan pengukuran parameter fisis di permukaan. Data yang diperoleh dari pengukuran geofisika merupakan respons dari kondisi geologi bawah permukaan yang dipengaruhi oleh variasi sifat fisis material, seperti densitas, resistivitas, suseptibilitas magnetik, maupun kecepatan rambat gelombang seismik yang menghasilkan anomali untuk diinterpretasikan menjadi model bawah permukaan. Dalam geofisika, interpretasi data dilakukan melalui proses pemodelan. Model merupakan representasi matematis dari kondisi bawah permukaan yang dinyatakan dalam parameter fisis dan geometri tertentu. Melalui model, respons teoritis dapat dihitung menggunakan persamaan fisika sesuai metode geofisika yang digunakan.
</p>

<p>
Pemodelan geofisika terdiri atas dua proses utama, yaitu <i>forward modeling</i> dan <i>inverse modeling</i>. <i>Forward modeling</i> merupakan proses menghitung respons data berdasarkan model bawah permukaan yang telah diketahui. Sebaliknya, <i>inverse modeling</i> atau inversi merupakan proses memperkirakan parameter model berdasarkan data observasi. Dalam inversi, parameter model dicari sedemikian rupa sehingga respons hasil perhitungan memiliki kesesuaian terbaik dengan data pengamatan. Oleh karena itu, inversi sering disebut sebagai proses <i>data fitting</i>. Secara matematis, proses inversi dilakukan dengan meminimalkan fungsi objektif atau fungsi <i>misfit</i> yang menyatakan selisih antara data observasi dan data hasil perhitungan. Model optimum diperoleh ketika nilai <i>misfit</i> mencapai minimum.
</p>

<h3 style="text-align:left; font-size:13pt"><b>1.3.2 Inversi Non Linear</b></h3>

<p>
Inversi non-linear merupakan metode inversi yang digunakan ketika hubungan antara data observasi dan parameter model tidak dapat dinyatakan dalam bentuk persamaan linear. Secara umum hubungan tersebut dinyatakan sebagai:
</p>

$$
d = g(m)
$$

<p>
dengan d merupakan data observasi, m adalah parameter model, dan g(m) merupakan fungsi <i>forward modeling</i> non-linear.
</p>

<p>
Pada inversi non-linear, perubahan kecil pada parameter model tidak selalu menghasilkan perubahan linear pada data. Hal ini menyebabkan solusi model tidak dapat diperoleh secara langsung menggunakan invers matriks sederhana seperti pada inversi linear, melainkan memerlukan proses iteratif atau pencarian dalam ruang model. Karakteristik utama inversi non-linear adalah kemungkinan adanya banyak solusi lokal (<i>local minimum</i>) maupun solusi global (<i>global minimum</i>). Kondisi ini terjadi karena fungsi objektif dapat memiliki lebih dari satu nilai minimum.
</p>

<h3 style="text-align:left; font-size:13pt"><b>1.3.3 Fungsi Objektif dan Misfit Error</b></h3>

<p>
Dalam pemodelan inversi, tujuan utama yang ingin dicapai adalah memperoleh model yang mampu menghasilkan respons teoritis yang sesuai dengan data observasi.
</p>

Residual dinyatakan sebagai:

$$
r = d - g(m)
$$

Fungsi objektif:

$$
E = ||d - g(m)||^2
$$

Nilai <i>misfit error</i> menggunakan <i>Root Mean Square Error</i> (<i>RMSE</i>):

$$
RMSE = \sqrt{\frac{1}{N}\sum_{i=1}^{N}(d_i - g_i(m))^2}
$$

dengan:

<ul>
<li>N = jumlah data pengamatan</li>
<li>dᵢ = data observasi ke-i</li>
<li>gᵢ(m) = respons model pada data ke-i</li>
</ul>

<p>
Nilai <i>misfit</i> yang kecil menunjukkan bahwa model mampu merepresentasikan data observasi dengan baik.
</p>
<h3 style="text-align:left; font-size:13pt"><b>1.3.4 Informasi A Priori</b></h3>
Seperti yang diketahui, informasi “a priori” merupakan kuantifikasi ekspetasi atau harapan mengenai karakteristik solusi yang tidak didasarkan pada data pengukuran. Pengaruh dari informasi “a priori” adalah untuk mengurangi hasil inversi yang tidak proporsional. Pada dasarnya, informasi “a priori” digunakan ketika jumlah data observasi (N) lebih sedikit dibandingkan jumlah parameter model yang dicari (M). Hal ini biasa dikenal dengan istilah under-determined. Pada beberapa kasus, informasi “a priori” terkadang bersifat subjektif sehingga solusi inversinya menjadi bias. Selain itu, jumlah solusi inversi pada kasus under-determined adanya non-unique. Untuk menghindari masalah-masalah tersebut, maka solusi inversi diasumsikan sederhana. Secara matematis, persamaan informasi “a priori” dapat ditulis sebagai berikut

$$
L = m^T m = \sum_{i=1}^{M} m_i^2
$$

Metode inversi yang baik hendaknya dapat memperhitungkan tingkat kesalahan data atau ketelitian data dalam proses penyelesaian inversi. Dengan demikian, solusi yang diperoleh secara objektif sesuai dengan kualitas data. Umumnya, pengaruh data dengan kesalahan cukup besar (atau tingkat ketelitian rendah) harus diminimalisasi agar hasil inversinya merupakan representasi data dengan tingkat ketelitian yang baik. Hal tersebut dapat dilakukan melalui pembobotan data dalam penyelesaian masalah inversi. Pada dasarnya pemberian bobot relatif pada data sehingga data dengan kesalahan besar tidak akan berpengaruh pada solusi inversi. Dengan kata lain, solusi inversi akan lebih ditentukan oleh data yang berkualitas baik. Solusi inversi dikatakan sederhana jika L bernilai minimum. Dengan menggunakan konsep ini, maka diperoleh solusi inversi linear under-determined sebagai berikut :
$$
m = G^T (GG^T)^{1} d
$$
Meskipun demikian, banyak kasus permasalahan dalam inversi geofisika yang merupakan gabungan antara under-determined dan over-determined sehingga disebut mixed-determined. Solusi akhir inversi linear mixed-determined dapat ditulis dalam persamaan berikut :
$$
m = (G^T G + \epsilon^2 I(GG^T)^{-1} G^T d)
$$
 Solusi inversi di atas disebut juga sebagai inversi linear teredam (damped linear inversion) karena digunakan untuk meredam ketidakstabilan solusi yang timbul pada perhitungan akibat keterbatasan pada bagian under-determined. Nilai ε^2 disebut sebagai faktor redaman atau parameter regularisasi. Sementara itu, model referensi merupakan model yang mendekati atau mirip dengan model tertentu. Model referensi ditentukan berdasarkan informasi “a priori” yang tersedia. Pada dasarnya, model referensi dapat terjadi dalam dua kondisi, yaitu dalam kondisi flat dan dalam kondisi smooth. Salah satu contoh model referensi adalah model rata-rata yang disusun dengan harga konstan. Solusi inversi dengan model referensi ditulis dalam persamaan berikut.
 $$
m = m + \left[G^T G + \epsilon^2 I\right]^{-1} G^T \left[d - Gm\right]
$$
 Berdasarkan persamaan di atas, terlihat bahwa solusi inversi (model) merupakan model referensi yang diperbarui oleh faktor koreksi. Jika model yang diharapkan adalah model yang bersifat flat, maka minimumkan selisih model yang berdekatan sehingga gradien nilai model pada lokasi yang berdekatan tidak terlampau jauh. Secara matematis, model referensi dalam kondisi flat ditulis pada persamaan berikut :
 $$
L(m) = \sum_{j=2}^{M} (m_j - m_{j-1})^2
$$

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import copy
from numpy.linalg import inv

# Station Position
x = [0, 10, 20,30]
z = [0, 0, 0, 0]
n = len(x)

# True hypocenter
x0 = 12
z0 = 10
t0 = 0

# Model Velocity 1 Layer
v = 7

# Forward Modelling to calculate traveltime from hypocenter to stataion
t = []
for i in range(n):
  t.append((1/v) * np.sqrt((z[i] - z0)**2 + (x[i] - x0)**2))

# Convert t to a numpy array after it's fully populated
t = np.array(t)

# Add Noise to travel time data
noise = np.random.rand(n) * 0.03 * np.mean(t)
ti = t + noise

# Guess the initial position
M = [15,30]
init_position = copy.deepcopy(M)
M1 = []

# Create weighted matrix
W = np.eye(n)

# Assign Weighted matrix
W[n-1][n-1] = 0.1
# W[i][i] = abs(ti[i] - (ti[i] = np.mean(ti)))

print(W)

xhipo = [init_position[0]]
yhipo = [init_position[1]]
Erms = []

# Inversion without weighted factor
for niter in range(1000): #Ganti jumlah iterasi nnt
   gm = []
   dgm_dm1 = []
   dgm_dm2 = []

   for i in range(n):
    gm.append(t0 + (1/v) * (np.sqrt((x[i] - M[0])**2 +(z[i] - M[1])**2)))
    dgm_dm1.append((1/v) * (-1*(x[i] - M[0]))/(np.sqrt((x[i] - M[0])**2 +(z[i] - M[1])**2)))
    dgm_dm2.append((1/v) * (-1*(z[i] - M[1]))/(np.sqrt((x[i] - M[0])**2 +(z[i] - M[1])**2)))

   J = np.zeros((n, 2))

   for i in range(n):
     J[i][0] = dgm_dm1[i].item()
     J[i][1] = dgm_dm2[i].item()

   Mo = copy.deepcopy(M)

   # Add model pertubation from inv to Mo
   Mo = np.reshape(Mo, (2,1))
   ti = np.reshape(ti, (n,1))
   gm = np.reshape(gm, (n,1))

   Model = Mo + inv(J.T @ J) @ J.T @ (ti - gm)
   M = [Model[0][0], Model[1][0]]

   Data = Model.tolist()
   xhipo.append(Data[0][0])
   yhipo.append(Data[1][0])


   # Calculate rms error
   dt = ti - gm # (np.mat(ti) - np.mat(gm))
   Erms.append(np.sqrt(np.average(dt**2)))

plt.figure(figsize=(7,5))
plt.plot(15, 30, 'o', label='Initial Position', color='black')
plt.plot(xhipo, yhipo, 'o', label='Hypocenter Inversion')
plt.plot(xhipo, yhipo, '-', label='Hypocenter Inversion')
plt.plot(x0, z0, 'o', alpha=0.5, color='red', label='True Hypocenter')
plt.title('Hypocenter Determination_155')
plt.xlabel('X Coordinate')
plt.ylabel('Y Coordinate')
plt.legend()
plt.grid()

# Plot kurva error RMS terhadap iterasi
plt.figure(figsize=(7,5))
plt.plot(Erms, color='blue')
plt.title('Kurva RMS Error Vs N Iterasi_155')
plt.ylabel('RMS Error')
plt.xlabel('N Iterasi')
plt.grid()
plt.show()

print(Erms)
```

```python
#Dengan Bobot Matriks W
import numpy as np
import matplotlib.pyplot as plt
import copy
from numpy.linalg import inv

# Station Position
x = [0, 10, 20, 30]
z = [0, 0, 0, 0]
n = len(x)

# True Hypocenter
x0 = 12
z0 = 10
t0 = 0

# Model Velocity 1 Layer
v = 7

# Forward modelling to calculate traveltime from hypocenter to station
t = []
for i in range(n):
    t.append((1/v) * np.sqrt((z[i] - z0)**2 + (x[i] - x0)**2))

# Add noise to traveltime data
noise = np.random.rand(n) * 0.03 * np.mean(t)
ti = t + noise

print(t)

# Add outlier data
ti[n-1] = ((1/v) * np.sqrt((z[n-1] - z0)**2 + (x[n-1] - x0)**2)) + np.random.rand() * 0.5 * np.mean(t)

# Guess the initial position
M = [15, 30]
init_position = copy.deepcopy(M)
M1 = []

# Create weighted matrix
W = np.eye(n)

#Assign weight matrix
W[n-1][n-1] = 0.1
#W[i][i] = abs(ti[i] - (ti[i]- np.mean(ti)))

print (W)

xhipo = [init_position[0]]
yhipo = [init_position[1]]
Erms = []

#Inversion without weighted factor
for niter in range(1000):           #ganti jumlah iterasi
    gm = []
    dgm_dm1 = []
    dgm_dm2 = []

    for i in range(n):
        gm.append(t0 + (1/v) * (np.sqrt((x[i] - M[0])**2 + (z[i] - M[1])**2)))
        dgm_dm1.append((1/v) * (-1*(x[i] - M[0]))/(np.sqrt((x[i] - M[0])**2 + (z[i] - M[1])**2)))
        dgm_dm2.append((1/v) * (-1*(z[i] - M[1]))/(np.sqrt((x[i] - M[0])**2 + (z[i] - M[1])**2)))

    J = np.zeros((n, 2))

    for i in range(n):
        J[i][0] = dgm_dm1[i].item()
        J[i][1] = dgm_dm2[i].item()

    Mo = copy.deepcopy(M)

    # Add model perturbation from inv to Mo
    Mo = np.reshape(Mo, (2,1))
    ti = np.reshape(ti, (n,1))
    gm = np.reshape(gm, (n,1))

    Model = np.asmatrix(Mo) + np.asmatrix(inv(np.asmatrix(J).transpose() * np.asmatrix(J))) * np.asmatrix(J.transpose()) * np.asmatrix(W) * (np.asmatrix(ti) - np.asmatrix(gm))
    M = [Model[0][0], Model[1][0]]

    Data = Model.tolist()
    xhipo.append(Data[0][0])
    yhipo.append(Data[1][0])

    # Calculate rms error
    dt = ti - gm # (np.mat(ti) - np.mat(gm))
    Erms.append(np.sqrt(np.average(dt**2)))

plt.figure(figsize=(7,5))
plt.plot(15, 30, 'o', label='Initial Position', color='black')
plt.plot(xhipo, yhipo, 'o', label='Hypocenter Inversion')
plt.plot(xhipo, yhipo, '-', label='Hypocenter Inversion')
plt.plot(x0, z0, 'o', alpha=0.5, color='red', label='True Hypocenter')
plt.title('Hypocenter Determination_123120155')
plt.xlabel('X Coordinate')
plt.ylabel('Y Coordinate')
plt.legend()
plt.grid()

# Plot kurva error RMS terhadap iterasi
plt.figure(figsize=(7,5))
plt.plot(Erms, color='blue')
plt.title('Kurva RMS Error Vs N Iterasi_123120155')
plt.ylabel('RMS Error')
plt.xlabel('N Iterasi')
plt.grid()
plt.show()

print(Erms)
```

```python

#Inversi Linear Under-determined
import numpy as np
import matplotlib.pyplot as plt
from numpy.linalg import inv

# Data Observasi
theta_deg = np.array([10, 25])     # sudut datang (derajat)
R_obs = np.array([0.1523, 0.0947]) # koefisien refleksi observasi

# Ubah ke radian
theta = np.radians(theta_deg)

# Forward Modelling Matrix (G)
G = []

for i in range(len(theta)):
    
    g1 = 1
    g2 = np.sin(theta[i])**2
    g3 = (np.sin(theta[i])**2) * (np.tan(theta[i])**2)
    
    G.append([g1, g2, g3])

G = np.array(G)

# Data vector
d = np.reshape(R_obs,(len(R_obs),1))

print("Matrix Kernel (G):")
print(G)

print("Data Observasi:")
print(d)

# Inversi Linear Under-determined
# Minimum Norm Solution
GT = G.transpose()

Model = GT @ inv(G @ GT) @ d

A = Model[0][0]
B = Model[1][0]
C = Model[2][0]

print("\nParameter Model:")
print("A =",A)
print("B =",B)
print("C =",C)


# Forward modelling hasil inversi
theta_model = np.linspace(0,40,100)
theta_model_rad = np.radians(theta_model)

R_model = A + B*(np.sin(theta_model_rad)**2) + \
          C*(np.sin(theta_model_rad)**2)*(np.tan(theta_model_rad)**2)


# Plot Hasil
plt.figure(figsize=(7,5))

plt.plot(theta_model,R_model,label='Model Inversi')
plt.plot(theta_deg,R_obs,'o',color='red',label='Data Observasi')

plt.xlabel('Sudut Datang (derajat)')
plt.ylabel('Koefisien Refleksi')
plt.title('Inversi Linear Under-determined_123120155')
plt.legend()
plt.grid()

plt.show()
```

<p>
Pada kode pertama, inversi dilakukan tanpa matriks bobot yang besar untuk menentukan lokasi hiposenter gempa. Hiposenter berada di koordinat (12, 10) dengan posisi awal tebakan di (15, 30). Dari Gambar 1, terlihat bahwa proses iterasi berhasil mengkonvergensi posisi hiposenter mendekati titik sebenarnya. Lintasan pergerakan model dari titik awal (15, 30) menuju titik akhir yang sangat dekat dengan (12, 10) menunjukkan bahwa algoritma Gauss-Newton yang digunakan bekerja dengan baik dalam meminimumkan residual antara traveltime observasi dan traveltime kalkulasi. Dari Gambar 2, kurva RMS error menunjukkan penurunan yang sangat tajam pada iterasi-iterasi awal, kemudian langsung mendatar mendekati nol setelah sekitar 10-20 iterasi pertama. Ini menandakan konvergensi yang cepat dan stabil. Nilai RMS error akhir yang sangat kecil mengindikasikan bahwa model hasil inversi mampu menghasilkan data traveltime observasi dengan baik. Hal ini karena pada kode pertama tidak ada outlier yang ditambahkan ke data, sehingga noise yang ada hanya berupa gangguan kecil sebesar 3% dari rata-rata traveltime.</p>
<p>Pada kode kedua, ditambahkan dua perubahan yaitu outlier pada data stasiun ke-4 dan matriks bobot W dengan nilai W[3][3] = 0.1. Outlier ditambahkan dengan noise sebesar 50% dari rata-rata traveltime, jauh lebih besar dibanding noise normal 3%. Tujuannya adalah mensimulasikan kondisi nyata di mana salah satu stasiun merekam data yang tidak reliable. Dari Gambar 3, lintasan konvergensi terlihat lebih panjang dan berliku dibanding kode pertama. Model bergerak dari (15, 30) melalui beberapa titik intermediat sebelum akhirnya mendekati posisi sekitar (12, 10). Meskipun hasil akhirnya belum tepat berimpit dengan hiposenter sejati, arah pergerakannya sudah menuju lokasi yang benar. Pemberian bobot kecil W = 0.1 pada stasiun ke-4 yang mengandung outlier terbukti membantu mengurangi pengaruh data yang tidak akurat tersebut terhadap hasil inversi. Dari Gambar 4, kurva RMS error menunjukkan pola berbeda, nilai RMS error turun cepat di awal namun kemudian stabil pada nilai sekitar 0.3 dan tidak mendekati nol seperti pada kode pertama. Ini disebabkan adanya outlier, meskipun bobot sudah diberikan, residual tidak bisa ditekan hingga nol karena data outlier tetap ada dalam sistem persamaan meski dengan kontribusi yang dikurangi. Nilai RMS error yang tersisa ini mencerminkan ketidakcocokan yang inheren antara model dan data yang terkontaminasi outlier. Perbandingan antara kode pertama dan kedua menunjukkan, tanpa bobot dan tanpa outlier, hasil inversi sangat baik dengan RMS mendekati nol. Dengan adanya outlier tanpa bobot, hasil inversi akan terdistorsi. Pemberian bobot kecil pada data yang dicurigai mengandung error terbukti dapat mengurangi distorsi tersebut, meskipun tidak sepenuhnya menghilangkannya.</p>
<p>Pada kode ketiga, dilakukan inversi linear untuk menentukan parameter AVO (Amplitude Versus Offset) menggunakan persamaan Shuey yang disederhanakan. Sistem ini bersifat under-determined karena hanya tersedia 2 data observasi yaitu koefisien refleksi pada sudut 10° dan 25°, sementara ada 3 parameter yang harus ditentukan yaitu A, B, dan C. Pendekatan yang digunakan adalah minimum norm solution dengan formula M = Gᵀ(GGᵀ)⁻¹d. Solusi ini memilih nilai parameter yang paling kecil normanya di antara seluruh kemungkinan solusi yang memenuhi data. Dari Gambar 5, model hasil inversi berupa kurva yang melewati kedua titik observasi dengan baik, menunjukkan bahwa data fit tercapai. Kurva menunjukkan nilai koefisien refleksi yang menurun seiring bertambahnya sudut datang, dari sekitar 0.16 pada sudut 0° hingga mendekati negatif pada sudut di atas 38°. Pola ini fisiknya masuk akal karena koefisien refleksi memang umumnya berubah dengan sudut datang tergantung pada kontras impedansi antar lapisan. Namun solusi under-determined juga, terdapat infinitely many solutions yang bisa memenuhi kedua titik data tersebut. Solusi minimum norm hanya memilih yang paling "kecil" secara norma, bukan yang paling benar secara fisik. Untuk mendapatkan hasil yang lebih dapat dipercaya, idealnya diperlukan lebih banyak data observasi pada berbagai sudut datang agar sistem menjadi over-determined dan solusinya lebih terkonstrain.</p>
