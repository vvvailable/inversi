        1.1 Latar Belakang
    </p>
    <p style="text-align: justify;">
       Dalam kajian geofisika, khususnya pada metode inversi, tujuan utama analisis adalah memperoleh model bawah permukaan yang mampu menjelaskan data pengamatan secara optimal. Namun, permasalahan inversi pada umumnya bersifat ill-posed, yaitu memiliki solusi yang tidak unik, tidak stabil terhadap kesalahan data, dan sering kali sulit ditentukan secara langsung. Kondisi ini menyebabkan model hasil inversi dapat menghasilkan banyak kemungkinan solusi yang berbeda meskipun menggunakan data yang sama. Oleh karena itu, diperlukan suatu pendekatan tambahan untuk mengontrol solusi inversi agar lebih stabil dan realistis secara geologi (Santosa, 2010). Salah satu pendekatan yang umum digunakan dalam proses inversi adalah dengan menambahkan informasi a priori atau informasi awal mengenai model bawah permukaan. Informasi a priori berfungsi sebagai batasan atau kendala (constraint) yang membantu proses inversi menghasilkan model yang lebih masuk akal secara fisik dan geologis. Informasi ini dapat berasal dari berbagai sumber seperti data sumur, pengetahuan geologi regional, maupun asumsi matematis tertentu mengenai karakteristik model yang diharapkan (Grandis, 2009).
    </p>
      <p style="text-align: justify;">
      Dalam praktik inversi geofisika, terdapat beberapa pendekatan dalam membangun model referensi berdasarkan informasi a priori. Dua pendekatan yang sering digunakan adalah model referensi flatness dan model referensi smoothness. Pendekatan flatness mengasumsikan bahwa nilai parameter model cenderung konstan atau memiliki variasi yang sangat kecil antar sel model. Dengan kata lain, metode ini berupaya meminimalkan perubahan nilai parameter model sehingga menghasilkan model yang relatif seragam. Pendekatan ini sering digunakan ketika informasi detail mengenai variasi parameter bawah permukaan masih terbatas (Grandis, 2009).Sementara itu, pendekatan smoothness menekankan pada kehalusan perubahan parameter model antar sel model. Metode ini tidak memaksakan nilai parameter menjadi konstan, tetapi mengontrol agar perubahan nilai parameter antar sel model berlangsung secara bertahap dan tidak menunjukkan perubahan yang terlalu tajam. Pendekatan smoothness banyak digunakan dalam inversi geofisika karena mampu menghasilkan model yang lebih stabil serta tetap merepresentasikan variasi geologi bawah permukaan secara lebih realistis (Santosa, 2010)Pemilihan jenis model referensi yang tepat sangat penting karena akan mempengaruhi karakteristik model hasil inversi. Oleh karena itu, pemahaman mengenai konsep informasi a priori serta penerapan model referensi flatness dan smoothness menjadi bagian penting dalam proses inversi geofisika agar model yang dihasilkan tidak hanya sesuai dengan data pengamatan, tetapi juga konsisten dengan kondisi geologi yang sebenarnya.
    </p>

        1.2 Dasar Teori
    </p>
    <p style="text-align: justify;">
<p style="text-align: justify;">
        1.2.1 Konsep Informasi “A Priori” dalam Inversi
    </p>
    <p style="text-align: justify;">
    Konsep informasi a priori dalam inversi geofisika merupakan salah satu komponen penting untuk mengatasi permasalahan non-keunikan (non-uniqueness) dan ketidakstabilan solusi. Dalam banyak kasus, solusi inversi tidak bersifat tunggal karena keterbatasan data dan adanya noise, sehingga diperlukan informasi tambahan di luar data observasi untuk membatasi ruang solusi. Informasi a priori ini dapat berupa model awal, batasan fisik parameter, maupun asumsi geologi yang relevan dengan kondisi bawah permukaan. Dengan memasukkan informasi tersebut, proses inversi menjadi lebih terarah dan menghasilkan model yang lebih realistis secara geologi. Secara konseptual, inversi geofisika bertujuan untuk memperoleh model bawah permukaan berdasarkan data pengamatan, namun hubungan antara data dan model seringkali bersifat tidak langsung dan kompleks. Oleh karena itu, pendekatan inversi umumnya dirumuskan sebagai optimasi yang menggabungkan kesesuaian data (data misfit) dan kendala tambahan berupa informasi a priori. Hal ini sejalan dengan pernyataan bahwa metode inversi digunakan untuk menentukan parameter model seperti posisi, densitas, atau kecepatan berdasarkan data observasi, yang memerlukan asumsi atau parameter awal untuk mempercepat dan menstabilkan proses pencarian solusi (Al Hakim, Y. U., Arman, Y., & Zulfian. 2023). 
    <p style="text-align: justify;">
 </p>
 <p style="text-align: justify;">
    Dalam praktiknya, informasi a priori dapat diimplementasikan dalam bentuk model referensi, regularisasi, maupun pembobotan parameter. Model referensi sering digunakan sebagai acuan awal yang akan diperbaiki melalui iterasi inversi. Selain itu, teknik regularisasi seperti damping, smoothness, atau flatness constraint merupakan bentuk formal dari informasi a priori yang bertujuan mengontrol kompleksitas model agar tidak terlalu berosilasi. Penggunaan kendala ini penting terutama pada kondisi geologi kompleks, di mana data saja tidak cukup untuk memberikan gambaran bawah permukaan yang akurat (Jatnika, J., Nugraha, A. D., & Wandono. 2015). Lebih lanjut, dalam studi inversi seperti tomografi seismik atau relokasi hiposenter, pemilihan model awal (initial model) sebagai bagian dari informasi a priori sangat mempengaruhi hasil akhir inversi. Model awal yang kurang representatif dapat menyebabkan solusi konvergen ke minimum lokal yang tidak sesuai dengan kondisi sebenarnya. Oleh karena itu, integrasi informasi geologi, data sumur, atau hasil penelitian sebelumnya menjadi krusial dalam membangun model a priori yang baik. Hal ini menunjukkan bahwa kualitas hasil inversi tidak hanya bergantung pada data, tetapi juga pada ketepatan informasi tambahan yang digunakan dalam proses inversi (Surya, T. M. A., Utama, W., Syaifuddin, F., & Novitasari, L. 2016)
    </p>
<p style="text-align: justify;">
        1.2.2 Regularisasi dan Fungsi Objektif
    </p>
    <p style="text-align: justify;">
    Regularisasi dan fungsi objektif merupakan dua komponen utama dalam formulasi inversi geofisika yang digunakan untuk memperoleh solusi model yang stabil dan realistis. Fungsi objektif (objective function) pada dasarnya adalah suatu ukuran kuantitatif yang menyatakan tingkat kesesuaian antara data hasil observasi dengan data hasil perhitungan dari model. Umumnya, fungsi objektif dirumuskan dalam bentuk kuadrat selisih (least squares) antara data observasi dan data sintetik, sehingga proses inversi dilakukan dengan meminimalkan nilai fungsi tersebut. Pendekatan ini banyak digunakan karena sifatnya yang matematis sederhana dan mampu memberikan solusi optimal dalam sistem linear, meskipun tetap rentan terhadap pengaruh noise dan ketidakstabilan jika tidak dikombinasikan dengan teknik tambahan (Putra & Sutarno, 2019). Namun, dalam praktiknya, permasalahan inversi sering bersifat ill-posed, yaitu tidak memiliki solusi tunggal atau sangat sensitif terhadap perubahan kecil pada data. Untuk mengatasi hal tersebut, digunakan teknik regularisasi yang berfungsi sebagai kendala tambahan dalam fungsi objektif. Regularisasi bekerja dengan menambahkan komponen penalti terhadap model agar solusi yang dihasilkan tidak terlalu kompleks atau berosilasi tinggi. Dengan kata lain, regularisasi membantu menyeimbangkan antara kesesuaian data (data misfit) dan kesederhanaan model (model norm). Hal ini sejalan dengan konsep bahwa solusi inversi tidak hanya harus cocok dengan data, tetapi juga harus masuk akal secara fisik dan geologi (Santosa & Nugraha, 2017).
   
</p>
<p style="text-align: justify;">
    Secara matematis, fungsi objektif terregularisasi biasanya dinyatakan sebagai kombinasi antara norma residual data dan norma model, yang dikontrol oleh parameter regularisasi. Parameter ini menentukan seberapa besar pengaruh informasi tambahan terhadap solusi akhir. Jika parameter regularisasi terlalu kecil, solusi cenderung mengikuti data secara berlebihan dan menjadi tidak stabil, sedangkan jika terlalu besar, solusi menjadi terlalu halus dan kehilangan detail penting. Oleh karena itu, pemilihan parameter regularisasi menjadi aspek krusial dalam proses inversi. Beberapa metode seperti trade-off curve atau L-curve sering digunakan untuk menentukan nilai parameter yang optimal (Wahyudi et al., 2020). Dalam aplikasi geofisika, regularisasi dapat diwujudkan dalam berbagai bentuk, seperti damping (pembatasan besar parameter), smoothness constraint (kehalusan model), maupun flatness constraint (kecenderungan model konstan). Pemilihan jenis regularisasi ini bergantung pada tujuan inversi dan karakteristik geologi daerah penelitian. Misalnya, pada pemodelan reservoir, smoothness constraint sering digunakan untuk menghasilkan distribusi parameter yang kontinu, sedangkan pada studi struktur geologi yang tajam, diperlukan regularisasi yang tetap mampu mempertahankan kontras tinggi. Dengan demikian, kombinasi yang tepat antara fungsi objektif dan regularisasi akan menghasilkan model inversi yang tidak hanya sesuai dengan data, tetapi juga representatif terhadap kondisi bawah permukaan (Pratama et al., 2018).
</p>
<p style="text-align: justify;">
        1.2.3 Model Referensi dan Flatness Constraint
    </p>
    <p style="text-align: justify;">
    Model referensi (model reference),  merupakan salah satu bentuk informasi a priori yang digunakan dalam inversi geofisika untuk mengarahkan solusi menuju kondisi yang lebih realistis. Dalam banyak kasus, permasalahan inversi bersifat tidak unik sehingga diperlukan model awal sebagai acuan agar solusi tidak menyimpang jauh dari kondisi geologi yang masuk akal. Model referensi ini biasanya diperoleh dari data pendukung seperti log sumur, interpretasi geologi, atau hasil penelitian sebelumnya. Dalam formulasi fungsi objektif, model referensi dimasukkan sebagai bagian dari kendala (constraint) yang membatasi perubahan model selama proses iterasi inversi berlangsung, sehingga solusi yang dihasilkan tetap stabil dan konvergen (Santosa & Nugraha, 2017). Salah satu pendekatan regularisasi yang sering dikombinasikan dengan model referensi adalah flatness constraint. Konsep ini mengasumsikan bahwa model bawah permukaan memiliki kecenderungan mendekati nilai konstan, sehingga variasi parameter antar sel model diminimalkan. Pendekatan ini sangat sesuai untuk kondisi geologi yang relatif sederhana atau homogen, di mana perubahan sifat fisik batuan tidak terlalu signifikan. Dengan menerapkan flatness constraint, solusi inversi menjadi lebih halus dan tidak dipengaruhi secara berlebihan oleh noise pada data, sehingga interpretasi yang dihasilkan lebih stabil (Pratama et al., 2018).
    </p>
     <p style="text-align: justify;">
    Secara matematis, flatness constraint umumnya direpresentasikan menggunakan operator identitas, yang berarti bahwa penalti diberikan langsung terhadap nilai model itu sendiri dibandingkan dengan model referensi. Dengan demikian, fungsi objektif tidak hanya meminimalkan selisih antara data observasi dan data sintetik, tetapi juga meminimalkan deviasi model terhadap model referensi. Hal ini berbeda dengan pendekatan smoothness constraint yang menggunakan operator turunan untuk mengontrol perubahan antar parameter model. Penggunaan operator identitas menjadikan metode ini lebih sederhana dan efektif untuk menjaga kestabilan solusi, terutama pada data yang terbatas atau memiliki tingkat ketidakpastian tinggi (Wahyudi et al., 2020). Meskipun demikian, penerapan flatness constraint perlu disesuaikan dengan tujuan inversi dan kompleksitas geologi daerah penelitian. Pada kondisi bawah permukaan yang kompleks dengan kontras parameter yang tinggi, penggunaan kendala ini secara berlebihan dapat menyebabkan hilangnya detail penting dalam model. Oleh karena itu, dalam praktiknya, flatness constraint sering dikombinasikan dengan jenis regularisasi lain atau diatur bobotnya agar tetap mampu merepresentasikan variasi geologi secara optimal tanpa mengorbankan kestabilan solusi (Putra & Sutarno, 2019).
    </p>

<p><b>Langkah Pengerjaan Inversi Linear Under-determined:</b></p>

<p><b>1. Menyiapkan Data Observasi</b><br>
Menentukan data sudut datang (<i>theta</i>) dan koefisien refleksi (<i>R_obs</i>) yang akan digunakan sebagai input dalam proses inversi.</p>

<p><b>2. Konversi Sudut</b><br>
Mengubah sudut dari satuan derajat ke radian agar dapat digunakan dalam fungsi trigonometri pada perhitungan matematis.</p>

<p><b>3. Membentuk Matriks Kernel (G)</b><br>
Menyusun matriks kernel berdasarkan fungsi sudut datang, yaitu komponen konstanta, sin²(θ), dan kombinasi sin²(θ)·tan²(θ). Matriks ini merepresentasikan hubungan antara model dan data.</p>

<p><b>4. Menyusun Vektor Data</b><br>
Mengubah data observasi menjadi bentuk vektor kolom agar sesuai dengan operasi matriks dalam inversi linear.</p>

<p><b>5. Melakukan Inversi Minimum Norm</b><br>
Menghitung parameter model menggunakan metode minimum norm karena sistem bersifat under-determined (jumlah data lebih sedikit dari parameter). Metode ini menghasilkan solusi dengan norma terkecil.</p>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import copy
from numpy.linalg import inv

# Station Position
x = [0, 10, 20,30]
z = [0, 0, 0, 0]
n = len(x)

# True hypocenter
x0 = 12
z0 = 10
t0 = 0

# Model Velocity 1 Layer
v = 7

# Forward Modelling to calculate traveltime from hypocenter to stataion
t = []
for i in range(n):
  t.append((1/v) * np.sqrt((z[i] - z0)**2 + (x[i] - x0)**2))

# Convert t to a numpy array after it's fully populated
t = np.array(t)

# Add Noise to travel time data
noise = np.random.rand(n) * 0.03 * np.mean(t)
ti = t + noise

# Guess the initial position
M = [15,30]
init_position = copy.deepcopy(M)
M1 = []

# Create weighted matrix
W = np.eye(n)

# Assign Weighted matrix
W[n-1][n-1] = 0.1
# W[i][i] = abs(ti[i] - (ti[i] = np.mean(ti)))

print(W)

xhipo = [init_position[0]]
yhipo = [init_position[1]]
Erms = []

# Inversion without weighted factor
for niter in range(1000): #Ganti jumlah iterasi nnt
   gm = []
   dgm_dm1 = []
   dgm_dm2 = []

   for i in range(n):
    gm.append(t0 + (1/v) * (np.sqrt((x[i] - M[0])**2 +(z[i] - M[1])**2)))
    dgm_dm1.append((1/v) * (-1*(x[i] - M[0]))/(np.sqrt((x[i] - M[0])**2 +(z[i] - M[1])**2)))
    dgm_dm2.append((1/v) * (-1*(z[i] - M[1]))/(np.sqrt((x[i] - M[0])**2 +(z[i] - M[1])**2)))

   J = np.zeros((n, 2))

   for i in range(n):
     J[i][0] = dgm_dm1[i].item()
     J[i][1] = dgm_dm2[i].item()

   Mo = copy.deepcopy(M)

   # Add model pertubation from inv to Mo
   Mo = np.reshape(Mo, (2,1))
   ti = np.reshape(ti, (n,1))
   gm = np.reshape(gm, (n,1))

   Model = Mo + inv(J.T @ J) @ J.T @ (ti - gm)
   M = [Model[0][0], Model[1][0]]

   Data = Model.tolist()
   xhipo.append(Data[0][0])
   yhipo.append(Data[1][0])


   # Calculate rms error
   dt = ti - gm # (np.mat(ti) - np.mat(gm))
   Erms.append(np.sqrt(np.average(dt**2)))

plt.figure(figsize=(7,5))
plt.plot(15, 30, 'o', label='Initial Position', color='black')
plt.plot(xhipo, yhipo, 'o', label='Hypocenter Inversion')
plt.plot(xhipo, yhipo, '-', label='Hypocenter Inversion')
plt.plot(x0, z0, 'o', alpha=0.5, color='red', label='True Hypocenter')
plt.title('Hypocenter Determination_072')
plt.xlabel('X Coordinate')
plt.ylabel('Y Coordinate')
plt.legend()
plt.grid()

# Plot kurva error RMS terhadap iterasi
plt.figure(figsize=(7,5))
plt.plot(Erms, color='blue')
plt.title('Kurva RMS Error Vs N Iterasi_072')
plt.ylabel('RMS Error')
plt.xlabel('N Iterasi')
plt.grid()
plt.show()

print(Erms)
```

```python
# Zaky Dwi Putra_123120081
# Problem 3 : Inversi Linear Under-determined
import numpy as np
import matplotlib.pyplot as plt
from numpy.linalg import inv

# Data Observasi
theta_deg = np.array([10, 25])     # sudut datang (derajat)
R_obs = np.array([0.1523, 0.0947]) # koefisien refleksi observasi

# Ubah ke radian
theta = np.radians(theta_deg)

# Forward Modelling Matrix (G)
G = []

for i in range(len(theta)):
    
    g1 = 1
    g2 = np.sin(theta[i])**2
    g3 = (np.sin(theta[i])**2) * (np.tan(theta[i])**2)
    
    G.append([g1, g2, g3])

G = np.array(G)

# Data vector
d = np.reshape(R_obs,(len(R_obs),1))

print("Matrix Kernel (G):")
print(G)

print("Data Observasi:")
print(d)

# Inversi Linear Under-determined
# Minimum Norm Solution
GT = G.transpose()

Model = GT @ inv(G @ GT) @ d

A = Model[0][0]
B = Model[1][0]
C = Model[2][0]

print("\nParameter Model:")
print("A =",A)
print("B =",B)
print("C =",C)


# Forward modelling hasil inversi
theta_model = np.linspace(0,40,100)
theta_model_rad = np.radians(theta_model)

R_model = A + B*(np.sin(theta_model_rad)**2) + \
          C*(np.sin(theta_model_rad)**2)*(np.tan(theta_model_rad)**2)


# Plot Hasil
plt.figure(figsize=(7,5))

plt.plot(theta_model,R_model,label='Model Inversi')
plt.plot(theta_deg,R_obs,'o',color='red',label='Data Observasi')

plt.xlabel('Sudut Datang (derajat)')
plt.ylabel('Koefisien Refleksi')
plt.title('Inversi Linear Under-determined')
plt.legend()
plt.grid()

plt.show()
```
