    <p style="text-align: justify;">
        I. Teori Dasar
        </p>
        <p style="text-align: justify;">
        Inversi non-linear dengan pendekatan global merupakan metode optimasi yang digunakan untuk memperoleh parameter model terbaik tanpa bergantung pada model awal secara kuat. Metode ini bekerja dengan mengeksplorasi seluruh ruang model sehingga mampu mengurangi kemungkinan solusi terjebak pada minimum lokal. Berbeda dengan pendekatan lokal yang hanya mencari solusi di sekitar model awal, pendekatan global melakukan pencarian secara lebih luas melalui proses acak maupun sistematis. Beberapa metode yang termasuk dalam pendekatan global yaitu Grid Search, Monte Carlo, dan Simulated Annealing. Pendekatan ini banyak digunakan dalam geofisika untuk menyelesaikan permasalahan inversi yang kompleks, terutama ketika hubungan antara data observasi dan parameter model bersifat non-linear (Tarantola, 2005).
        </p>
         <p style="text-align: justify;">
        Metode global search memiliki keunggulan dalam menemukan solusi optimum pada fungsi objektif yang kompleks dan memiliki banyak minimum lokal. Dalam prosesnya, metode ini mengevaluasi berbagai kombinasi parameter model untuk memperoleh nilai misfit terkecil antara data hasil observasi dan data sintetik. Walaupun membutuhkan waktu komputasi yang lebih besar dibanding pendekatan lokal, metode inversi non-linear global mampu menghasilkan solusi yang lebih stabil dan representatif terhadap kondisi bawah permukaan sebenarnya (Menke, 2012).
        <p style="text-align: justify;">
        A. Invers Modelling
        </p>
        <p style="text-align: justify;">
        Pemodelan inversi (inverse modeling) sering dianggap sebagai kebalikan dari pemodelan maju (forward modeling) karena parameter model diperoleh langsung dari data observasi. Teori inversi merupakan kumpulan metode matematika dan statistika yang digunakan untuk mendapatkan informasi mengenai suatu sistem fisika berdasarkan hasil pengamatan. Dalam proses ini, data observasi digunakan untuk menentukan model atau parameter model yang paling sesuai dengan kondisi sebenarnya. Proses inversi dilakukan dengan memodifikasi model secara otomatis agar respons hasil perhitungan mendekati data pengamatan. Oleh sebab itu, inversi juga dikenal sebagai data fitting karena bertujuan mencari model yang menghasilkan kecocokan terbaik terhadap data observasi. Tingkat kecocokan tersebut biasanya dinyatakan dalam fungsi objektif yang harus diminimalkan.
        </p>
         <p style="text-align: justify;">
        Secara umum, inversi dibedakan menjadi inversi linear dan inversi non-linear. Pada inversi linear, hubungan antara data dan parameter model dapat dinyatakan melalui persamaan linear sehingga solusi lebih mudah diperoleh. Sementara itu, inversi non-linear memiliki fungsi objektif yang lebih kompleks dan dapat diselesaikan menggunakan pendekatan lokal maupun global. Pendekatan lokal hanya memanfaatkan informasi di sekitar model awal atau model hasil iterasi sebelumnya, sehingga pencarian solusi berisiko terjebak pada minimum lokal. Semakin kompleks fungsi non-linear dan semakin banyak parameter model, maka bentuk fungsi objektif juga semakin rumit.
         </p>
         <p style="text-align: justify;">
         Dalam inversi non-linear, fungsi objektif sering digambarkan sebagai permukaan pada ruang model berdimensi tertentu. Karena adanya banyak kemungkinan minimum lokal, solusi yang diperoleh melalui pendekatan lokal belum tentu merupakan solusi optimum. Oleh karena itu, digunakan pendekatan global (global search approach) yang mampu mengeksplorasi ruang model secara lebih luas. Beberapa metode yang termasuk pendekatan global yaitu Grid Search dan Monte Carlo, yang digunakan untuk memperoleh solusi inversi yang lebih optimal dan representatif.
         </p>
         <p style="text-align: justify;">
         B. Grid Search
         </p>
         <p style="text-align: justify;">
         Metode Grid Search merupakan pendekatan inversi yang digunakan dalam analisis data geofisika untuk memperkirakan model bawah permukaan berdasarkan data observasi. Metode ini bekerja dengan meminimalkan selisih antara data hasil pengamatan dan data hasil perhitungan model. Pada prosesnya, ruang pencarian dibagi menjadi beberapa grid atau sel, di mana setiap grid mewakili kemungkinan kombinasi parameter model. Seluruh kombinasi parameter kemudian dievaluasi menggunakan fungsi objektif untuk mendapatkan solusi yang paling sesuai.
         </p>
         <p style="text-align: justify;">
         Grid Search memiliki keunggulan karena mampu mengeksplorasi ruang model secara sistematis dan terstruktur. Namun, metode ini membutuhkan waktu komputasi yang cukup besar, terutama jika jumlah parameter atau dimensi model yang dicari semakin kompleks. Meskipun demikian, metode ini tetap efektif untuk memperoleh solusi inversi yang representatif terhadap kondisi bawah permukaan.
         </p>
         <p style="text-align: justify;">
         C. Metode Simulated Annealing
         </p>
         <p style="text-align: justify;">
         PMetode Simulated Annealing (SA) merupakan teknik optimasi global yang termasuk dalam guided random search. Pada metode ini, model yang mendekati solusi optimum memiliki peluang lebih besar untuk dipilih dibanding model lainnya. Berbeda dengan purely random search yang memberikan peluang sama pada semua sampel, SA menggunakan nilai probabilitas untuk menentukan solusi yang lebih baik selama proses pencarian. Prinsip Simulated Annealing diadaptasi dari proses termodinamika pada pembentukan kristal. Pada suhu tinggi, suatu material berada dalam kondisi cair, kemudian didinginkan secara perlahan hingga mencapai keadaan energi minimum yang stabil. Konsep tersebut diterapkan dalam proses inversi agar solusi yang diperoleh dapat mendekati nilai optimum global dan mengurangi kemungkinan terjebak pada minimum lokal.
         </p>

Permasalahan 1

```python
import numpy as np                      # Import NumPy untuk perhitungan numerik (array, matematika)
import matplotlib.pyplot as plt         # Import matplotlib untuk visualisasi

stations = np.array([                  # Koordinat stasiun (x, y)
    [20, 10],                          # Stasiun 1
    [50, 25],                          # Stasiun 2
    [40, 50],                          # Stasiun 3
    [10, 40]                           # Stasiun 4
])

t_obs = np.array([7.1, 1.8, 5.0, 7.9]) # Data waktu tempuh observasi (sudah dikurangi origin time)
v = 4.0                                # Kecepatan gelombang

def misfit(x, y):                      # Fungsi untuk menghitung misfit (error model)
    t_calc = np.sqrt(                  # Hitung waktu tempuh model
        (stations[:,0]-x)**2 +         # Selisih koordinat x, dikuadratkan
        (stations[:,1]-y)**2           # Selisih koordinat y, dikuadratkan
    ) / v                              # Dibagi kecepatan untuk mendapatkan waktu
    return np.sum(                     # Jumlahkan semua error
        (t_calc - t_obs)**2            # Selisih model vs observasi, lalu kuadratkan
    )

def grid_search(x_range, y_range, step):  # Fungsi grid search (pencarian menyeluruh)
    best_m = np.inf                    # Nilai misfit terbaik (awal = tak hingga)
    best_pos = None                    # Posisi terbaik
    
    for x in np.arange(x_range[0], x_range[1], step):  # Loop x dalam range
        for y in np.arange(y_range[0], y_range[1], step):  # Loop y dalam range
            m = misfit(x, y)           # Hitung misfit pada titik (x,y)
            
            if m < best_m:             # Jika lebih baik dari sebelumnya
                best_m = m             # Update nilai misfit terbaik
                best_pos = (x, y)      # Simpan posisi terbaik
                
    return best_pos, best_m            # Return posisi dan misfit terbaik

def simulated_annealing(init_pos, bounds, T=1.0, alpha=0.95, iter_max=1000):  # Fungsi SA
    current = np.array(init_pos)       # Posisi awal
    current_m = misfit(current[0], current[1])  # Misfit awal
    
    best = current.copy()              # Simpan solusi terbaik
    best_m = current_m                # Misfit terbaik
    
    history = []                      # List untuk menyimpan jejak solusi
    
    for i in range(iter_max):         # Loop iterasi
        
        new = current + np.random.uniform(-1, 1, 2)  # Generate kandidat baru acak
        
        new = np.clip(new, bounds[0], bounds[1])     # Batasi agar tetap dalam domain
        
        new_m = misfit(new[0], new[1])  # Hitung misfit kandidat baru
        
        if new_m < current_m or np.random.rand() < np.exp(-(new_m-current_m)/T):
            current = new             # Terima solusi baru
            current_m = new_m         # Update misfit
            
            if new_m < best_m:        # Jika terbaik sejauh ini
                best = new            # Update solusi terbaik
                best_m = new_m        # Update nilai terbaik
        
        history.append(current.copy())  # Simpan posisi ke histori
        
        T *= alpha                    # Turunkan suhu (cooling schedule)
    
    return best, best_m, np.array(history)  # Return hasil terbaik dan histori

bounds = ([0,0],[60,60])              # Batas domain pencarian (x dan y)

gs_pos, gs_m = grid_search((0,60),(0,60),1)  # Jalankan grid search dengan step 1

sa_pos, sa_m, history = simulated_annealing(gs_pos, bounds)  # Refinement dengan SA

print("Grid Search:", gs_pos, gs_m)  # Tampilkan hasil grid search
print("Simulated Annealing:", sa_pos, sa_m)  # Tampilkan hasil SA

x = np.arange(0,60,1)                # Grid nilai x untuk plotting
y = np.arange(0,60,1)                # Grid nilai y
X, Y = np.meshgrid(x, y)             # Membuat grid 2D
Z = np.zeros_like(X)                 # Inisialisasi matriks misfit

for i in range(len(x)):              # Loop untuk setiap x
    for j in range(len(y)):          # Loop untuk setiap y
        Z[j,i] = misfit(x[i], y[j])  # Hitung misfit di tiap titik grid

plt.figure()                         # Buat figure baru

plt.contourf(X, Y, Z)               # Plot kontur misfit (heatmap)

plt.scatter(stations[:,0], stations[:,1], marker='^')  # Plot posisi stasiun

plt.scatter(gs_pos[0], gs_pos[1], marker='x')  # Plot hasil grid search

plt.scatter(sa_pos[0], sa_pos[1], marker='o')  # Plot hasil simulated annealing

plt.plot(history[:,0], history[:,1])  # Plot jalur pencarian SA

plt.title("2D Hypocenter Estimation")  # Judul grafik
plt.xlabel("X")                        # Label sumbu X
plt.ylabel("Y")                        # Label sumbu Y

plt.show()                             # Tampilkan plot
```

Permasalahan 2

```python
import numpy as np                      # Import library NumPy untuk operasi numerik (array, matematika)
import matplotlib.pyplot as plt         # Import matplotlib untuk visualisasi/plot grafik

np.random.seed(0)                      # Set seed random agar hasil selalu sama (reproducible)

NM = 35                                # Nilai parameter (berdasarkan NIM)

true_hypo = np.array([13*NM, 15*NM, -5*NM])  # Koordinat hiposenter sebenarnya (x, y, z)

v = 5.1                                # Kecepatan gelombang (konstan)

stations = np.array([                  # Array posisi stasiun pengamat
    [700, 1900, 0],                   # Stasiun 1
    [2500, 2000, 0],                  # Stasiun 2
    [1500, 2500, 0],                  # Stasiun 3
    [1850, 800, 0]                    # Stasiun 4
])

def travel_time(h):                    # Fungsi untuk menghitung waktu tempuh dari hiposenter h
    return np.sqrt(                   # Hitung akar kuadrat (jarak Euclidean)
        np.sum((stations - h)**2, axis=1)  # Hitung selisih koordinat, kuadratkan, lalu jumlahkan per stasiun
    ) / v                             # Bagi jarak dengan kecepatan untuk mendapatkan waktu

t_true = travel_time(true_hypo)        # Hitung waktu tempuh sebenarnya dari hiposenter asli

noise = np.random.normal(0, 0.005 * t_true)  # Generate noise Gaussian (mean=0, std=0.5% dari waktu)

t_obs = t_true + noise                # Data observasi = waktu sebenarnya + noise

def misfit(h):                        # Fungsi untuk menghitung misfit (error model)
    return np.sum(                    # Jumlahkan semua error
        (travel_time(h) - t_obs)**2   # Selisih waktu model dan observasi, lalu kuadratkan
    )

def grid_search_coarse():             # Fungsi grid search kasar
    best = None                       # Variabel untuk menyimpan posisi terbaik
    best_m = np.inf                   # Misfit terbaik (awal = tak hingga)
    
    for x in range(0,3000,200):       # Loop x dari 0 sampai 3000 dengan step 200
        for y in range(0,3000,200):   # Loop y
            for z in range(-500,0,100): # Loop z (kedalaman negatif)
                h = np.array([x,y,z]) # Kandidat hiposenter
                m = misfit(h)         # Hitung misfit
                
                if m < best_m:        # Jika lebih baik (lebih kecil)
                    best = h          # Simpan posisi
                    best_m = m        # Simpan nilai misfit
    
    return best                       # Kembalikan hasil terbaik

def grid_search_refine(center):       # Grid search halus di sekitar titik center
    best = center                     # Mulai dari hasil coarse
    best_m = misfit(center)           # Hitung misfit awal
    
    for x in np.arange(center[0]-200, center[0]+200, 50):  # Loop x lebih detail
        for y in np.arange(center[1]-200, center[1]+200, 50):  # Loop y
            for z in np.arange(center[2]-100, center[2]+100, 25): # Loop z
                h = np.array([x,y,z]) # Kandidat baru
                m = misfit(h)         # Hitung misfit
                
                if m < best_m:        # Jika lebih baik
                    best = h          # Update posisi
                    best_m = m        # Update misfit
    
    return best, best_m               # Return hasil terbaik dan nilainya

def simulated_annealing(init):        # Fungsi optimasi Simulated Annealing
    current = init.copy()             # Posisi awal (dari grid search)
    current_m = misfit(current)       # Misfit awal
    
    best = current.copy()             # Simpan solusi terbaik
    best_m = current_m                # Simpan misfit terbaik
    
    history = []                      # List untuk menyimpan jejak pencarian
    
    T = 1.0                           # Suhu awal
    
    for i in range(2000):             # Iterasi sebanyak 2000
        
        new = current + np.random.uniform(-50,50,3)  # Generate kandidat baru secara acak
        
        new[0] = np.clip(new[0], 0, 3000)  # Batasi x agar tetap dalam domain
        new[1] = np.clip(new[1], 0, 3000)  # Batasi y
        new[2] = np.clip(new[2], -500, 0)  # Batasi z
        
        new_m = misfit(new)           # Hitung misfit kandidat baru
        
        if new_m < current_m or np.random.rand() < np.exp(-(new_m-current_m)/T):
            current = new             # Terima solusi baru
            current_m = new_m         # Update misfit
            
            if new_m < best_m:        # Jika terbaik sejauh ini
                best = new            # Update best
                best_m = new_m        # Update best misfit
        
        history.append(current.copy())  # Simpan posisi ke histori
        
        T *= 0.95                    # Turunkan suhu (cooling schedule)
    
    return best, best_m, np.array(history)  # Return hasil terbaik dan histori

coarse = grid_search_coarse()         # Jalankan grid search kasar

gs_pos, gs_m = grid_search_refine(coarse)  # Refinement hasil coarse

sa_pos, sa_m, history = simulated_annealing(gs_pos)  # Optimasi dengan SA

error = np.linalg.norm(sa_pos - true_hypo)  # Hitung error Euclidean terhadap nilai asli

print("True:", true_hypo)            # Print hiposenter sebenarnya
print("Grid Search:", gs_pos)        # Print hasil grid search
print("Simulated Annealing:", sa_pos) # Print hasil SA
print("Error:", error)               # Print error

from mpl_toolkits.mplot3d import Axes3D  # Import modul 3D plotting

fig = plt.figure()                   # Buat figure baru
ax = fig.add_subplot(111, projection='3d')  # Tambahkan subplot 3D

ax.scatter(stations[:,0], stations[:,1], stations[:,2], marker='^')  # Plot stasiun

ax.scatter(true_hypo[0], true_hypo[1], true_hypo[2], marker='o')  # Plot posisi sebenarnya

ax.scatter(sa_pos[0], sa_pos[1], sa_pos[2], marker='x')  # Plot hasil estimasi

ax.plot(history[:,0], history[:,1], history[:,2])  # Plot jalur pencarian SA

ax.set_title("3D Hypocenter Estimation")  # Judul plot
ax.set_xlabel("X")                        # Label sumbu X
ax.set_ylabel("Y")                        # Label sumbu Y
ax.set_zlabel("Z")                        # Label sumbu Z

plt.show()                               # Tampilkan plot
```

Pada praktikum ini digunakan dua metode pencarian global (global search), yaitu Grid Search (GS) dan Simulated Annealing (SA), pada medium 2D dan 3D.
        <p style="text-align: justify;">
    </p>
    Pada kasus 2D, parameter yang dicari hanya koordinat (x,y) sehingga ruang model masih relatif kecil. Metode Grid Search menghasilkan solusi yang sangat baik karena menghitung seluruh kemungkinan model pada grid secara menyeluruh. Metode ini bersifat deterministik sehingga hasilnya cenderung stabil, tetapi membutuhkan waktu komputasi lebih lama. Di sisi lain, Simulated Annealing juga mampu memberikan hasil yang mendekati nilai sebenarnya dengan proses pencarian yang lebih cepat melalui pendekatan acak terarah.
<p style="text-align: justify;">
    </p>
    Pada kasus 3D, parameter model bertambah menjadi (x,y,z) sehingga ruang model menjadi lebih luas dan kompleks. Selain itu, data yang digunakan mengandung noise sehingga kondisi permasalahan lebih mendekati keadaan nyata. Pada kondisi ini, Grid Search menjadi kurang efisien karena banyaknya kombinasi parameter yang harus dihitung menyebabkan waktu komputasi meningkat. Sebaliknya, Simulated Annealing lebih efektif karena dapat mengeksplorasi ruang model secara efisien tanpa harus mencoba semua kemungkinan solusi serta mampu menghindari minimum lokal.
 <p style="text-align: justify;">
    </p>
    Berdasarkan hasil praktikum, dapat diketahui bahwa Grid Search lebih cocok digunakan pada permasalahan sederhana dengan dimensi rendah karena hasilnya akurat dan stabil. Namun, untuk kasus yang lebih kompleks seperti medium 3D, Simulated Annealing lebih unggul karena lebih cepat dan efisien dalam memperoleh solusi optimum. Oleh karena itu, metode Simulated Annealing lebih direkomendasikan dalam penyelesaian inversi geofisika non-linear yang kompleks.

    4. Simulated Annealing lebih efektif karena menggunakan pendekatan guided random search dan mampu menghindari local minimum.
    <p style="text-align: justify;">
    5. Pemahaman karakteristik setiap metode penting untuk menentukan metode inversi yang paling tepat dan efisien dalam permasalahan geofisika.
    </p>
