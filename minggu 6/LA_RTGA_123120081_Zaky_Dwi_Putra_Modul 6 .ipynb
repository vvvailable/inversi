    <p style="text-align: justify;">
      Inversi merupakan salah satu metode penting dalam geofisika untuk memperoleh parameter bawah permukaan berdasarkan data observasi di permukaan. Permasalahan inversi sering kali bersifat non-linear, sehingga solusi yang diperoleh tidak tunggal dan sangat bergantung pada model awal yang digunakan. Hal ini menyebabkan metode inversi linear konvensional sering kali tidak mampu memberikan hasil yang optimal, terutama pada sistem yang kompleks dan memiliki banyak parameter. Oleh karena itu, diperlukan pendekatan global yang mampu mengeksplorasi ruang solusi secara lebih luas tanpa terlalu bergantung pada tebakan awal. Salah satu metode yang banyak digunakan adalah Simulated Annealing (SA), yang terinspirasi dari proses pendinginan logam dalam ilmu material. Metode ini memungkinkan pencarian solusi global dengan menghindari jebakan minimum lokal melalui mekanisme probabilistik dalam proses optimasi. Pendekatan ini dinilai lebih fleksibel dan robust dalam menyelesaikan permasalahan inversi non-linear dibandingkan metode deterministik lainnya (Sutopo & Widodo, 2018).
    </p>
      <p style="text-align: justify;">
     Simulated Annealing bekerja dengan prinsip menurunkan “temperatur” secara bertahap untuk mengontrol eksplorasi dan eksploitasi dalam pencarian solusi terbaik. Pada temperatur tinggi, sistem memiliki peluang besar untuk menerima solusi yang lebih buruk, sehingga mampu keluar dari minimum lokal. Seiring penurunan temperatur, pencarian menjadi lebih terfokus pada solusi optimal. Dalam konteks geofisika, metode ini telah banyak diterapkan pada inversi data seismik, gravitasi, dan resistivitas untuk meningkatkan akurasi model bawah permukaan. Keunggulan utama metode ini adalah kemampuannya dalam menangani kompleksitas model serta ketidaklinieran hubungan antara parameter dan data. Namun demikian, pemilihan parameter seperti skema pendinginan dan jumlah iterasi sangat mempengaruhi hasil inversi yang diperoleh. Oleh karena itu, pemahaman yang baik terhadap karakteristik metode Simulated Annealing sangat diperlukan agar dapat menghasilkan model yang representatif dan akurat (Pratama, Santoso, & Nugroho, 2020).
    </p>

        1.2 Dasar Teori

    <p style="text-align: justify;">
<p style="text-align: justify;">
        1.2.1 Fondasi Inversi Geofisika: Dari Data ke Model Bawah Permukaan
    </p>
    <p style="text-align: justify;">
    Inversi geofisika merupakan suatu proses matematis yang digunakan untuk mengestimasi parameter bawah permukaan berdasarkan data pengamatan di permukaan. Dalam praktiknya, inversi menjadi tahap lanjutan setelah pemodelan maju (forward modeling) yang bertujuan untuk mencocokkan respon model terhadap data observasi. Hubungan antara data dan model dalam inversi umumnya dinyatakan dalam bentuk persamaan non-linear yang kompleks. Oleh karena itu, proses inversi memerlukan pendekatan numerik untuk memperoleh solusi yang representatif. Konsep utama dalam inversi adalah bagaimana mengubah data geofisika menjadi informasi fisik seperti densitas, kecepatan gelombang, atau resistivitas. Proses ini tidak sederhana karena satu set data dapat dihasilkan oleh banyak kemungkinan model bawah permukaan. Hal ini menyebabkan masalah non-keunikan yang menjadi tantangan utama dalam inversi geofisika. Selain itu, kualitas data yang digunakan juga sangat mempengaruhi hasil inversi yang diperoleh. Dengan demikian, inversi geofisika menjadi alat penting dalam interpretasi kondisi bawah permukaan secara kuantitatif (Santoso & Prasetyo, 2019).
    <p style="text-align: justify;">
    </p>
 </p>
 <p style="text-align: justify;">
    Hubungan antara data dan model dalam inversi geofisika biasanya dinyatakan dalam bentuk fungsi matematis yang menghubungkan parameter model dengan respon fisik yang terukur. Dalam konteks ini, model bawah permukaan direpresentasikan sebagai sekumpulan parameter yang ingin ditentukan nilainya melalui proses inversi. Data observasi kemudian dibandingkan dengan data hasil perhitungan model untuk menilai tingkat kecocokan keduanya. Proses ini dilakukan secara iteratif hingga diperoleh model yang memberikan kesesuaian terbaik. Salah satu tujuan utama inversi adalah meminimalkan selisih antara data observasi dan data sintetik yang dihasilkan model. Pendekatan ini sering disebut sebagai minimisasi fungsi objektif atau misfit. Semakin kecil nilai misfit, maka model dianggap semakin mendekati kondisi sebenarnya di bawah permukaan. Namun, solusi yang diperoleh tidak selalu unik sehingga diperlukan kendala tambahan atau regularisasi. Oleh karena itu, pemahaman yang baik tentang hubungan data dan model sangat penting dalam menghasilkan interpretasi geofisika yang akurat dan dapat dipertanggungjawabkan (Widodo, Nugraha, & Putra, 2021).
    </p>
<p style="text-align: justify;">
        1.2.2 Kompleksitas Inversi Non-Linear: Tantangan dan Karakteristiknya
    <p style="text-align: justify;">
   Inversi non-linear dalam geofisika merupakan permasalahan yang kompleks karena hubungan antara parameter model dan data observasi tidak bersifat linier. Hal ini menyebabkan perubahan kecil pada parameter model dapat menghasilkan perubahan yang tidak proporsional pada respon data. Kondisi tersebut membuat proses pencarian solusi menjadi lebih sulit dibandingkan dengan inversi linear. Salah satu karakteristik utama inversi non-linear adalah adanya banyak kemungkinan solusi yang dapat memenuhi data observasi. Fenomena ini dikenal sebagai non-uniqueness, di mana satu set data dapat dijelaskan oleh berbagai model bawah permukaan yang berbeda. Akibatnya, interpretasi hasil inversi memerlukan pertimbangan tambahan seperti informasi geologi atau kendala fisik. Selain itu, inversi non-linear sangat sensitif terhadap model awal yang digunakan dalam proses iterasi. Pemilihan model awal yang kurang tepat dapat menyebabkan solusi terjebak pada hasil yang tidak optimal. Oleh karena itu, pemahaman terhadap kompleksitas ini menjadi kunci dalam penerapan metode inversi yang efektif (Sutanto & Raharjo, 2020).
</p>
<p style="text-align: justify;">
   Selain permasalahan non-keunikan dan sensitivitas terhadap model awal, inversi non-linear juga menghadapi tantangan berupa keberadaan minimum lokal dalam ruang solusi. Minimum lokal adalah kondisi di mana solusi yang diperoleh tampak optimal secara lokal, namun bukan merupakan solusi terbaik secara global. Hal ini sering terjadi pada fungsi objektif yang memiliki banyak lembah dan puncak dalam ruang parameter. Metode optimasi konvensional berbasis gradien cenderung terjebak pada minimum lokal sehingga gagal menemukan solusi global. Oleh karena itu, diperlukan pendekatan yang mampu mengeksplorasi ruang solusi secara lebih luas. Salah satu cara untuk mengatasi masalah ini adalah dengan menggunakan metode optimasi global. Metode ini memungkinkan pencarian solusi yang tidak hanya bergantung pada arah gradien lokal. Dengan demikian, peluang untuk menemukan solusi global menjadi lebih besar. Kompleksitas ini menunjukkan bahwa inversi non-linear memerlukan strategi khusus dalam implementasinya. Pemilihan metode yang tepat sangat berpengaruh terhadap kualitas hasil inversi yang diperoleh (Prasetyo, Nugroho, & Santoso, 2021).
</p>
<p style="text-align: justify;">
        1.2.3 Mengukur Ketepatan Model: Peran Fungsi Objektif (Misfit)
    </p>
    <p style="text-align: justify;">
    Evaluasi kualitas model dalam inversi geofisika dilakukan menggunakan fungsi objektif atau misfit yang mengukur perbedaan antara data observasi dan data sintetik hasil pemodelan. Secara umum, hubungan antara data dan model dapat dinyatakan sebagai d=g(m), di mana d adalah data observasi, m adalah parameter model, dan g(m) adalah fungsi forward. Dalam praktiknya, karena adanya kesalahan pengukuran dan keterbatasan model, data observasi dobs tidak akan sama persis dengan data hasil model dcalc. Oleh karena itu, didefinisikan fungsi misfit untuk mengukur selisih tersebut. Bentuk paling umum dari fungsi objektif adalah metode kuadrat terkecil (least squares), yang dinyatakan sebagai berikut: 
    </p>
    <p style="text-align: center;">
    ϕ(m)=∑i=1N​(dobs,i​−dcalc,i​(m))2
    </p>
    </p>
    <p style="text-align: justify;">
    Fungsi tersebut menyatakan bahwa total error diperoleh dari penjumlahan kuadrat selisih antara data observasi dan data hasil model untuk setiap titik data. Pendekatan ini banyak digunakan karena memberikan solusi yang stabil dan memiliki dasar matematis yang kuat. Selain itu, untuk mempertimbangkan kualitas data yang berbeda-beda, sering digunakan bentuk berbobot (weighted least squares), sehingga setiap data memiliki kontribusi yang berbeda dalam perhitungan misfit. Dengan demikian, fungsi objektif menjadi dasar dalam menentukan seberapa baik suatu model merepresentasikan kondisi bawah permukaan (Santoso & Widodo, 2019).Dalam implementasinya, proses inversi bertujuan untuk mencari model m yang meminimalkan nilai fungsi objektif tersebut. Namun, dalam banyak kasus, minimisasi misfit saja tidak cukup karena dapat menghasilkan model yang terlalu kompleks atau tidak stabil. Oleh karena itu, ditambahkan istilah regularisasi untuk mengontrol solusi agar lebih realistis. Fungsi objektif yang telah ditambahkan regularisasi umumnya dinyatakan sebagai: 
    </p>
    <p style="text-align: center;">
     Φ(m)=∥dobs​−g(m)∥2+λ∥Lm∥2. 
     </p>
    <p style="text-align: justify;">
     Pada persamaan tersebut, λ adalah parameter regularisasi yang mengatur keseimbangan antara kesesuaian data dan kestabilan model, sedangkan L adalah  operator regularisasi. Nilai λ yang terlalu besar akan menghasilkan model yang terlalu halus, sedangkan nilai yang terlalu kecil dapat menyebabkan model mengikuti noise data. Oleh karena itu, pemilihan parameter ini sangat krusial dalam proses inversi. Selain itu, bentuk fungsi objektif juga dapat dimodifikasi menggunakan norma lain seperti norma L1 untuk meningkatkan ketahanan terhadap outlier. Pemahaman yang baik terhadap formulasi matematis fungsi objektif sangat penting dalam menghasilkan model inversi yang akurat dan dapat dipertanggungjawabkan secara ilmiah. Dengan demikian, fungsi misfit tidak hanya menjadi alat evaluasi, tetapi juga penentu arah optimasi dalam proses inversi geofisika (Prasetyo, Nugroho, & Santoso, 2021).
    </p>
<p style="text-align: justify;">
        1.2.4 Pendekatan Linear (Linearisasi)
    </p>
    <p style="text-align: justify;">
      Mekanisme algoritma Simulated Annealing merupakan proses optimasi global yang meniru fenomena pendinginan logam dalam ilmu fisika. Proses ini dimulai dari tahap inisialisasi model awal, yaitu menentukan parameter awal m0m_0m0​ secara acak atau berdasarkan informasi awal yang tersedia. Model awal ini kemudian digunakan untuk menghitung respon data sintetik melalui fungsi forward modeling. Selanjutnya dilakukan evaluasi terhadap fungsi objektif atau misfit untuk mengetahui tingkat kesesuaian antara data observasi dan model. Setelah itu, algoritma akan membangkitkan model baru m′m'm′ dengan cara memberikan perturbasi acak terhadap model sebelumnya. Perubahan ini bertujuan untuk mengeksplorasi ruang solusi secara luas tanpa terjebak pada satu titik tertentu. Setiap model baru yang dihasilkan akan kembali dievaluasi nilai misfit-nya. Proses ini dilakukan secara berulang untuk menghasilkan kandidat solusi yang lebih baik. Dengan demikian, tahap awal algoritma berfungsi sebagai eksplorasi ruang parameter secara global (Santoso & Prasetyo, 2019). Keunikan utama Simulated Annealing terletak pada mekanisme probabilitas penerimaan solusi yang dikenal sebagai kriteria Metropolis. Jika model baru menghasilkan nilai misfit yang lebih kecil, maka solusi tersebut akan langsung diterima. Namun, jika nilai misfit lebih besar, solusi tersebut masih memiliki peluang untuk diterima berdasarkan probabilitas tertentu. Probabilitas ini bergantung pada perubahan energi (misfit) dan temperatur sistem, yang secara matematis dinyatakan sebagai berikut:
    </p>
    <p style="text-align: center;">
     P=exp(−ΔE/T​)
     </p>
     <p style="text-align: justify;">
      di mana ΔE adalah selisih misfit antara model baru dan model sebelumnya, serta T adalah temperatur kontrol. Pada temperatur tinggi, peluang menerima solusi yang lebih buruk masih besar sehingga algoritma mampu keluar dari jebakan minimum lokal. Seiring penurunan temperatur, probabilitas ini semakin kecil sehingga pencarian menjadi lebih fokus pada solusi optimal. Proses iterasi terus dilakukan hingga temperatur mencapai nilai minimum atau kriteria konvergensi terpenuhi. Dengan mekanisme ini, Simulated Annealing mampu menyeimbangkan eksplorasi dan eksploitasi dalam pencarian solusi global. Oleh karena itu, metode ini sangat efektif digunakan dalam permasalahan inversi non-linear yang kompleks (Widodo, Nugraha, & Putra, 2021).
    </p>

```python
# Grid Search
# Zaky Dwi Putra_123120081

import numpy as np # Library untuk perhitungan numerik (array, operasi matematika)
import matplotlib.pyplot as plt # Library untuk plotting / visualisasi grafik
from numpy.linalg import inv # Fungsi untuk invers matriks (tidak dipakai di kode ini)

Vavg = 4  # Kecepatan rata-rata gelombang (km/s atau satuan lain)
tobs = [7.1,1.8,5.0,7.9]  # Waktu tiba gelombang hasil observasi di tiap stasiun
tobs = np.reshape(tobs, (len(tobs), 1)) # Mengubah bentuk array jadi kolom (n x 1)
Xsts = [20,50,40,10]  # Koordinat X stasiun seismik
Ysts = [10,25,50,40] # Koordinat Y stasiun seismik

# Definisikan ruang model
L = 1  # Lebar Grid
X0 = np.arange(0,50,L) # Rentang koordinat X model
Y0 = np.arange(0,50,L) # Rentang koordinat Y model

# Lakukan looping pencarian model solusi yang menghasilkan misfit minimum
P = 10  # nilai awal taksiran misfit minimum (dalam sekon)
HypoX = [] # Menyimpan koordinat X kandidat hiposenter terbaik
HypoY = [] # Menyimpan koordinat Y kandidat hiposenter terbaik
erms  = [] # Menyimpan nilai error RMS tiap update solusi

for i in range(len(Y0)): # Loop semua grid Y
    for j in range(len(X0)):  # Loop semua grid X
        Dt = []  # Menyimpan selisih waktu kuadrat (residual)
        for k in range(len(Xsts)):  # Loop tiap stasiun
            tcal = (np.sqrt((X0[j] - Xsts[k])**2 + (Y0[i] - Ysts[k])**2)/Vavg) # Hitung waktu tempuh teoritis dari titik grid ke stasiun
            Dt.append((tobs[k] - tcal)**2) 

        misfit = np.sqrt(np.mean(Dt))

        if misfit < P:
            HypoX.append(X0[j])  # Simpan koordinat X terbaik sementara
            HypoY.append(Y0[i])  # Simpan koordinat Y terbaik sementara
            erms.append(misfit)   # Simpan nilai error
            P = misfit  # Update nilai misfit minimum
        else:
            pass   # Jika tidak lebih baik, abaikan

# Plot hasil
plt.figure(figsize=(10, 8))

# Plot grid
for x in X0:
    plt.plot([x, x], [Y0[0], Y0[-1]], 'k-', alpha=0.2)  # Garis vertikal
for y in Y0:
    plt.plot([X0[0], X0[-1]], [y, y], 'k-', alpha=0.2)  # Garis horizontal

# Plot lokasi hiposenter terbaik
plt.plot(HypoX, HypoY, '-r', label='Jalur Pencarian') # Plot jalur pencarian solusi (perubahan solusi terbaik)
plt.plot(HypoX[0], HypoY[0], '*c', markersize=10, label='Awal')  # Titik awal pencarian solusi
plt.plot(HypoX[-1], HypoY[-1], '*y', markersize=10, label='Akhir') # Titik akhir (hiposenter terbaik

# Plot stasiun observasi
plt.plot(Xsts, Ysts, 'v', markersize=8, color='blue', label='Stasiun')

# Label dan Legend
plt.xlabel('Koordinat X') # Memberi nama sumbu Horizontal sebagai koordinat X
plt.ylabel('Koordinat Y') # Memberi nama sumbu vertikal sebagai koordinat Y
plt.title('Pencarian Lokasi Hiposenter dengan Grid Search') # Menampilkan judul grafik sesuai isi visualisasi
plt.legend() # Menampilkan keterangan simbol (jalur, awal, akhir, stasiun)
plt.grid(True) # Menampilkan garis bantu agar pembacaan koordinat lebih mudah
plt.show() # Menampilkan hasil plot

# Output hasil
print('Koordinat X Hiposenter Terbaik:', HypoX[-1]) # X terbaik
print('Koordinat Y Hiposenter Terbaik:', HypoY[-1]) # Y terbaik
print('Misfit Terbaik:', P) # Nilai error minimum
print('Jumlah Iterasi:', len(HypoX)) # Jumlah update solusi
# Hypo = np.array([HypoX, HypoY])
# print(Hypo)
plt.plot(HypoX, HypoY, '-r') # Menggambar garis merah yang menunjukkan jalur perubahan solusi hiposenter terbaik
plt.plot(HypoX[0], HypoY[0], '*c', label='awal') # Menandai titik pertama (awal iterasi) dengan simbol bintang warna
plt.plot(HypoX[len(HypoX)-1], HypoY[len(HypoY)-1], '*y', label='akhir') # Menandai titik terakhir (hiposenter terbaik) dengan simbol bintang warna kuning
plt.plot(Xsts, Ysts, 'v', label='stasiun') # Menampilkan posisi stasiun dengan simbol segitiga
plt.xlabel('') 
plt.legend() # Menampilkan grafik jalur pencarian
plt.show() # Menampilkan perubahan error RMS setiap kali solusi diperbarui
plt.plot(erms) # Menampilkan perubahan error RMS setiap kali solusi diperbarui
plt.title('kurva RMS VS N Iterasi') # Menjelaskan bahwa grafik menunjukkan hubungan RMS error vs jumlah iterasi
```

```python
#Simulated Anealing
#Zaky Dwi Putra_123120081

import numpy as np
import matplotlib.pyplot as plt

# Data observasi
Vavg = 4  # Kecepatan rata-rata gelombang dalam medium
tobs = [7.1, 1.8, 5.0, 7.9]  # Waktu tiba gelombang P yang diamati di setiap stasiun
Xsts = [20, 50, 40, 10]  # Koordinat X stasiun
Ysts = [10, 25, 50, 40]  # Koordinat Y stasiun

# Parameter Simulated Annealing
Xmin, Ymin = 0, 0      # Batas bawah ruang pencarian
Xmax, Ymax = 50, 50    # Batas atas ruang pencarian
T0 = 100               # Temperatur awal
Tmin = 0.1             # Temperatur minimum
alpha = 0.99           # Faktor pendinginan (cooling rate)
max_iter = 1000        # Maksimum iterasi

# Inisialisasi variabel
X = np.random.uniform(Xmin, Xmax)  # Koordinat X awal (acak)
Y = np.random.uniform(Ymin, Ymax)  # Koordinat Y awal (acak)
HypoX, HypoY = [X], [Y]  # List untuk menyimpan koordinat hiposenter

# Algoritma Simulated Annealing
T = T0  # Inisialisasi suhu awal (temperature awal)
for k in range(max_iter):  # Loop iterasi hingga batas maksimum iterasi
    
    # Perturbasi solusi saat ini
    dX = np.random.uniform(-1, 1) * (Xmax - Xmin) / 10  # Perubahan acak kecil pada koordinat X
    dY = np.random.uniform(-1, 1) * (Ymax - Ymin) / 10  # Perubahan acak kecil pada koordinat Y
    X1 = X + dX  # Kandidat solusi baru untuk X
    Y1 = Y + dY  # Kandidat solusi baru untuk Y

    # Hitung misfit untuk solusi saat ini dan solusi baru
    E = np.mean([(tobs[i] - np.sqrt((X - Xsts[i])**2 + (Y - Ysts[i])**2) / Vavg)**2 for i in range(len(Xsts))])  # Misfit solusi lama (error rata-rata)
    E1 = np.mean([(tobs[i] - np.sqrt((X1 - Xsts[i])**2 + (Y1 - Ysts[i])**2) / Vavg)**2 for i in range(len(Xsts))])  # Misfit solusi baru
    dE = E1 - E  # Selisih misfit (perubahan energi)

    # Terima atau tolak solusi baru
    if dE < 0 or np.exp(-dE / T) > np.random.rand():  # Jika solusi lebih baik atau diterima secara probabilistik
        X, Y = X1, Y1  # Update solusi ke solusi baru
        HypoX.append(X)  # Simpan riwayat koordinat X
        HypoY.append(Y)  # Simpan riwayat koordinat Y

    # Pendinginan (cooling schedule)
    T *= alpha  # Menurunkan suhu secara bertahap
    if T < Tmin:  # Jika suhu sudah terlalu kecil
        break  # Hentikan iterasi

# Plot hasil Simulated Annealing
plt.plot(HypoX, HypoY, '-r', label='Jalur Simulated Annealing')  # Plot jalur perpindahan solusi
plt.plot(HypoX[0], HypoY[0], '*c', label='Awal')  # Titik awal
plt.plot(HypoX[-1], HypoY[-1], '*y', label='Akhir')  # Titik akhir (hasil terbaik)
plt.plot(Xsts, Ysts, 'v', label='Stasiun')  # Lokasi stasiun pengamatan
plt.xlabel('Koordinat X')  # Label sumbu X
plt.ylabel('Koordinat Y')  # Label sumbu Y
plt.legend()  # Menampilkan legenda
plt.title ('Simulated Annealing Untuk penentuan hiposenter ')  # Judul grafik
plt.show()  # Menampilkan plot

print('Final Hypocenter:', HypoX[-1], HypoY[-1])  # Menampilkan koordinat hiposenter akhir
print('Final Misfit:', E1)  # Menampilkan nilai misfit akhir
print('Iterations:', len(HypoX))  # Menampilkan jumlah iterasi yang diterima

```

```python
#Zaky Dwi Putra_123120081
#Problem 2

import numpy as np
import matplotlib.pyplot as plt
import time

# DATA NIM 81

vavg = 5.1
tobs = np.array([171.36094527, 332.47252819, 278.04102498, 193.38347066])

Xsts = np.array([700, 2500, 1500, 1850])
Ysts = np.array([1900, 2000, 2500, 800])
Zsts = np.array([0, 0, 0, 0])

# FUNGSI MISFIT

def misfit(x, y, z):
    tcal = []
    for i in range(len(Xsts)):
        r = np.sqrt((x - Xsts[i])**2 + (y - Ysts[i])**2 + (z - Zsts[i])**2)
        tcal.append(r / vavg)
    return np.sqrt(np.mean((tobs - np.array(tcal))**2))

# GRID SEARCH
start_gs = time.time()

Xgrid = np.arange(0, 3000, 200)
Ygrid = np.arange(0, 3000, 200)
Zgrid = np.arange(-1000, 0, 100)

best_misfit = 1e10
best_model = None

for x in Xgrid:
    for y in Ygrid:
        for z in Zgrid:
            m = misfit(x, y, z)
            if m < best_misfit:
                best_misfit = m
                best_model = (x, y, z)

end_gs = time.time()
time_gs = end_gs - start_gs


# SIMULATED ANNEALING
start_sa = time.time()

T, Tmin, alpha = 100, 0.1, 0.95

x = np.random.uniform(0, 3000)
y = np.random.uniform(0, 3000)
z = np.random.uniform(-1000, 0)

current_misfit = misfit(x, y, z)

while T > Tmin:
    x_new = x + np.random.uniform(-200, 200)
    y_new = y + np.random.uniform(-200, 200)
    z_new = z + np.random.uniform(-100, 100)

    new_misfit = misfit(x_new, y_new, z_new)
    dE = new_misfit - current_misfit

    if dE < 0 or np.exp(-dE / T) > np.random.rand():
        x, y, z = x_new, y_new, z_new
        current_misfit = new_misfit

    T *= alpha

end_sa = time.time()
time_sa = end_sa - start_sa

# HASIL
print("=== GRID SEARCH ===")
print("Hiposenter:", best_model)
print("Misfit:", best_misfit)
print("Waktu komputasi:", time_gs, "detik")

print("\n=== SIMULATED ANNEALING ===")
print("Hiposenter:", (x, y, z))
print("Misfit:", current_misfit)
print("Waktu komputasi:", time_sa, "detik")

# PLOT 3D
from mpl_toolkits.mplot3d import Axes3D

x_gs, y_gs, z_gs = best_model
x_sa, y_sa, z_sa = x, y, z

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(Xsts, Ysts, Zsts, marker='^', s=80, label='Stasiun')
ax.scatter(x_gs, y_gs, z_gs, marker='o', s=100, label='Grid Search')
ax.scatter(x_sa, y_sa, z_sa, marker='x', s=100, label='Simulated Annealing')

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z (Depth)')
ax.set_title('Perbandingan Hiposenter 3D')
ax.invert_zaxis()
ax.legend()

plt.show()
```
