1.1 Latar Belakang</b>
<p>Dalam geofisika, data yang diperoleh dari pengukuran di permukaan bumi merupakan respons dari kondisi bawah permukaan yang tidak dapat diamati secara langsung. Oleh karena itu, diperlukan suatu metode untuk menginterpretasikan data tersebut menjadi informasi mengenai parameter fisis dan struktur geologi bawah permukaan. Proses ini dikenal sebagai pemodelan inversi, yaitu teknik untuk memperkirakan parameter model berdasarkan data observasi .

Secara umum, hubungan antara data dan parameter model dapat dinyatakan melalui proses forward modeling, yaitu perhitungan respons data berdasarkan model tertentu. Sebaliknya, dalam inversi, parameter model dicari sehingga respons model mendekati data observasi. Dalam praktiknya, kesesuaian antara data dan model dinyatakan melalui fungsi objektif yang diminimalkan, biasanya menggunakan metode least squares (kuadrat terkecil) .

Salah satu bentuk sederhana dari inversi adalah inversi linear, di mana hubungan antara data dan parameter model bersifat linear. Contoh klasiknya adalah regresi garis lurus, seperti hubungan temperatur terhadap kedalaman. Namun, dalam data geofisika nyata sering dijumpai adanya noise (kesalahan) dan outlier, yang dapat menyebabkan solusi inversi menjadi tidak stabil atau tidak representatif terhadap kondisi sebenarnya .

Selain itu, permasalahan inversi geofisika umumnya bersifat tidak unik (non-uniqueness), di mana lebih dari satu model dapat menghasilkan respons yang sama atau hampir sama terhadap data. Ketidakpastian ini dapat diperparah oleh kualitas data yang buruk atau distribusi data yang tidak merata. Oleh karena itu, diperlukan pendekatan untuk meningkatkan kestabilan solusi dan mengurangi pengaruh data yang tidak representatif .

Salah satu metode yang digunakan untuk mengatasi masalah tersebut adalah inversi linear berbobot (weighted least squares). Pada metode ini, setiap data diberi bobot sesuai tingkat kepercayaannya. Data dengan kualitas baik diberi bobot lebih besar, sedangkan data yang mengandung noise atau outlier diberi bobot kecil. Dengan demikian, solusi inversi akan lebih dipengaruhi oleh data yang reliabel dan menghasilkan model yang lebih stabil dan realistis.

Melalui praktikum ini, dilakukan penerapan metode inversi linear berbobot dan dibandingkan dengan inversi tanpa bobot. Tujuannya adalah untuk memahami pengaruh pembobotan terhadap hasil inversi, khususnya dalam menghadapi data yang mengandung outlier, serta mengevaluasi kualitas model berdasarkan nilai misfit dan kesesuaian terhadap tren data.</p>

<h3 style="text-align:left; font-size:13pt"><b>1.3.1 Konsep Dasar Inversi dalam Geofisika</b></h3>
<p>Dalam bidang geofisika, pengumpulan data di permukaan bumi dilakukan untuk memperkirakan kondisi yang ada di bawah permukaan. Data yang diperoleh merupakan respons dari struktur geologi bawah permukaan yang dihasilkan oleh variasi sifat fisik yang relevan, seperti densitas massa, resistivitas, karakteristik kemagnetan, dan kecepatan gelombang seismik. Untuk memperoleh distribusi sifat fisik di kedalaman secara kuantitatif, biasanya dilakukan pemodelan. Model tersebut berfungsi sebagai representasi keadaan geologi bawah permukaan oleh objek anomali dengan karakteristik fisik dan bentuk tertentu (Grandis, 2009).

Proses pemodelan inversi sering dianggap sebagai kebalikan dari pemodelan maju (forward modeling), karena dalam pemodelan inversi parameter model diperoleh langsung dari data. Teori inversi didefinisikan sebagai kumpulan teknik atau metode matematika dan statistik untuk memperoleh informasi mengenai suatu sistem fisik berdasarkan pengamatan terhadap sistem tersebut (Menke, 1984). Hubungan antara data dan parameter model dapat dinyatakan sebagai:
$$
d = Gm
$$
di mana:

- \(d\) adalah data 
- \(m\) adalah parameter model  
- \(G\) adalah matriks kernel

Solusi parameter model dapat diperoleh melalui manipulasi persamaan matriks sehingga menjadi:
$$
G^T G m = G^T d
$$

$$
m = (G^T G)^{-1} G^T d
$$
Persamaan tersebut merupakan solusi umum untuk masalah inversi linier menggunakan metode kuadrat terkecil yang banyak digunakan dalam pemodelan geofisika (Aster, Borchers, & Thurber, 2013).</p>

<h3 style="text-align:left; font-size:13pt"><b>1.3.2 Inversi Linear dan Metode Least Squares</b></h3>
<p>Dalam permasalahan geofisika, hubungan antara data observasi dan parameter model umumnya dapat dinyatakan dalam bentuk matematis. Untuk kasus linear, hubungan tersebut dituliskan sebagai:

d=Gm

di mana d adalah vektor data observasi, G adalah matriks kernel yang merepresentasikan hubungan fisika (forward modeling), dan m adalah parameter model yang ingin ditentukan.

Namun, dalam kondisi nyata, data observasi hampir selalu mengandung kesalahan (noise), sehingga persamaan tersebut tidak dapat dipenuhi secara eksak. Oleh karena itu, digunakan pendekatan optimasi dengan meminimalkan selisih antara data observasi (d obs) dan data hasil perhitungan (d cal). Selisih ini disebut residual.

<h3 style="text-align:left; font-size:13pt"><b>1.3.3 Permasalahan dalam Inversi Geofisika</b></h3>
<p> Umumnya data geofisika yang didapatkan dari lapangan mengandung noise dan outlier, maka dari itu sebelum di inversikan data harus bersih dari noise dan outlier. Noise pada data bisa berasal dari kesalahan alat ukur, lingkungan sekitar pengukuran dilakukan, adanya gangguan eksternal, dan terbatasnya resolusi. Noise pada data dapat menyebabkan data menyimpang dan mempengaruhi hasil representasi data. Outlier pada data merupakan nilai yang menyimpang dibandingkan data lain. Pada metode least squares, outlier dapat mempengaruhi hasil karena jika error dikuadratkan maka nilai besar menjadi dominan dan satu titik outlier bisa mempengaruhi solusi secara signifikan. Selain noise dan outlier, non-uniqueness juga menjadi permasalahan dalam menginversi data geofisika. Non-uniqueness adalah keadaan dimana ada lebih dari satu model untuk satu data. Umumnya hal ini disebabkan oleh terbatasnya jumlah data, adanya noise, dan ambiguitas. Hal ini dapat memengaruhi hasil interpretasi geofisika dan untuk memvalidasinya dibutuhkan informasi tambahan (a priori).

<h3 style="text-align:left; font-size:13pt"><b>3.1 Problem 1</b></h3> 

```python
import numpy as np                              #import library numpy untuk operasi matematis
import matplotlib.pyplot as plt                 #import library matplotlib untuk memvisualisasikan grafik
from numpy.linalg import inv                    #dari numpy untuk import fungsi invers matriks

# Generated data dummy (misal data kedalaman)
N = 20                                          #Mendefinisikan n titik data sebanyak 20

zmin = 0                                        #Mendefinisikan kedalaman minimum = 0
zmax = 10                                       #Mendefinisikan kedalaman max = 10

Z = np.sort(np.random.uniform(zmin, zmax, N))   #menambahkan titik kedalaman acak dari keil ke besar

# Generated data observasi temperatur
a = 2                                           #mendefinisikan a (intercept)
b = 2                                           #mendefinisikan b (gradien)

sd = 0.5                                        #Mendefinisikan standar deviasi

Tobs = a + b * Z + np.random.normal(0, sd, N)   #menghitung Tobs ditambah noise

# Membuat data outlier
Tobs[N-1] = 1                                   #mengganti data terakhir dengan nilai = 1 untuk mensimulasikan outlier 

# Membuat matriks bobot (subjective weight)
w = np.eye(N)                                   #membuat weighted matrix
w[N-1][N-1] = 0.1                               #mengubah bobot data outlier  menjadi 0.1 agar pengaruhnya terhadap solusi diminimalkan

# Membuat matriks kernel G
G = np.zeros((N, 2))                            #membuat matrix kernel nx2

for i in range(N):
    G[i][0] = 1                                 #matrix kolom pertama adalah 1
    G[i][1] = Z[i]                              #matrix kolom kedua koefisien parameter gradien b

# Mengubah data menjadi matriks kolom
d = np.reshape(Tobs, (N, 1))                    #mengubah matrix Tobs menjadi matrix kolom nx1

# Perhitungan inversi linear berbobot
m = inv(G.T @ w @ G) @ G.T @ w @ d              #menghitung m dengan weighted least square

a_est = m[0][0]                                 #mengekstrak nilai estimasi a dari hasil inversi berbobot
b_est = m[1][0]                                 #mengekstrak nilai estimasi b dari hasil inversi berbobot

# Model hasil inversi
Tcalc_weighted = a_est + b_est * Z              #menghitung T kalkulasi weighted dengan parameter hasil inversi

# Inversi tanpa bobot
m2 = inv(G.T @ G) @ G.T @ d                     #menghitung m dengan Least Squares

a2 = m2[0][0]                                   #mengekstrak nilai estimasi a dari hasil inversi tanpa bobot
b2 = m2[1][0]                                   #mengekstrak nilai estimasi b dari hasil inversi tanpa bobot

Tcalc_unweighted = a2 + b2 * Z                  #menghitung T kalkulasi unwighted 

# Menghitung misfit
misfit_weighted = np.sum((Tobs - Tcalc_weighted)**2)        #menghitung misfit weighted sebagai jumlah kuadrat selisih data observasi dan model berbobot
misfit_unweighted = np.sum((Tobs - Tcalc_unweighted)**2)    #menghitung misfit unweighted sebagai jumlah kuadrat selisih data observasi dan model 

print("Misfit with weighted :", misfit_weighted)            #menampilkan misfit hasil weighted inversion
print("Misfit without weighted :", misfit_unweighted)       #menampilkan misfit hasil unweighted inversion

# Plot hasil
plt.scatter(Z, Tobs, label='Observed Data') #memplot label data observasi
plt.plot(Z, Tcalc_weighted, label='Weighted', linewidth=2) #memplot label weighted
plt.plot(Z, Tcalc_unweighted, label='Without Weighted', linewidth=2) #memplot label without weighted

plt.xlabel("Depth") #menyatakan plot sumbu x adalah depth
plt.ylabel("Temperature") #menyatakan plot sumbu y adalah temperatur
plt.title("Inversi Linear Berbobot vs Non Bobot") #membuat judul 
plt.legend() #plot legenda
plt.show() #plot show
```

```python
# Problem 2 : Inversi Penentuan Hiposenter dengan Data Outlier

import numpy as np
import matplotlib.pyplot as plt
from numpy.linalg import inv

# ===============================
# 1. MODEL SEBENARNYA (TRUE MODEL)
# ===============================

# Hiposenter sebenarnya
x_true = 500
z_true = -400
t0_true = 0

# Kecepatan gelombang P (m/s)
V = 2000

# Koordinat 4 stasiun
stations = np.array([
    [0,0],
    [1000,0],
    [2000,0],
    [3000,0]
])

N = len(stations)

# ===============================
# 2. FORWARD MODELING
# ===============================

T_true = []

for i in range(N):
    
    dx = x_true - stations[i,0]
    dz = z_true - stations[i,1]
    
    r = np.sqrt(dx**2 + dz**2)
    
    T_true.append(t0_true + r/V)

T_true = np.array(T_true)

# ===============================
# 3. MENAMBAHKAN NOISE
# ===============================

noise = np.random.normal(0,0.01,N)
T_obs = T_true + noise

# membuat outlier pada stasiun ke-4
T_obs[3] = T_obs[3] + 0.05

# ===============================
# 4. TEBAKAN AWAL HIPOSENTER
# ===============================

x0 = 100
z0 = -100
t0 = 0

# untuk weighted
x_w = x0
z_w = z0
t_w = t0

# untuk unweighted
x_u = x0
z_u = z0
t_u = t0

# ===============================
# 5. MATRIKS BOBOT
# ===============================

W = np.eye(N)
W[3,3] = 0.1

# ===============================
# 6. MENYIMPAN ITERASI
# ===============================

path_x_w = []
path_z_w = []

path_x_u = []
path_z_u = []

# ===============================
# 7. ITERASI INVERSI (1000 ITERASI)
# ===============================

for it in range(1000):

    # ----- TANPA BOBOT -----
    
    G = np.zeros((N,3))
    d = np.zeros((N,1))
    
    for i in range(N):
        
        dx = x_u - stations[i,0]
        dz = z_u - stations[i,1]
        
        r = np.sqrt(dx**2 + dz**2)
        
        T_calc = t_u + r/V
        
        d[i] = T_obs[i] - T_calc
        
        G[i,0] = dx/(V*r)
        G[i,1] = dz/(V*r)
        G[i,2] = 1
    
    m = inv(G.T @ G) @ G.T @ d
    
    x_u = x_u + m[0,0]
    z_u = z_u + m[1,0]
    t_u = t_u + m[2,0]
    
    path_x_u.append(x_u)
    path_z_u.append(z_u)

    # ----- DENGAN BOBOT -----
    
    Gw = np.zeros((N,3))
    dw = np.zeros((N,1))
    
    for i in range(N):
        
        dx = x_w - stations[i,0]
        dz = z_w - stations[i,1]
        
        r = np.sqrt(dx**2 + dz**2)
        
        T_calc = t_w + r/V
        
        dw[i] = T_obs[i] - T_calc
        
        Gw[i,0] = dx/(V*r)
        Gw[i,1] = dz/(V*r)
        Gw[i,2] = 1
    
    mw = inv(Gw.T @ W @ Gw) @ Gw.T @ W @ dw
    
    x_w = x_w + mw[0,0]
    z_w = z_w + mw[1,0]
    t_w = t_w + mw[2,0]
    
    path_x_w.append(x_w)
    path_z_w.append(z_w)

# ===============================
# 8. HASIL INVERSI
# ===============================

print("Hiposenter sebenarnya :", x_true, z_true)

print("\nTanpa Bobot")
print("Hiposenter hasil :", x_u, z_u)

print("\nDengan Bobot")
print("Hiposenter hasil :", x_w, z_w)

# ===============================
# 9. PLOT ITERASI
# ===============================

plt.figure(figsize=(7,6))

plt.scatter(stations[:,0],stations[:,1],color='blue',label='Stasiun')
plt.scatter(x_true,z_true,color='green',label='Hiposenter True')

plt.plot(path_x_u,path_z_u,'r--',label='Tanpa Bobot')
plt.plot(path_x_w,path_z_w,'k-',label='Dengan Bobot')

plt.scatter(x_u,z_u,color='red')
plt.scatter(x_w,z_w,color='black')

plt.xlabel("X (m)")
plt.ylabel("Z (m)")
plt.title("Perbandingan Inversi Hiposenter (Weighted vs Unweighted)")

plt.legend()
plt.grid()

plt.show()
```

```python
import numpy as np                              #import library numpy untuk operasi matematis
import matplotlib.pyplot as plt                 #import library matplotlib untuk memvisualisasikan grafik
from numpy.linalg import inv                    #dari numpy untuk import fungsi invers matriks

#koordinat hiposenter (s1, s2,s3) dalam meter
hiposenter = np.array([         #mendefinisikan koordinat hiposenter
    [0, -250],                  #s 1
    [0, -633],                  #s 2
    [1000, 0]                   #s 3
])

#koordinat stasiun seismik (r1, r2) dalam meter
stasiun = np.array([            #mendefinisikan koordinat stasiun seismik
    [2000, -1000],              #r1
    [0, -1000]                  #r2
])

# Menghitung panjang jalur (ray path) menggunakan rumus Pythagoras
l_11 = np.sqrt((hiposenter[0, 0] - stasiun[0, 0])**2 + (hiposenter[0, 1] - stasiun[0, 1])**2)
l_21 = np.sqrt((hiposenter[1, 0] - stasiun[0, 0])**2 + (hiposenter[1, 1] - stasiun[0, 1])**2)
l_31 = np.sqrt((hiposenter[2, 0] - stasiun[0, 0])**2 + (hiposenter[2, 1] - stasiun[0, 1])**2)
#menghitung panjang ray path dari sumber gempa 

# Rasio R1b terhadap R1R2 (Setengah panjang R1R2)
r1b_ratio = 0.5 #nilai ini untuk membagi lintasan gelombang

#menghitung panjang jalur rib untuk tiap sumber gempa
delta_l1 = r1b_ratio * l_11         #menghitung panjang jalur rib untuk sumber gempa ke-1
delta_l2 = r1b_ratio * l_21         #menghitung panjang jalur rib untuk sumber gempa ke-2
delta_l3 = r1b_ratio * l_31         #menghitung panjang jalur rib untuk sumber gempa ke-3

#matriks desain G (panjang jalur untuk tiap blok)
G = np.array([                      #menyatakan matrix G (jalur ray) masing masing model
    [delta_l1, l_11 - delta_l1],    
    [delta_l2, l_21 - delta_l2],
    [delta_l3, l_31 - delta_l3]
])

#waktu tempuh observasi setelah dikoreksi
d_obs = np.array([1.544, 1.444, 0.744]) #s3 sudah dikoreksi

#matriks bobot (misal, bobot berdasarkan ketelitian data)
W = np.diag([1, 0.8, 1]) #tidak ada bobot khusus disini, jika ada kesalahan pada s3 maka nilai <1, dis

#menghitung inversi linear berbobot
GTWG = G.T @ W @ G
GTWd = G.T @ W @ d_obs
m = np.linalg.pinv(GTWG) @ GTWd #menggunakan pseudoinverse untuk menghindari singular matriks

#hasil estimasi kecepatan v1 dan v2
v1 = 1 / m[0] #kecepatan lapisan pertama
v2 = 1 / m[1] #kecepatan lapisan kedua

print(f"hasil estimasi v1 setelah koreksi: {v1: .4f} m/s")
print(f"hasil estimasi v2 setelah koreksi: {v1: .4f} m/s")

#plot regresi linear antara panjang jalur dan waktu tempuh
plt.figure(figsize=(8, 5))
plt.scatter(G[:, 0] + G[:, 1], d_obs, color='blue', label='Data Observasi (Dikoreksi)')
m_reg, b_reg = np.polyfit(G[:, 0] + G[:, 1], d_obs, 1)  # Regresi linear
plt.plot(G[:, 0] + G[:, 1], m_reg * (G[:, 0] + G[:, 1]) + b_reg, color='red', linestyle='--', label='Regresi Linear')

plt.xlabel('Panjang Jalur Total (m)')   #plot sumbu x menyatakan panjang jalur total
plt.ylabel('Waktu Tempuh (s)')          #plot sumbu y sebagai waktu tempuh
plt.title('Regresi Linear Waktu Tempuh vs Panjang Jalur (Afifah_123120155)') #plot judul
plt.legend()                            #plot legenda
plt.grid()                              #plot bentuk grid
plt.show()                              #plot 


#hasil estimasi sama karena v1=v2, dan data observasi terlihat fit dengan model
```

<h1 style="text-align:center; font-size:20pt"><b>BAB IV ANALISIS</b></h1>
<p>
Pada permasalahan pertama dilakukan inversi linear  antara temperatur dan kedalaman menggunakan model linear T = a + bZ. Model linear dapat dituliskan dalam bentuk umum d = Gm,  parameter model dicari dengan meminimalkan residual antara data observasi dan data kalkulasi.
Data dibuat 20 titik dan noise random ditambahkan. Lalu satu data terakhir dibuat outlier dan agar outlier tersebut tidak terlalu berpengaruh maka dibuat matriks dengan bobot kecil W[20,20] = 0.1.

Pada grafik terlihat garis weighted mengikuti tren data observasi, sedangkan unweighted menuju outlier. Hal ini terjadi karena data yang diberi weight menghasilkan estimasi yang stabil terhadap noise non-Gaussian. Hasil misfit dengan bobot menghasilkan 416.69969695221124 sedangkan misfit tanpa bobot menghasilkan 313.7572030652588, yang mana hasil ini menunjukkan selisih antara data ooobservasi dan hasil model. Karena misfit dengan bobot lebih besar artinya selisih model dengan data juga besar, begitupun sebaliknya. Akan tetapi data terakhir diubah Tobs[N-1] = 1 dan misfit dihitung secara keseluruhan. Nilai misfit weighted lebih besar dibanding unweighted karena model berbobot mengurangi pengaruh data outlier, sehingga tidak memaksakan fitting pada data anomali. Walaupun misfit total meningkat, model weighted lebih merepresentasikan tren utama data dan menghasilkan solusi yang lebih stabil. </p>
<p>
Pada permasalahan kedua dilakukan inversi nonlinear untuk menentukan posisi hiposenter. Digunakan inversi nonlinear karena waktu tempuh bergantung pada jarak Euclidean T = t0 + r/V dengan r = √((x-xi)<sup>2</sup> + (z-zi)<sup>2</sup>). Umumnya model nonlinear harus dilinearisasi terlebih dahulu dengan pendekatan Taylor expansion d ≈ GΔm lalu parameter diperbarui secara iteratif m<sub>baru</sub> = m<sub>lama</sub> + Δm. Pada permasalahan kedua dilakukan 1000 iterasi hingga mendapatkan hasil yang konvergen. Terlihat pada stasiun keempat mengandung outlier T_obs[3] = T_obs[3] + 0.05 maka dibuat <i>weighted matrix</i> W[4,4] = 0.1. Weighted inversion digunakan untuk mengurangi pengaruh <i>error picking arrival time</i>. Hasil menunjukkan tanpa bobot 517.44,-163.64 sedangkan dengan bobot 516.52,-271.26. Yang mana hasil dengan wighted lebih representatif terhadap model sebenarnya (500,-400). Hal ini membuktikan bahwa weighting meningkatkan kestabilan solusi nonlinear dan mempercepat konvergensi menuju minimum global.
</p>

<p>Pada permasalahan ketiga dilakukan estimasi kecepatan gelombang menggunakan hubungan linear travel time t = sL denngan t menyatakan travel time, s menyatkan slowness dan L menyatakan panjang ray path. Karena lowness lebih linear dibanding velocity maka dilakukan pembobotan matriks diagonal W = diag(1,0.8,1) dan menghasilkan v1 = 1473.3617 m/s dan v2 = 1473.3617 m/s. Nilai ini sama karena geometri model simetris, ry path seimbang maka sensitivitas v1 dan v2 sama. Plot regresi linear menunjukkan titik observasi hampir tepat pada garis tren model, sehingga residual kecil dan model fit dengan baik.
