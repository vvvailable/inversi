    <h1>Teori Dasar</h1>
</div>

<div style="font-family: 'Times New Roman', Times, serif; margin: 20px auto; width: 80%;">
    <p style="text-align: justify;">
        Inversi non-linear merupakan metode dalam geofisika yang digunakan untuk menentukan parameter model bawah permukaan ketika hubungan antara data observasi dan parameter model tidak bersifat linear. Berbeda dengan inversi linear, pada inversi non-linear hubungan antara data dan model dinyatakan dalam fungsi non-linear sehingga solusi tidak dapat diperoleh secara langsung, melainkan harus melalui proses iterasi. Tujuan utama dari inversi ini adalah meminimalkan selisih antara data hasil pengamatan dan data hasil perhitungan model (misfit) (Tarantola, 2005).
         </p>
    <p style="text-align: justify;">
        Dalam prosesnya, inversi non-linear biasanya dimulai dengan model awal (initial model) yang kemudian diperbaiki secara bertahap menggunakan algoritma iteratif. Beberapa metode yang umum digunakan antara lain metode Gauss-Newton method dan Levenberg-Marquardt algorithm, yang bekerja dengan melakukan linearisasi lokal terhadap fungsi non-linear di sekitar model saat ini. Pendekatan ini memungkinkan masalah non-linear dipecahkan sebagai rangkaian masalah linear hingga konvergen ke solusi optimum (Aster et al., 2013).
    <p style="text-align: justify;">
         Karakteristik utama inversi non-linear adalah adanya kemungkinan solusi yang tidak unik dan ketergantungan yang kuat terhadap model awal. Hal ini terjadi karena fungsi objektif dapat memiliki banyak minimum lokal, sehingga hasil inversi sangat dipengaruhi oleh tebakan awal yang digunakan. Oleh karena itu, pemilihan model awal yang baik dan penggunaan teknik regularisasi menjadi penting untuk memperoleh solusi yang stabil dan realistis (Menke, 2012).
    <p style="text-align: justify;">
        Selain itu, inversi non-linear juga sering dikombinasikan dengan konsep regularisasi untuk mengontrol kompleksitas model dan mengurangi efek noise pada data. Regularisasi dapat berupa penambahan kendala seperti smoothness atau kedekatan terhadap model referensi. Dengan demikian, solusi yang diperoleh tidak hanya cocok dengan data tetapi juga secara geologis masuk akal (Tikhonov & Arsenin, 1977).
    <p style="text-align: justify;">
        Secara umum, inversi non-linear banyak digunakan dalam berbagai metode eksplorasi geofisika seperti seismik, gravitasi, dan resistivitas. Kemampuannya dalam menangani hubungan kompleks antara data dan model menjadikannya alat yang sangat penting dalam interpretasi bawah permukaan, meskipun membutuhkan perhitungan yang lebih kompleks dibandingkan inversi linear (Tarantola, 2005).
    </p>

Problem 1

```python
import numpy as np              # Library untuk perhitungan numerik
import matplotlib.pyplot as plt # Library untuk visualisasi grafik
import copy                     # Untuk menyalin variabel
from numpy.linalg import inv    # Fungsi invers matriks

x = [0, 10, 20, 30]   # Koordinat x stasiun
z = [0, 0, 0, 0]      # Koordinat z stasiun (permukaan)
n = len(x)            # Jumlah stasiun

xo = 12   # Posisi x hiposenter sebenarnya
zo = 10   # Posisi z hiposenter sebenarnya
to = 0    # Waktu origin

v = 7     # Kecepatan gelombang

t = []
for i in range(n):
    t.append((1/v) * np.sqrt((z[i] * zo)**2 + (x[i] - xo)**2))  # Hitung waktu tempuh teoritis

noise = np.random.rand() * 0.03 * np.mean(t)  # Noise acak 3%
ti = t + noise  # Data observasi (data sintetik + noise)

M = [15, 30]                      # Tebakan awal hiposenter
init_position = copy.deepcopy(M)  # Simpan posisi awal
M1 = []                           # Variabel tidak terpakai

xhipo = [init_position[0]]  # Simpan lintasan x hasil inversi
yhipo = [init_position[1]]  # Simpan lintasan z hasil inversi
erms = []                   # Simpan RMS error

m = M  # Model awal

for niter in range(1000):   # Iterasi maksimum
    gm = []         # Data hasil model
    dgm_dm1 = []    # Turunan terhadap x
    dgm_dm2 = []    # Turunan terhadap z

    for i in range(n):
        gm.append(to + (1/v) * (np.sqrt((x[i] - m[0])**2 + (z[i] - m[1])**2)))  # Forward modelling

        dgm_dm1.append((1/v) * (-1 * (x[i] - m[0])) /
                       (np.sqrt((x[i] - m[0])**2 + (z[i] - m[1])**2)))  # Turunan parsial x

        dgm_dm2.append((1/v) * (-1 * (z[i] - m[1])) /
                       (np.sqrt((x[i] - m[0])**2 + (z[i] - m[1])**2)))  # Turunan parsial z

    j = np.zeros((n, 2))  # Matriks Jacobian

    for i in range(n):
        j[i][0] = dgm_dm1[i].item()  # Isi kolom x dengan nilai skalar
        j[i][1] = dgm_dm2[i].item()  # Isi kolom z dengan nilai skalar

    mo = copy.deepcopy(m)
    mo = np.reshape(mo, (2, 1))  # Model jadi vektor kolom
    ti = np.reshape(ti, (n, 1))  # Data observasi
    gm = np.reshape(gm, (n, 1))  # Data model

    model = np.asmatrix(mo) + np.asmatrix(inv(np.asmatrix(j.transpose()) * np.asmatrix(j))) * \
            np.asmatrix(j.transpose()) * (np.asmatrix(ti) - np.asmatrix(gm))  # Update model Gauss-Newton

    data = model.tolist()           # Konversi ke list
    m = [data[0][0], data[1][0]]    # Update parameter model dengan nilai skalar

    xhipo.append(data[0][0])  # Simpan x hasil iterasi
    yhipo.append(data[1][0])  # Simpan z hasil iterasi

    dt = ti - gm
    erms.append(np.sqrt(np.average(dt**2)))  # Hitung RMS error

plt.plot(12, 35, 'o', label='Initial Position', color='black')  # Plot posisi awal
plt.plot(xhipo, yhipo, 'o', label='Hypocenter Inversion')       # Plot titik hasil inversi
plt.plot(xhipo, yhipo, label='Hypocenter Inversion')            # Plot lintasan
plt.plot(xo, zo, 'o', alpha=0.5, color='red', label='True Hypocenter')  # Plot nilai sebenarnya

plt.title('Hypocenter Determination')
plt.xlabel('X Coordinate')
plt.ylabel('Y Coordinate')
plt.legend()

plt.figure()
plt.plot(erms)  # Plot RMS error
plt.title('Kurva RMS Error vs N Iterasi')
plt.ylabel('RMS Error')
plt.xlabel('N Iterasi')

plt.show()
```

Problem 2

```python
import numpy as np              # Library komputasi numerik
import matplotlib.pyplot as plt # Library untuk plotting grafik
import copy                     # Untuk copy variabel

x = [0, 10, 20, 30]   # Koordinat x stasiun
z = [0, 0, 0, 0]      # Koordinat z stasiun (permukaan)
n = len(x)            # Jumlah stasiun

xo = 12   # Posisi x hiposenter sebenarnya
zo = 10   # Posisi z hiposenter sebenarnya
to = 0    # Waktu origin

v = 7     # Kecepatan gelombang

t = []
for i in range(n):
    t.append((1/v) * np.sqrt((x[i] - xo)**2 + (z[i] - zo)**2))  # Hitung waktu tempuh teoritis

noise = np.random.rand() * 0.03 * np.mean(t)  # Noise acak 3%
ti = t + noise  # Data observasi (sudah diberi noise)

m = [15, 30]  # Tebakan awal hiposenter

alpha = 0.5   # Learning rate (besar langkah update)
niter = 100   # Jumlah iterasi maksimum

xhipo = [m[0]]  # Simpan lintasan x
yhipo = [m[1]]  # Simpan lintasan z
erms = []       # Simpan RMS error

for k in range(niter):   # Loop iterasi gradient descent
    gm = []         # Data hasil model
    dgm_dm1 = []    # Turunan terhadap x
    dgm_dm2 = []    # Turunan terhadap z

    for i in range(n):
        r = np.sqrt((x[i] - m[0])**2 + (z[i] - m[1])**2)  # Jarak sumber ke stasiun
        
        gm.append(to + (1/v) * r)  # Forward modelling (data sintetik)
        
        dgm_dm1.append((1/v) * (-(x[i] - m[0]) / r))  # Turunan parsial terhadap x
        dgm_dm2.append((1/v) * (-(z[i] - m[1]) / r))  # Turunan parsial terhadap z

    gm = np.array(gm)      # Ubah ke array
    ti_arr = np.array(ti)  # Data observasi jadi array

    grad_m1 = 0  # Gradien terhadap x
    grad_m2 = 0  # Gradien terhadap z

    for i in range(n):
        grad_m1 += dgm_dm1[i] * (gm[i] - ti_arr[i])  # Akumulasi gradien x
        grad_m2 += dgm_dm2[i] * (gm[i] - ti_arr[i])  # Akumulasi gradien z

    grad = np.array([grad_m1, grad_m2]) * 2  # Gradien total fungsi objektif

    m = m - alpha * grad  # Update model (gradient descent)

    xhipo.append(m[0])  # Simpan hasil x
    yhipo.append(m[1])  # Simpan hasil z

    error = gm - ti_arr
    erms.append(np.sqrt(np.mean(error**2)))  # Hitung RMS error

plt.figure()
plt.plot(xhipo, yhipo, 'o-', label='Lintasan Inversi')  # Plot lintasan iterasi
plt.plot(xo, zo, 'ro', label='True Hypocenter')         # Plot nilai sebenarnya
plt.plot(15, 30, 'ko', label='Initial Guess')           # Plot tebakan awal
plt.xlabel('X Coordinate')
plt.ylabel('Z Coordinate')
plt.title('Penentuan Hiposenter - Metode Gradien')
plt.legend()

plt.figure()
plt.plot(erms)  # Plot RMS error
plt.xlabel('Iterasi')
plt.ylabel('RMS Error')
plt.title('Kurva RMS Error vs Iterasi')

plt.show()  # Tampilkan semua grafik
```

<div style="text-align: center; font-family: 'Times New Roman', Times, serif;">
    <h1>Analisis</h1>
</div>

<div style="font-family: 'Times New Roman', Times, serif; margin: 20px auto; width: 80%;">
    <p style="text-align: justify;">
      Hasil penerapan metode Gauss-Newton dan Gradient Descent menunjukkan bahwa keduanya mampu mengestimasi posisi hiposenter dari data yang mengandung noise. Gauss-Newton memiliki konvergensi lebih cepat karena langsung mendekati solusi dalam beberapa iterasi awal, ditandai dengan penurunan RMS error yang tajam. Hal ini karena metode ini memanfaatkan matriks Jacobian melalui pendekatan least squares, sehingga arah pembaruan model lebih akurat. Sebaliknya, Gradient Descent memiliki konvergensi yang lebih lambat karena hanya mengikuti arah gradien tanpa mempertimbangkan kurvatur fungsi. Penurunan RMS error berlangsung bertahap dan sangat bergantung pada learning rate, di mana nilai yang tidak tepat dapat menyebabkan proses menjadi lambat atau tidak stabil. Perbedaan utama keduanya terletak pada mekanisme pembaruan model. Gauss-Newton menggunakan invers matriks sehingga lebih efisien dan cepat, tetapi lebih kompleks dan sensitif terhadap kondisi matriks. Sementara itu, Gradient Descent lebih sederhana dan stabil, namun membutuhkan iterasi lebih banyak. Secara umum, Gauss-Newton unggul dalam kecepatan dan akurasi, sedangkan Gradient Descent lebih fleksibel dan mudah diterapkan. Oleh karena itu, pemilihan metode perlu disesuaikan dengan kondisi data dan kebutuhan komputasi.
