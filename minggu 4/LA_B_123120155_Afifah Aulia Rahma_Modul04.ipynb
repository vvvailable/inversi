"><b>1.1 Latar Belakang</b></h2>
Metode inversi merupakan salah satu pendekatan penting dalam geofisika untuk memperoleh model bawah permukaan berdasarkan data observasi lapangan. Dalam praktiknya, proses inversi dilakukan dengan mencari parameter model yang menghasilkan respons sintetik mendekati data hasil pengukuran. Sebagian besar permasalahan inversi dalam geofisika pada dasarnya bersifat non-linear, artinya hubungan antara data dan parameter model tidak dapat dinyatakan secara sederhana dalam bentuk persamaan linear. Contoh nyata dari permasalahan inversi non-linear adalah penentuan lokasi hiposenter gempa bumi. Dalam kasus ini, hubungan antara waktu tempuh gelombang seismik dan posisi hiposenter merupakan fungsi non-linear yang melibatkan jarak Euclidean antara sumber gempa dan stasiun penerima. Karena sifatnya yang non-linear inilah diperlukan metode khusus yang mampu menyelesaikan permasalahan tersebut secara iteratif.
Terdapat beberapa pendekatan yang dikembangkan untuk menyelesaikan inversi non-linear, di antaranya adalah pendekatan linear, metode gradien, dan metode Newton. Pendekatan linear menyelesaikan masalah non-linear dengan melakukan linearisasi menggunakan ekspansi Taylor orde pertama di sekitar model awal, kemudian memperbaiki model secara iteratif melalui perturbasi yang dihitung dari matriks Jacobian. Metode gradien bekerja dengan mengoreksi model ke arah negatif gradien fungsi objektif, sehingga secara bertahap model bergerak menuju minimum error. Sementara itu, metode Newton menggunakan informasi turunan kedua fungsi objektif untuk mencari solusi yang lebih presisi, meskipun secara komputasi lebih mahal dibanding dua metode sebelumnya.
Masing-masing metode memiliki karakteristik konvergensi yang berbeda. Metode gradien cenderung cepat di iterasi awal namun melambat mendekati solusi, sedangkan metode Gauss-Newton justru sebaliknya yaitu lebih cepat di iterasi akhir. Kombinasi dari keduanya menghasilkan metode Levenberg-Marquardt yang memanfaatkan keunggulan masing-masing pendekatan tersebut. Pemahaman terhadap karakteristik ini penting agar praktisi dapat memilih metode yang tepat sesuai dengan kondisi data dan kompleksitas model yang dihadapi.
Oleh karena itu, praktikum ini dilakukan untuk memberikan pemahaman secara langsung mengenai cara kerja inversi non-linear dengan ketiga pendekatan tersebut melalui implementasi komputasi menggunakan Python. Dengan mempelajari kasus penentuan lokasi hiposenter gempa secara nyata, diharapkan pemahaman mengenai konsep inversi non-linear dapat lebih mudah dipahami dan diaplikasikan dalam permasalahan geofisika yang sesungguhnya.

"><b>1.3 Teori Dasar</b></h2>

<h3 style="text-align:left; font-size:13pt"><b>1.3.1 Konsep Dasar Inversi dalam Geofisika</b></h3>

<p>
Geofisika merupakan ilmu yang mempelajari kondisi bawah permukaan bumi berdasarkan pengukuran parameter fisis di permukaan. Data yang diperoleh dari pengukuran geofisika merupakan respons dari kondisi geologi bawah permukaan yang dipengaruhi oleh variasi sifat fisis material, seperti densitas, resistivitas, suseptibilitas magnetik, maupun kecepatan rambat gelombang seismik yang menghasilkan anomali untuk diinterpretasikan menjadi model bawah permukaan. Dalam geofisika, interpretasi data dilakukan melalui proses pemodelan. Model merupakan representasi matematis dari kondisi bawah permukaan yang dinyatakan dalam parameter fisis dan geometri tertentu. Melalui model, respons teoritis dapat dihitung menggunakan persamaan fisika sesuai metode geofisika yang digunakan.
</p>

<p>
Pemodelan geofisika terdiri atas dua proses utama, yaitu <i>forward modeling</i> dan <i>inverse modeling</i>. <i>Forward modeling</i> merupakan proses menghitung respons data berdasarkan model bawah permukaan yang telah diketahui. Sebaliknya, <i>inverse modeling</i> atau inversi merupakan proses memperkirakan parameter model berdasarkan data observasi. Dalam inversi, parameter model dicari sedemikian rupa sehingga respons hasil perhitungan memiliki kesesuaian terbaik dengan data pengamatan. Oleh karena itu, inversi sering disebut sebagai proses <i>data fitting</i>. Secara matematis, proses inversi dilakukan dengan meminimalkan fungsi objektif atau fungsi <i>misfit</i> yang menyatakan selisih antara data observasi dan data hasil perhitungan. Model optimum diperoleh ketika nilai <i>misfit</i> mencapai minimum.
</p>

<h3 style="text-align:left; font-size:13pt"><b>1.3.2 Inversi Non Linear</b></h3>

<p>
Inversi non-linear merupakan metode inversi yang digunakan ketika hubungan antara data observasi dan parameter model tidak dapat dinyatakan dalam bentuk persamaan linear. Secara umum hubungan tersebut dinyatakan sebagai:
</p>

$$
d = g(m)
$$

<p>
dengan d merupakan data observasi, m adalah parameter model, dan g(m) merupakan fungsi <i>forward modeling</i> non-linear.
</p>

<p>
Pada inversi non-linear, perubahan kecil pada parameter model tidak selalu menghasilkan perubahan linear pada data. Hal ini menyebabkan solusi model tidak dapat diperoleh secara langsung menggunakan invers matriks sederhana seperti pada inversi linear, melainkan memerlukan proses iteratif atau pencarian dalam ruang model. Karakteristik utama inversi non-linear adalah kemungkinan adanya banyak solusi lokal (<i>local minimum</i>) maupun solusi global (<i>global minimum</i>). Kondisi ini terjadi karena fungsi objektif dapat memiliki lebih dari satu nilai minimum.
</p>

Fungsi *forward modeling* digunakan untuk menghitung respons teoritis berdasarkan parameter model tertentu. Dalam praktikum Modul 4, fungsi tersebut digunakan untuk menentukan waktu tempuh gelombang seismik berdasarkan posisi hiposenter gempa.

Persamaan waktu tempuh gelombang pada medium homogen dapat dinyatakan sebagai:

$$
T_{cal}=T_0+\frac{1}{v}\sqrt{(x_i-x_0)^2+(z_i-z_0)^2}
$$

dengan:
- $T_{cal}$ = waktu tempuh hasil perhitungan,
- $T_0$ = origin time,
- $v$ = kecepatan gelombang,
- $(x_i,z_i)$ = koordinat stasiun,
- $(x_0,z_0)$ = koordinat hiposenter.

Karena hubungan waktu tempuh dengan posisi hiposenter bersifat non-linear, maka solusi model diperoleh menggunakan proses iteratif.

<h3 style="text-align:left; font-size:13pt"><b>1.3.3 Inversi Non-Linear dengan Pendekatan Linear</b></h3>

<p style="text-align: justify;">
Menurut Grandis (2009), beberapa permasalahan inversi non-linear dapat diselesaikan menggunakan pendekatan linear melalui proses linearisasi. Pendekatan ini dilakukan dengan mengubah hubungan non-linear menjadi bentuk linear di sekitar model awal sehingga solusi dapat diperoleh menggunakan metode <i>least squares</i>.
</p>

<p style="text-align: justify;">
Solusi inversi dinyatakan sebagai model awal yang diperturbasi dengan suatu koreksi model:

$$
m = m_0 + \Delta m
$$

</p>

<p style="text-align: justify;">
dengan:
</p>

<ul>
<li>m<sub>0</sub> = model awal,</li>
<li>Δm = perturbasi model.</li>
</ul>

<p style="text-align: justify;">
Jika persamaan tersebut disubstitusikan ke dalam hubungan non-linear, maka diperoleh:

$$ m = m_0 + \Delta m $$
</p>

<p style="text-align: justify;">
Dalam bentuk komponen dapat dituliskan sebagai:

$$
d_i = g_i(m_0^{(j)} + \Delta m_j)
$$
</p>

<p style="text-align: justify;">
dengan:
</p>

<ul>
<li>i = 1,2,3,\ldots,N sebagai jumlah data,</li>
<li>j = 1,2,3,\ldots,M sebagai jumlah parameter model.</li>
</ul>

<p style="text-align: justify;">
Untuk memperoleh bentuk linear dilakukan ekspansi Taylor orde pertama di sekitar model awal m<sub>0</sub> sehingga diperoleh:

$$
g_i(m_0^{(j)}+\Delta m_j)
\approx
g_i(m_0^{(j)})
+
\frac{\partial g_i}{\partial m_j}\Bigg|_{m_0}\Delta m_j
+
O(\Delta m_j)
$$

Suku O(Δm<sub>j</sub>) merupakan sisa suku turunan orde kedua dan orde lebih tinggi. Jika suku tersebut diabaikan maka diperoleh bentuk linear:

$$
d_i =
g_i(m_0^{(j)})
+
\frac{\partial g_i}{\partial m_j}\Bigg|_{m_0}\Delta m_j
$$
</p>


<h3 style="text-align:left; font-size:13pt"><b>1.3.4 Metode Newton</h3></b>
Metode Newton untuk inversi non-linear didasarkan pada prinsip minimisasi fungsi objektif E yang menyatakan misfit antara data observasi dan respons model:

<p align="center">E = [d - g(m)]ᵀ [d - g(m)]</p>

Solusi minimum dicari dengan menetapkan turunan pertama fungsi objektif terhadap parameter model sama dengan nol:

<p align="center">∂E/∂m = 0</p>

Turunan pertama dan kedua fungsi objektif masing-masing adalah:

<p align="center">∂E/∂m = 2Jᵀ(g(m) - d)</p>

<p align="center">∂²E/∂m² = 2(JᵀJ + Hᵀ(g(m) - d))</p>

di mana H adalah matriks Hessian yang berisi turunan kedua fungsi g. Dengan mengabaikan suku Hessian untuk menghindari kesulitan komputasi, persamaan iterasi Newton menjadi:

<p align="center">m_(n+1) = mₙ - [Jₙᵀ Jₙ]⁻¹ Jₙᵀ (g(mₙ) - d)</p>

Metode Newton memiliki kecepatan konvergensi yang lebih tinggi dibanding metode gradien, terutama ketika model sudah mendekati solusi optimum.



<h3 style="text-align:left; font-size:13pt"><b>1.3.5 Metode Gradien</h3></b>
<p>
Metode gradien bekerja berdasarkan prinsip bahwa gradien suatu fungsi menunjukkan arah peningkatan terbesar fungsi tersebut, sehingga arah negatif gradien bersesuaian dengan arah penurunan nilai fungsi. Pada setiap iterasi, model dikoreksi dengan bergerak ke arah negatif gradien fungsi objektif E:</p>

<p align="center">Δmₙ = -k [∂E/∂m] (m=mₙ)</p>

sehingga persamaan iterasinya menjadi:

<p align="center">m(n+1) = mₙ - 2k (Jₙᵀ(g(mₙ) - d))</p>

di mana k adalah step length yang menentukan seberapa besar langkah koreksi yang diambil pada setiap iterasi. Metode gradien cenderung cepat pada iterasi awal karena bergerak langsung ke arah penurunan terbesar, namun melambat ketika mendekati solusi karena gradien mengecil di sekitar minimum.

<h3 style="text-align:left; font-size:13pt"><b>1.3.6 Metode Lavenberg-MMarquardt</h3></b>
<p>
Mengingat bahwa metode gradien cepat di awal namun lambat di akhir, sedangkan metode Gauss-Newton lambat di awal namun cepat di akhir, kombinasi keduanya menghasilkan metode Levenberg-Marquardt yang memanfaatkan keunggulan masing-masing:

<p align="center">m(n+1) = mₙ + [Jₙᵀ Jₙ + ε²I]⁻¹ Jₙᵀ (d - g(mₙ))</p>

Parameter ε² berfungsi sebagai faktor damping yang mengontrol transisi antara perilaku metode gradien dan Gauss-Newton. Ketika ε² besar, metode berperilaku seperti gradien, sedangkan ketika ε² kecil, metode berperilaku seperti Gauss-Newton.

<h3 style="text-align:left; font-size:13pt"><b>1.3.7 RMS Error</h3></b>
<p>
Kualitas hasil inversi dinilai menggunakan Root Mean Square (RMS) error yang menyatakan rata-rata perbedaan antara data observasi dan respons model:

<p align="center">ERMS = √( (1/N) Σᵢ (dᵢ - gᵢ(m))² )</p>

Nilai RMS error yang semakin kecil menunjukkan bahwa model hasil inversi semakin baik dalam mencocokkan data observasi. Konvergensi tercapai ketika nilai RMS error sudah tidak berubah secara signifikan antar iterasi atau sudah berada di bawah nilai misfit yang ditetapkan.

```python
import numpy as np                    # Library untuk komputasi numerik/matriks
import matplotlib.pyplot as plt        # Library untuk visualisasi/plotting
import copy                           # Library untuk menyalin objek secara mendalam
from numpy.linalg import inv          # Fungsi khusus untuk menghitung invers matriks

# =========================
# DATA AWAL (STASIUN & PARAMETER)
# =========================
x = np.array([0, 10, 20, 30])         # Koordinat X dari 4 stasiun penerima
z = np.array([0, 0, 0, 0])             # Koordinat Z (kedalaman) dari 4 stasiun
n = len(x)                             # Jumlah stasiun yang digunakan

# True hypocenter (Posisi asli yang ingin dicari)
xo = 12                                # Koordinat X asli gempa
zo = 10                                # Koordinat Z asli gempa
to = 0                                 # Waktu asal (origin time) asli

# Kecepatan gelombang
v = 7                                  # Kecepatan rambat gelombang (km/s)

# =========================
# FORWARD MODELING (MENCARI DATA PENGAMATAN)
# =========================
t = []                                 # List kosong untuk menyimpan waktu tempuh
for i in range(n):                     # Loop sebanyak jumlah stasiun
    # Menghitung t = (jarak / kecepatan) menggunakan rumus Pythagoras
    t.append((1/v) * np.sqrt((x[i]-xo)**2 + (z[i]-zo)**2))

t = np.array(t)                        # Mengubah list waktu ke bentuk array NumPy

# Tambahkan noise (Agar data terlihat seperti data lapangan asli yang tidak sempurna)
noise = np.random.rand() * 0.03 * np.mean(t) # Membuat noise acak kecil
ti = t + noise                         # ti adalah waktu tiba yang "terkontaminasi" noise

# =========================
# INITIAL GUESS (TEBAKAN AWAL)
# =========================
m = np.array([15.0, 30.0])             # Tebakan posisi awal [x, z] (masih jauh dari asli)

xhipo = [m[0]]                         # List untuk mencatat riwayat pergerakan X
yhipo = [m[1]]                         # List untuk mencatat riwayat pergerakan Z
erms = []                              # List untuk mencatat nilai error setiap iterasi

# =========================
# ITERASI INVERSI (GAUSS-NEWTON)
# =========================
for niter in range(50):                # Melakukan pengulangan inversi sebanyak 50 kali

    gm = []                            # List waktu tempuh kalkulasi (model)
    dgm_dm1 = []                       # List turunan parsial terhadap x (Jacobian kolom 1)
    dgm_dm2 = []                       # List turunan parsial terhadap z (Jacobian kolom 2)

    for i in range(n):                 # Menghitung nilai untuk setiap stasiun
        # Menghitung jarak dari tebakan model ke stasiun
        r = np.sqrt((x[i]-m[0])**2 + (z[i]-m[1])**2)

        # Forward model berdasarkan parameter model saat ini
        gm.append(to + (1/v) * r)

        # Menghitung gradien/turunan fungsi (Jacobian)
        dgm_dm1.append((1/v) * (-(x[i]-m[0]) / r)) # Turunan terhadap x
        dgm_dm2.append((1/v) * (-(z[i]-m[1]) / r)) # Turunan terhadap z

    gm = np.array(gm).reshape(n, 1)    # Mengubah gm menjadi matriks kolom (n x 1)
    ti_reshape = ti.reshape(n, 1)      # Mengubah data observasi menjadi matriks kolom

    # Membangun Matriks Jacobian (n x 2)
    J = np.zeros((n, 2))               # Inisialisasi matriks nol ukuran n stasiun x 2 parameter
    J[:, 0] = dgm_dm1                  # Mengisi kolom 1 dengan turunan terhadap x
    J[:, 1] = dgm_dm2                  # Mengisi kolom 2 dengan turunan terhadap z

    # Update model (Persamaan Gauss-Newton: dm = (J^T J)^-1 J^T * delta_d)
    dm = inv(J.T @ J) @ J.T @ (ti_reshape - gm) # Menghitung perubahan model
    m = m + dm.flatten()               # Memperbarui model (model baru = model lama + dm)

    # Simpan hasil per iterasi untuk di-plot
    xhipo.append(m[0])
    yhipo.append(m[1])

    # Hitung error RMS (Root Mean Square) untuk melihat konvergensi
    dt = ti_reshape - gm               # Selisih data observasi dan kalkulasi
    erms.append(np.sqrt(np.mean(dt**2))) # Rumus RMS Error

# =========================
# PLOTTING (VISUALISASI HASIL)
# =========================
plt.figure()                           # Membuat jendela grafik pertama
plt.plot(15, 30, 'o', label='Initial Position', color='black') # Plot tebakan awal
plt.plot(xhipo, yhipo, 'o-', label='Hypocenter Inversion')    # Plot jejak inversi
plt.plot(xo, zo, 'o', color='red', label='True Hypocenter')   # Plot posisi asli

plt.title('Hypocenter Determination_123120155')  # Judul grafik
plt.xlabel('X Coordinate')             # Label sumbu X
plt.ylabel('Z Coordinate')             # Label sumbu Z (Kedalaman)
plt.legend()                           # Menampilkan keterangan label
plt.grid()                             # Menampilkan garis kotak-kotak

plt.figure()                           # Membuat jendela grafik kedua
plt.plot(erms)                         # Plot grafik penurunan error
plt.title('RMS Error vs Iteration_123120155')    # Judul grafik error
plt.xlabel('Iteration')                # Label sumbu X (Iterasi)
plt.ylabel('RMS Error')                # Label sumbu Y (Besar Error)
plt.grid()                             # Menampilkan garis kotak-kotak

plt.show()                             # Perintah untuk menampilkan semua grafik
```

```python
import pandas as pd   # Library untuk pengolahan data (tidak terlalu dipakai di sini)
import numpy as np    # Library utama untuk perhitungan numerik
import matplotlib.pyplot as plt  # Untuk plotting grafik
import copy   # Untuk copy variabel tanpa referensi


# DATA AWAL

# Posisi stasiun (receiver)
x = [0, 10, 20, 30]   # koordinat x stasiun
z = [0, 0, 0, 0]      # semua stasiun di permukaan (z=0)
n = len(x)            # jumlah stasiun

# Hiposenter sebenarnya (target)
x0 = 12
z0 = 10
t0 = 0                # waktu origin gempa

# Kecepatan gelombang (km/s)
v = 7


# FORWARD MODELLING


t = []
for i in range(n):
    # Menghitung waktu tempuh gelombang dari hiposenter ke stasiun
    t.append((1/v) * np.sqrt((z[i] - z0)**2 + (x[i] - x0)**2))

t = np.array(t)   # ubah ke array numpy


# TAMBAH NOISE

# Membuat noise acak sebesar 3% dari rata-rata waktu
noise = np.random.rand(n) * 0.03 * np.mean(t)

# Data observasi (data sintetik + noise)
ti = t + noise


# Tebakan AWAL MODEL

M = [15, 30]   # tebakan awal posisi hiposenter
init_position = copy.deepcopy(M)  # simpan posisi awal

# Learning rate (langkah update)
alpha = 0.1   # makin besar = cepat tapi bisa tidak stabil


# PENYIMPANAN HASIL

xhipo = [M[0]]   # menyimpan posisi x tiap iterasi
yhipo = [M[1]]   # menyimpan posisi z tiap iterasi
Erms = []        # menyimpan error tiap iterasi

# ITERASI GRADIENT DESCENT

for niter in range(200):  # jumlah iterasi
    gm = []          # data model (hasil perhitungan)
    dgm_dm1 = []     # turunan terhadap x
    dgm_dm2 = []     # turunan terhadap z

    for i in range(n):
        # jarak antara model hiposenter dan stasiun
        r = np.sqrt((x[i] - M[0])**2 + (z[i] - M[1])**2)

        # forward modelling (waktu tempuh)
        gm.append(t0 + (1/v) * r)

        # turunan parsial terhadap x (Jacobian)
        dgm_dm1.append((1/v) * (-(x[i] - M[0]) / r))

        # turunan parsial terhadap z (Jacobian)
        dgm_dm2.append((1/v) * (-(z[i] - M[1]) / r))

    # ubah ke bentuk matriks (vektor kolom)
    gm = np.array(gm).reshape(n,1)
    ti_vec = np.array(ti).reshape(n,1)


    # MATRIX JACOBIAN

    J = np.zeros((n,2))   # matriks Jacobian ukuran (n data x 2 parameter)

    for i in range(n):
        J[i,0] = dgm_dm1[i]   # turunan terhadap x
        J[i,1] = dgm_dm2[i]   # turunan terhadap z

    # RESIDUAL

    r = ti_vec - gm   # selisih data observasi dan model

    # GRADIENT

    grad = -2 * (J.T @ r)   # rumus gradien dari fungsi error

    # UPDATE MODEL

    M = [
        M[0] - alpha * grad[0][0],   # update koordinat x
        M[1] - alpha * grad[1][0]    # update koordinat z
    ]

    # simpan jalur iterasi
    xhipo.append(M[0])
    yhipo.append(M[1])

    # HITUNG ERROR (RMS)
    Erms.append(np.sqrt(np.mean(r**2)))  # RMS error

# PLOT HASIL

plt.figure(figsize=(7,5))

# titik tebakan awal
plt.plot(init_position[0], init_position[1], 'o', color='black', label='Initial')

# jalur pergerakan model
plt.plot(xhipo, yhipo, 'o-', label='Gradient Path')

# hiposenter sebenarnya
plt.plot(x0, z0, 'ro', label='True Hypocenter')

plt.title('Hypocenter - Gradient Descent_123120155')
plt.xlabel('X')
plt.ylabel('Z')
plt.legend()
plt.grid()

# PLOT ERROR

plt.figure(figsize=(7,5))
plt.plot(Erms)   # grafik error terhadap iterasi
plt.title('RMS Error vs Iteration (Gradient)_123120155')
plt.xlabel('Iteration')
plt.ylabel('Error')
plt.grid()

plt.show()

# HASIL AKHIR

print("Final Model:", M)   # hasil
```

<h1 style="text-align:center; font-size:20pt"><b>BAB IV ANALISIS</b></h1>
<p>Dari Gambar 1, terlihat bahwa proses inversi berhasil mengkonvergensi posisi hiposenter dari tebakan awal di koordinat (15, 30) menuju posisi yang sangat dekat dengan hiposenter sejati di (12, 10). Lintasan pergerakan model menunjukkan dua tahap yang menarik: pertama model bergerak ke arah kiri bawah menuju sekitar (9, 12), kemudian berbalik arah ke kanan menuju (12, 10). Pola ini terjadi karena pada iterasi awal algoritma Gauss-Newton melakukan koreksi besar berdasarkan gradien pertama, namun karena sifat linearisasi lokal dari ekspansi Taylor, koreksi tersebut sedikit melewati target sebelum akhirnya mengkonvergen. Secara keseluruhan, posisi akhir hasil inversi hampir berimpit dengan hiposenter sejati, menunjukkan bahwa metode ini berhasil bekerja dengan baik.
Dari Gambar 2, kurva RMS error menunjukkan penurunan yang sangat tajam dari nilai awal sekitar 2.5 pada iterasi pertama, kemudian turun drastis mendekati nol dalam sekitar 3 hingga 5 iterasi pertama saja. Setelah itu kurva menjadi mendatar hampir sempurna hingga iterasi ke-50. Pola konvergensi yang sangat cepat ini merupakan karakteristik khas metode Gauss-Newton, yaitu konvergensi kuadratik yang membuat error berkurang sangat drastis di iterasi awal. Nilai RMS error akhir yang mendekati nol mengindikasikan bahwa model hasil inversi mampu mencocokkan data traveltime observasi dengan sangat baik, dan 50 iterasi yang dilakukan jauh melebihi kebutuhan aktual karena konvergensi sudah tercapai sebelum iterasi ke-10.</p>
<p>Dari Gambar 3, terlihat hasil yang sangat berbeda dibanding Problem 1. Model dimulai dari tebakan awal di (15, 30) namun lintasan pergerakannya membentuk kurva melengkung yang bergerak ke kanan atas, bukan menuju hiposenter sejati di (12, 10) yang ditunjukkan oleh titik merah di pojok kiri bawah. Ini adalah indikasi yang jelas bahwa metode Gradient Descent dengan learning rate alpha = 0.1 dan 200 iterasi belum berhasil konvergen ke solusi yang benar. Hasil akhir yang tercetak di konsol menunjukkan posisi model berhenti di (13.20, 12.48), yang masih cukup jauh dari hiposenter sejati (12, 10). Penyebab utamanya adalah learning rate yang terlalu besar sehingga setiap langkah koreksi melompat melewati lembah minimum, dan jumlah iterasi 200 kali ternyata belum cukup untuk mencapai konvergensi dengan learning rate tersebut.
Dari Gambar 4, kurva RMS error memperlihatkan penurunan yang jauh lebih lambat dan gradual dibanding Gauss-Newton. Error turun dari sekitar 2.5 secara perlahan dan masih berada di sekitar 0.25 pada iterasi ke-200, belum mendekati nol. Ini membuktikan bahwa metode gradien memang lambat dalam konvergensi karena setiap langkah hanya bergerak sedikit demi sedikit mengikuti arah negatif gradien tanpa memanfaatkan informasi kelengkungan fungsi objektif seperti yang dilakukan Gauss-Newton. Pola penurunan yang halus dan eksponensial ini sesuai dengan teori bahwa metode gradien memiliki kecepatan konvergensi linear, jauh lebih lambat dibanding konvergensi kuadratik Gauss-Newton.</p>
<p>Secara keseluruhan, perbandingan kedua hasil ini menunjukkan perbedaan yang sangat signifikan dalam hal kecepatan dan kualitas konvergensi. Gauss-Newton berhasil mencapai solusi yang hampir tepat hanya dalam beberapa iterasi pertama dengan RMS error mendekati nol, sementara Gradient Descent dengan 200 iterasi sekalipun masih belum mencapai posisi hiposenter yang benar dan error masih tersisa sekitar 0.25. Hal ini terjadi karena Gauss-Newton memanfaatkan informasi matriks Jacobian secara penuh untuk menghitung arah dan besaran koreksi yang optimal, sedangkan Gradient Descent hanya menggunakan arah gradien tanpa mempertimbangkan kelengkungan fungsi sehingga langkahnya kurang efisien. Untuk memperbaiki hasil Gradient Descent, diperlukan learning rate yang lebih kecil misalnya alpha = 0.01, jumlah iterasi yang jauh lebih banyak, atau penggunaan metode Levenberg-Marquardt yang menggabungkan keunggulan keduanya.</p>

<p>Berdasarkan praktikum yang telah dilakukan, dapat disimpulkan.</p>
<ul>
<li>Persamaan inversi non-linear dalam bidang geofisika berhasil diselesaikan menggunakan dua pendekatan yang berbeda. Pada Problem 1 dengan pendekatan linear Gauss-Newton, algoritma berhasil mengkonvergensi posisi hiposenter dari tebakan awal (15, 30) menuju posisi yang hampir tepat berimpit dengan hiposenter sejati di (12, 10) hanya dalam beberapa iterasi pertama dari total 50 iterasi yang dilakukan. Pada Problem 2 dengan metode Gradient Descent, algoritma juga berhasil diimplementasikan namun posisi akhir model berhenti di (13.20, 12.48) setelah 200 iterasi, yang masih belum sepenuhnya mencapai hiposenter sejati akibat learning rate yang terlalu besar dan jumlah iterasi yang belum mencukupi.</li>
<li>Dari hasil analisis, fungsi dan karakteristik masing-masing metode dapat dipahami dengan baik. Pendekatan linear Gauss-Newton bekerja dengan melinearisasi fungsi non-linear menggunakan ekspansi Taylor orde pertama di sekitar model awal, kemudian menghitung perturbasi model secara iteratif menggunakan matriks Jacobian melalui persamaan Δm = (JᵀJ)⁻¹ Jᵀ(tᵢ - g(m)). Metode ini terbukti memiliki konvergensi yang sangat cepat karena memanfaatkan informasi kelengkungan fungsi objektif secara penuh, sehingga RMS error turun drastis dari 2.5 mendekati nol hanya dalam 3 hingga 5 iterasi pertama. Metode Gradien bekerja dengan mengoreksi model ke arah negatif gradien fungsi error menggunakan persamaan m_(n+1) = mₙ - α·grad, di mana setiap langkah hanya bergerak sedikit demi sedikit tanpa memanfaatkan informasi kelengkungan. Akibatnya konvergensinya jauh lebih lambat dan RMS error masih tersisa sekitar 0.25 setelah 200 iterasi. Metode Newton pada prinsipnya menggabungkan informasi turunan pertama dan kedua fungsi objektif untuk mencari solusi yang lebih presisi, namun membutuhkan komputasi yang lebih berat karena melibatkan matriks Hessian. Kombinasi metode Gradien dan Gauss-Newton menghasilkan metode Levenberg-Marquardt yang memanfaatkan kecepatan gradien di awal iterasi dan presisi Gauss-Newton di akhir iterasi, sehingga menjadi metode yang paling robust untuk permasalahan inversi non-linear dalam geofisika.</li>
</ul>
