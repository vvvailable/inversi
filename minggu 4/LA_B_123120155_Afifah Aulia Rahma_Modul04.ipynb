"><b>1.1 Latar Belakang</b></h2>
Metode inversi merupakan salah satu pendekatan penting dalam geofisika untuk memperoleh model bawah permukaan berdasarkan data observasi lapangan. Dalam praktiknya, proses inversi dilakukan dengan mencari parameter model yang menghasilkan respons sintetik mendekati data hasil pengukuran. Namun, permasalahan inversi geofisika umumnya bersifat tidak unik, sehingga satu set data observasi dapat dijelaskan oleh lebih dari satu model bawah permukaan. Ketidakunikan solusi menjadi semakin kompleks pada kasus under-determined, yaitu ketika jumlah parameter model lebih banyak dibandingkan jumlah data observasi. Kondisi tersebut menyebabkan ruang solusi menjadi sangat luas dan solusi inversi menjadi tidak stabil. Selain itu, keberadaan noise data dan keterbatasan observasi juga dapat memengaruhi kualitas model hasil inversi.

Untuk mengatasi permasalahan tersebut, diperlukan informasi tambahan atau a priori information yang berfungsi sebagai constraint dalam proses inversi. Informasi a priori dapat berupa model awal, batas parameter, maupun asumsi karakter model tertentu. Selain itu, penerapan regularisasi melalui parameter damping digunakan untuk menstabilkan solusi sehingga model yang diperoleh tidak hanya sesuai terhadap data, tetapi juga tetap realistis secara matematis dan fisis. Dalam proses inversi, penggunaan model referensi juga berperan penting sebagai acuan awal untuk mengarahkan iterasi solusi. Di samping itu, constraint flatness dan smoothness digunakan untuk mengontrol karakter model hasil inversi. Constraint flatness menghasilkan model yang relatif seragam antarparameter, sedangkan smoothness menghasilkan perubahan model yang lebih gradual dan halus. Oleh karena itu, pada praktikum Modul 3 ini dilakukan percobaan mengenai penerapan informasi a priori, regularisasi, model referensi, flatness, dan smoothness dalam penyelesaian masalah inversi linear untuk memahami pengaruh constraint terhadap kestabilan solusi dan kualitas model hasil inversi pada berbagai kondisi permasalahan geofisika.

<h3 style="text-align:left; font-size:13pt"><b>1.3.1 Konsep Dasar Inversi dalam Geofisika</b></h3>

<p>
Geofisika merupakan ilmu yang mempelajari kondisi bawah permukaan bumi berdasarkan pengukuran parameter fisis di permukaan. Data yang diperoleh dari pengukuran geofisika merupakan respons dari kondisi geologi bawah permukaan yang dipengaruhi oleh variasi sifat fisis material, seperti densitas, resistivitas, suseptibilitas magnetik, maupun kecepatan rambat gelombang seismik yang menghasilkan anomali untuk diinterpretasikan menjadi model bawah permukaan. Dalam geofisika, interpretasi data dilakukan melalui proses pemodelan. Model merupakan representasi matematis dari kondisi bawah permukaan yang dinyatakan dalam parameter fisis dan geometri tertentu. Melalui model, respons teoritis dapat dihitung menggunakan persamaan fisika sesuai metode geofisika yang digunakan.
</p>

<p>
Pemodelan geofisika terdiri atas dua proses utama, yaitu <i>forward modeling</i> dan <i>inverse modeling</i>. <i>Forward modeling</i> merupakan proses menghitung respons data berdasarkan model bawah permukaan yang telah diketahui. Sebaliknya, <i>inverse modeling</i> atau inversi merupakan proses memperkirakan parameter model berdasarkan data observasi. Dalam inversi, parameter model dicari sedemikian rupa sehingga respons hasil perhitungan memiliki kesesuaian terbaik dengan data pengamatan. Oleh karena itu, inversi sering disebut sebagai proses <i>data fitting</i>. Secara matematis, proses inversi dilakukan dengan meminimalkan fungsi objektif atau fungsi <i>misfit</i> yang menyatakan selisih antara data observasi dan data hasil perhitungan. Model optimum diperoleh ketika nilai <i>misfit</i> mencapai minimum.
</p>

<h3 style="text-align:left; font-size:13pt"><b>1.3.2 Inversi Non Linear</b></h3>

<p>
Inversi non-linear merupakan metode inversi yang digunakan ketika hubungan antara data observasi dan parameter model tidak dapat dinyatakan dalam bentuk persamaan linear. Secara umum hubungan tersebut dinyatakan sebagai:
</p>

$$
d = g(m)
$$

<p>
dengan d merupakan data observasi, m adalah parameter model, dan g(m) merupakan fungsi <i>forward modeling</i> non-linear.
</p>

<p>
Pada inversi non-linear, perubahan kecil pada parameter model tidak selalu menghasilkan perubahan linear pada data. Hal ini menyebabkan solusi model tidak dapat diperoleh secara langsung menggunakan invers matriks sederhana seperti pada inversi linear, melainkan memerlukan proses iteratif atau pencarian dalam ruang model. Karakteristik utama inversi non-linear adalah kemungkinan adanya banyak solusi lokal (<i>local minimum</i>) maupun solusi global (<i>global minimum</i>). Kondisi ini terjadi karena fungsi objektif dapat memiliki lebih dari satu nilai minimum.
</p>

Fungsi *forward modeling* digunakan untuk menghitung respons teoritis berdasarkan parameter model tertentu. Dalam praktikum Modul 4, fungsi tersebut digunakan untuk menentukan waktu tempuh gelombang seismik berdasarkan posisi hiposenter gempa.

Persamaan waktu tempuh gelombang pada medium homogen dapat dinyatakan sebagai:

$$
T_{cal}=T_0+\frac{1}{v}\sqrt{(x_i-x_0)^2+(z_i-z_0)^2}
$$

dengan:
- $T_{cal}$ = waktu tempuh hasil perhitungan,
- $T_0$ = origin time,
- $v$ = kecepatan gelombang,
- $(x_i,z_i)$ = koordinat stasiun,
- $(x_0,z_0)$ = koordinat hiposenter.

Karena hubungan waktu tempuh dengan posisi hiposenter bersifat non-linear, maka solusi model diperoleh menggunakan proses iteratif.

<h3 style="text-align:left; font-size:13pt"><b>1.3.3 Inversi Non-Linear dengan Pendekatan Linear</b></h3>

<p style="text-align: justify;">
Menurut Grandis (2009), beberapa permasalahan inversi non-linear dapat diselesaikan menggunakan pendekatan linear melalui proses linearisasi. Pendekatan ini dilakukan dengan mengubah hubungan non-linear menjadi bentuk linear di sekitar model awal sehingga solusi dapat diperoleh menggunakan metode <i>least squares</i>.
</p>

<p style="text-align: justify;">
Solusi inversi dinyatakan sebagai model awal yang diperturbasi dengan suatu koreksi model:

$$
m = m_0 + \Delta m
$$

</p>

<p style="text-align: justify;">
dengan:
</p>

<ul>
<li>m<sub>0</sub> = model awal,</li>
<li>Δm = perturbasi model.</li>
</ul>

<p style="text-align: justify;">
Jika persamaan tersebut disubstitusikan ke dalam hubungan non-linear, maka diperoleh:

$$ m = m_0 + \Delta m $$
</p>

<p style="text-align: justify;">
Dalam bentuk komponen dapat dituliskan sebagai:

$$
d_i = g_i(m_0^{(j)} + \Delta m_j)
$$
</p>

<p style="text-align: justify;">
dengan:
</p>

<ul>
<li>i = 1,2,3,\ldots,N sebagai jumlah data,</li>
<li>j = 1,2,3,\ldots,M sebagai jumlah parameter model.</li>
</ul>

<p style="text-align: justify;">
Untuk memperoleh bentuk linear dilakukan ekspansi Taylor orde pertama di sekitar model awal m<sub>0</sub> sehingga diperoleh:

$$
g_i(m_0^{(j)}+\Delta m_j)
\approx
g_i(m_0^{(j)})
+
\frac{\partial g_i}{\partial m_j}\Bigg|_{m_0}\Delta m_j
+
O(\Delta m_j)
$$

Suku O(Δm<sub>j</sub>) merupakan sisa suku turunan orde kedua dan orde lebih tinggi. Jika suku tersebut diabaikan maka diperoleh bentuk linear:

$$
d_i =
g_i(m_0^{(j)})
+
\frac{\partial g_i}{\partial m_j}\Bigg|_{m_0}\Delta m_j
$$
</p>


<h3 style="text-align:left; font-size:13pt"><b>1.3.4 Matriks Jacobian</h3></b>

<p style="text-align: justify;">
Turunan parsial pada persamaan linearisasi membentuk suatu matriks sensitivitas yang disebut matriks Jacobian. Matriks Jacobian menunjukkan sensitivitas perubahan data terhadap perubahan parameter model.
</p>

$$
J_{ij} =
\frac{\partial g_i}{\partial m_j}
$$
</p>

<p style="text-align: justify;">
dengan:
</p>

<ul>
<li>J<sub>ij</sub> = elemen matriks Jacobian,</li>
<li>g<sub>i</sub> = data hasil <i>forward modeling</i>,</li>
<li>m<sub>j</sub> = parameter model.</li>
</ul>

<p style="text-align: justify;">
Berdasarkan definisi Jacobian, hubungan linear antara perturbasi data dan perturbasi model dapat dituliskan sebagai:


$$
d-g(m_0)=J_0\Delta m_0
$$

$$
\Delta d_0 = d-g(m_0)
$$
</p>

<p style="text-align: justify;">
Menurut Grandis (2009), pada inversi non-linear matriks kernel digantikan oleh matriks Jacobian yang menyatakan sejauh mana data prediksi berubah akibat perturbasi model. Oleh karena itu, Jacobian sering disebut sebagai matriks sensitivitas.
</p>


<h3 style="text-align:left; font-size:13pt"><b>1.3.5 Metode Gauss-Newton</h3></b>

<p style="text-align: justify;">
Karena bentuk hubungan perturbasi data dan perturbasi model menyerupai inversi linear, maka solusi perturbasi model dapat dihitung menggunakan metode <i>least squares</i>.

$$
\Delta m =
[J_0^T J_0]^{-1}
J_0^T
(d-g(m_0))
$$
</p>

<p style="text-align: justify;">
Persamaan tersebut menyatakan koreksi model yang diperlukan terhadap model awal agar diperoleh model baru yang lebih sesuai dengan data observasi.

$$
m_1 = m_0 + \Delta m
$$
</p>

<p style="text-align: justify;">
Proses inversi dilakukan secara iteratif hingga diperoleh model optimum dengan nilai error minimum. Persamaan update model pada iterasi ke-(n+1) dinyatakan sebagai:

$$
m_{n+1}
=
m_n
+
[J_n^T J_n]^{-1}
J_n^T
(d-g(m_n))
$$
</p>

<p style="text-align: justify;">
Metode Gauss-Newton memiliki konvergensi yang relatif cepat karena memanfaatkan informasi sensitivitas model melalui matriks Jacobian.
</p>

```python
import numpy as np              # Library untuk perhitungan numerik
import matplotlib.pyplot as plt # Library untuk visualisasi grafik
import copy                     # Untuk menyalin variabel
from numpy.linalg import inv    # Fungsi invers matriks

x = [0, 10, 20, 30]   # Koordinat x stasiun
z = [0, 0, 0, 0]      # Koordinat z stasiun (permukaan)
n = len(x)            # Jumlah stasiun

xo = 12   # Posisi x hiposenter sebenarnya
zo = 10   # Posisi z hiposenter sebenarnya
to = 0    # Waktu origin

v = 7     # Kecepatan gelombang

t = []
for i in range(n):
    t.append((1/v) * np.sqrt((z[i] * zo)**2 + (x[i] - xo)**2))  # Hitung waktu tempuh teoritis

noise = np.random.rand() * 0.03 * np.mean(t)  # Noise acak 3%
ti = t + noise  # Data observasi (data sintetik + noise)

M = [15, 30]                      # Tebakan awal hiposenter
init_position = copy.deepcopy(M)  # Simpan posisi awal
M1 = []                           # Variabel tidak terpakai

xhipo = [init_position[0]]  # Simpan lintasan x hasil inversi
yhipo = [init_position[1]]  # Simpan lintasan z hasil inversi
erms = []                   # Simpan RMS error

m = M  # Model awal

for niter in range(1000):   # Iterasi maksimum
    gm = []         # Data hasil model
    dgm_dm1 = []    # Turunan terhadap x
    dgm_dm2 = []    # Turunan terhadap z

    for i in range(n):
        gm.append(to + (1/v) * (np.sqrt((x[i] - m[0])**2 + (z[i] - m[1])**2)))  # Forward modelling

        dgm_dm1.append((1/v) * (-1 * (x[i] - m[0])) /
                       (np.sqrt((x[i] - m[0])**2 + (z[i] - m[1])**2)))  # Turunan parsial x

        dgm_dm2.append((1/v) * (-1 * (z[i] - m[1])) /
                       (np.sqrt((x[i] - m[0])**2 + (z[i] - m[1])**2)))  # Turunan parsial z

    j = np.zeros((n, 2))  # Matriks Jacobian

    for i in range(n):
        j[i][0] = dgm_dm1[i].item()  # Isi kolom x dengan nilai skalar
        j[i][1] = dgm_dm2[i].item()  # Isi kolom z dengan nilai skalar

    mo = copy.deepcopy(m)
    mo = np.reshape(mo, (2, 1))  # Model jadi vektor kolom
    ti = np.reshape(ti, (n, 1))  # Data observasi
    gm = np.reshape(gm, (n, 1))  # Data model

    model = np.asmatrix(mo) + np.asmatrix(inv(np.asmatrix(j.transpose()) * np.asmatrix(j))) * \
            np.asmatrix(j.transpose()) * (np.asmatrix(ti) - np.asmatrix(gm))  # Update model Gauss-Newton

    data = model.tolist()           # Konversi ke list
    m = [data[0][0], data[1][0]]    # Update parameter model dengan nilai skalar

    xhipo.append(data[0][0])  # Simpan x hasil iterasi
    yhipo.append(data[1][0])  # Simpan z hasil iterasi

    dt = ti - gm
    erms.append(np.sqrt(np.average(dt**2)))  # Hitung RMS error

plt.plot(12, 35, 'o', label='Initial Position', color='black')  # Plot posisi awal
plt.plot(xhipo, yhipo, 'o', label='Hypocenter Inversion')       # Plot titik hasil inversi
plt.plot(xhipo, yhipo, label='Hypocenter Inversion')            # Plot lintasan
plt.plot(xo, zo, 'o', alpha=0.5, color='red', label='True Hypocenter')  # Plot nilai sebenarnya

plt.title('Hypocenter Determination')
plt.xlabel('X Coordinate')
plt.ylabel('Y Coordinate')
plt.legend()

plt.figure()
plt.plot(erms)  # Plot RMS error
plt.title('Kurva RMS Error vs N Iterasi')
plt.ylabel('RMS Error')
plt.xlabel('N Iterasi')

plt.show()
```

```python
import numpy as np              # Library komputasi numerik
import matplotlib.pyplot as plt # Library untuk plotting grafik
import copy                     # Untuk copy variabel

x = [0, 10, 20, 30]   # Koordinat x stasiun
z = [0, 0, 0, 0]      # Koordinat z stasiun (permukaan)
n = len(x)            # Jumlah stasiun

xo = 12   # Posisi x hiposenter sebenarnya
zo = 10   # Posisi z hiposenter sebenarnya
to = 0    # Waktu origin

v = 7     # Kecepatan gelombang

t = []
for i in range(n):
    t.append((1/v) * np.sqrt((x[i] - xo)**2 + (z[i] - zo)**2))  # Hitung waktu tempuh teoritis

noise = np.random.rand() * 0.03 * np.mean(t)  # Noise acak 3%
ti = t + noise  # Data observasi (sudah diberi noise)

m = [15, 30]  # Tebakan awal hiposenter

alpha = 0.5   # Learning rate (besar langkah update)
niter = 100   # Jumlah iterasi maksimum

xhipo = [m[0]]  # Simpan lintasan x
yhipo = [m[1]]  # Simpan lintasan z
erms = []       # Simpan RMS error

for k in range(niter):   # Loop iterasi gradient descent
    gm = []         # Data hasil model
    dgm_dm1 = []    # Turunan terhadap x
    dgm_dm2 = []    # Turunan terhadap z

    for i in range(n):
        r = np.sqrt((x[i] - m[0])**2 + (z[i] - m[1])**2)  # Jarak sumber ke stasiun
        
        gm.append(to + (1/v) * r)  # Forward modelling (data sintetik)
        
        dgm_dm1.append((1/v) * (-(x[i] - m[0]) / r))  # Turunan parsial terhadap x
        dgm_dm2.append((1/v) * (-(z[i] - m[1]) / r))  # Turunan parsial terhadap z

    gm = np.array(gm)      # Ubah ke array
    ti_arr = np.array(ti)  # Data observasi jadi array

    grad_m1 = 0  # Gradien terhadap x
    grad_m2 = 0  # Gradien terhadap z

    for i in range(n):
        grad_m1 += dgm_dm1[i] * (gm[i] - ti_arr[i])  # Akumulasi gradien x
        grad_m2 += dgm_dm2[i] * (gm[i] - ti_arr[i])  # Akumulasi gradien z

    grad = np.array([grad_m1, grad_m2]) * 2  # Gradien total fungsi objektif

    m = m - alpha * grad  # Update model (gradient descent)

    xhipo.append(m[0])  # Simpan hasil x
    yhipo.append(m[1])  # Simpan hasil z

    error = gm - ti_arr
    erms.append(np.sqrt(np.mean(error**2)))  # Hitung RMS error

plt.figure()
plt.plot(xhipo, yhipo, 'o-', label='Lintasan Inversi')  # Plot lintasan iterasi
plt.plot(xo, zo, 'ro', label='True Hypocenter')         # Plot nilai sebenarnya
plt.plot(15, 30, 'ko', label='Initial Guess')           # Plot tebakan awal
plt.xlabel('X Coordinate')
plt.ylabel('Z Coordinate')
plt.title('Penentuan Hiposenter - Metode Gradien')
plt.legend()

plt.figure()
plt.plot(erms)  # Plot RMS error
plt.xlabel('Iterasi')
plt.ylabel('RMS Error')
plt.title('Kurva RMS Error vs Iterasi')

plt.show()  # Tampilkan semua grafik
```
