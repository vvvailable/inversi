"><b>1.1 Latar Belakang</b></h2>

<p>
Dalam eksplorasi geofisika, pengukuran yang dilakukan di permukaan bumi menghasilkan respons fisika dari bawah permukaan bumi. Respons tersebut muncul akibat variasi parameter fisis seperti densitas, resistivitas, maupun kecepatan gelombang seismik yang berhubungan dengan kondisi bawah permukaan. Oleh karena itu, diperlukan suatu pendekatan untuk menginterpretasikan data tersebut menjadi model bawah permukaan yang representatif. Pendekatan yang digunakan untuk tujuan tersebut adalah pemodelan inversi geofisika, yaitu proses estimasi parameter model berdasarkan data observasi. Menurut Hendra Grandis, inversi merupakan kebalikan dari pemodelan ke depan (<i>forward modeling</i>), di mana pada <i>forward modeling</i> data dihitung dari model, sedangkan pada inversi model ditentukan dari data.
</p>

<p>
Data observasi sering mengandung <i>noise</i> (kesalahan pengukuran) dan <i>outlier</i> yang dapat menyebabkan kekeliruan hasil interpretasi. Selain itu, permasalahan inversi umumnya bersifat non-unik, di mana lebih dari satu model dapat menghasilkan respons data yang sama. Hal ini menyebabkan solusi inversi menjadi tidak stabil dan memerlukan pendekatan tambahan untuk memperoleh hasil yang lebih realistis.
</p>

<p>
Salah satu metode dasar dalam inversi adalah inversi linear dengan pendekatan <i>least squares</i>, yang bertujuan meminimalkan selisih antara data observasi dan data hasil perhitungan model. Metode ini banyak digunakan karena sederhana dan memiliki solusi matematis yang jelas. Akan tetapi, metode ini sangat sensitif terhadap data yang mengandung <i>outlier</i>, karena <i>error</i> dikuadratkan sehingga data ekstrem memiliki pengaruh yang besar terhadap solusi.
</p>

<p>
Untuk mengatasi kelemahan tersebut, digunakan metode inversi linear berbobot (<i>weighted least squares</i>). Pada metode ini, setiap data diberikan bobot sesuai tingkat kepercayaannya, sehingga data yang kurang reliabel dapat diminimalkan pengaruhnya terhadap hasil inversi. Pendekatan ini mampu meningkatkan kestabilan solusi dan menghasilkan model yang lebih representatif terhadap kondisi sebenarnya. Pada praktikum kali ini, konsep inversi linear berbobot diterapkan pada kasus regresi linear sederhana, penentuan hiposenter gempa secara iteratif, serta aplikasi pada tomografi seismik.
</p>

<h2 style="
font-size:14pt;
font-family:'Times New Roman';
margin-top:35px;
margin-bottom:10px;
"><b>1.3 Teori Dasar</b></h2>

<h3 style="text-align:left; font-size:13pt"><b>1.3.1 Konsep Dasar Inversi dalam Geofisika</b></h3>

<p>
Geofisika merupakan ilmu yang mempelajari kondisi bawah permukaan bumi berdasarkan pengukuran parameter fisis di permukaan. Data yang diperoleh dari pengukuran geofisika merupakan respons dari kondisi geologi bawah permukaan yang dipengaruhi oleh variasi sifat fisis material, seperti densitas, resistivitas, suseptibilitas magnetik, maupun kecepatan rambat gelombang seismik yang menghasilkan anomali untuk diinterpretasikan menjadi model bawah permukaan. Dalam geofisika, interpretasi data dilakukan melalui proses pemodelan. Model merupakan representasi matematis dari kondisi bawah permukaan yang dinyatakan dalam parameter fisis dan geometri tertentu. Melalui model, respons teoritis dapat dihitung menggunakan persamaan fisika sesuai metode geofisika yang digunakan.
</p>

<p>
Pemodelan geofisika terdiri atas dua proses utama, yaitu <i>forward modeling</i> dan <i>inverse modeling</i>. <i>Forward modeling</i> merupakan proses menghitung respons data berdasarkan model bawah permukaan yang telah diketahui. Sebaliknya, <i>inverse modeling</i> atau inversi merupakan proses memperkirakan parameter model berdasarkan data observasi. Dalam inversi, parameter model dicari sedemikian rupa sehingga respons hasil perhitungan memiliki kesesuaian terbaik dengan data pengamatan. Oleh karena itu, inversi sering disebut sebagai proses <i>data fitting</i>. Secara matematis, proses inversi dilakukan dengan meminimalkan fungsi objektif atau fungsi <i>misfit</i> yang menyatakan selisih antara data observasi dan data hasil perhitungan. Model optimum diperoleh ketika nilai <i>misfit</i> mencapai minimum.
</p>

<h3 style="text-align:left; font-size:13pt"><b>1.3.2 Inversi Non Linear</b></h3>

<p>
Inversi non-linear merupakan metode inversi yang digunakan ketika hubungan antara data observasi dan parameter model tidak dapat dinyatakan dalam bentuk persamaan linear. Secara umum hubungan tersebut dinyatakan sebagai:
</p>

$$
d = g(m)
$$

<p>
dengan d merupakan data observasi, m adalah parameter model, dan g(m) merupakan fungsi <i>forward modeling</i> non-linear.
</p>

<p>
Pada inversi non-linear, perubahan kecil pada parameter model tidak selalu menghasilkan perubahan linear pada data. Hal ini menyebabkan solusi model tidak dapat diperoleh secara langsung menggunakan invers matriks sederhana seperti pada inversi linear, melainkan memerlukan proses iteratif atau pencarian dalam ruang model. Karakteristik utama inversi non-linear adalah kemungkinan adanya banyak solusi lokal (<i>local minimum</i>) maupun solusi global (<i>global minimum</i>). Kondisi ini terjadi karena fungsi objektif dapat memiliki lebih dari satu nilai minimum.
</p>

<h3 style="text-align:left; font-size:13pt"><b>1.3.3 Fungsi Objektif dan Misfit Error</b></h3>

<p>
Dalam pemodelan inversi, tujuan utama yang ingin dicapai adalah memperoleh model yang mampu menghasilkan respons teoritis yang sesuai dengan data observasi.
</p>

Residual dinyatakan sebagai:

$$
r = d - g(m)
$$

Fungsi objektif:

$$
E = ||d - g(m)||^2
$$

Nilai <i>misfit error</i> menggunakan <i>Root Mean Square Error</i> (<i>RMSE</i>):

$$
RMSE = \sqrt{\frac{1}{N}\sum_{i=1}^{N}(d_i - g_i(m))^2}
$$

dengan:

<ul>
<li>N = jumlah data pengamatan</li>
<li>dᵢ = data observasi ke-i</li>
<li>gᵢ(m) = respons model pada data ke-i</li>
</ul>

<p>
Nilai <i>misfit</i> yang kecil menunjukkan bahwa model mampu merepresentasikan data observasi dengan baik.
</p>

<h3 style="text-align:left; font-size:13pt"><b>1.3.4 Metode Grid Search</b></h3>

<p>
Metode <i>grid search</i> merupakan salah satu metode optimasi global yang digunakan dalam pemodelan inversi non-linear untuk mencari kombinasi parameter model terbaik melalui evaluasi sistematis pada seluruh ruang parameter yang telah ditentukan.
</p>

Hubungan data dan model:

$$
d = g(m)
$$

Residual:

$$
r = d - g(m)
$$

Perhitungan <i>RMSE</i>:

$$
RMSE = \sqrt{\frac{1}{N}\sum_{i=1}^{N}(d_i - g_i(m))^2}
$$

Persamaan waktu tempuh teoritis:

$$
t_{cal} = \frac{\sqrt{(x-x_s)^2 + (y-y_s)^2}}{v}
$$

dengan:

<ul>
<li>t<sub>cal</sub> = waktu tempuh teoritis</li>
<li>(x,y) = koordinat hiposenter model</li>
<li>(x<sub>s</sub>,y<sub>s</sub>) = koordinat stasiun</li>
<li>v = kecepatan rata-rata gelombang</li>
</ul>

<p>
Model dengan selisih terkecil antara waktu observasi dan waktu kalkulasi dipilih sebagai solusi terbaik. Keunggulan metode <i>grid search</i> adalah sederhana, mudah diimplementasikan, dan tidak terlalu bergantung pada model awal sehingga mampu menghindari solusi lokal. Akan tetapi, metode ini memiliki kelemahan berupa kebutuhan komputasi yang besar, terutama jika jumlah parameter dan resolusi grid semakin tinggi.
</p>

<h3 style="text-align:left; font-size:13pt"><b>1.3.5 Metode Monte Carlo</b></h3>

<p>
Metode <i>Monte Carlo</i> merupakan salah satu metode optimasi global yang digunakan dalam inversi non-linear untuk mencari solusi optimum melalui proses sampling acak pada ruang parameter model. Berbeda dengan metode <i>grid search</i> yang mengevaluasi seluruh kombinasi parameter secara sistematis pada titik-titik grid tertentu, metode <i>Monte Carlo</i> melakukan pencarian dengan menghasilkan sejumlah model secara acak dalam batas parameter yang telah ditentukan.
</p>

<p>
Metode ini didasarkan pada prinsip probabilistik, yaitu peluang menemukan model optimum akan meningkat seiring bertambahnya jumlah sampel acak yang diuji. Dalam inversi geofisika, parameter model yang dicari dapat berupa posisi, kedalaman, kecepatan, resistivitas, maupun parameter fisik lainnya yang berkaitan dengan kondisi bawah permukaan.
</p>

<p>
Secara umum hubungan antara data observasi dan parameter model dinyatakan sebagai:
</p>

$$
d = g(m)
$$

<p>
dengan:
</p>

<ul>
<li>d = data observasi</li>
<li>g(m) = fungsi <i>forward modeling</i></li>
<li>m = parameter model</li>
</ul>

<p>
Pada metode <i>Monte Carlo</i>, model awal tidak terlalu menentukan hasil akhir karena parameter model dibangkitkan secara acak dalam rentang tertentu. Setiap model hasil sampling kemudian dihitung respons teoritisnya menggunakan fungsi <i>forward modeling</i>, lalu dibandingkan dengan data observasi.
</p>

<p>
Selisih antara data observasi dan data hasil perhitungan disebut residual, yang dinyatakan sebagai:
</p>

$$
r = d - g(m)
$$

<p>
Untuk mengevaluasi kualitas model, digunakan fungsi objektif atau <i>misfit error</i>. Pada praktikum ini, nilai <i>misfit</i> dihitung menggunakan <i>Root Mean Square Error</i> (<i>RMSE</i>):
</p>

$$
RMSE = \sqrt{\frac{1}{N}\sum_{i=1}^{N}(d_i - g_i(m))^2}
$$

<p>
dengan:
</p>

<ul>
<li>N = jumlah data pengamatan</li>
<li>dᵢ = data observasi ke-i</li>
<li>gᵢ(m) = respons model ke-i</li>
</ul>

<p>
Semakin kecil nilai <i>RMSE</i>, maka semakin baik model tersebut dalam merepresentasikan data observasi. Oleh karena itu, solusi optimum pada metode <i>Monte Carlo</i> ditentukan berdasarkan model yang memiliki nilai <i>misfit</i> minimum dari seluruh sampel acak yang dihasilkan.
</p>

<p>
Pada penentuan lokasi hiposenter gempa, metode <i>Monte Carlo</i> digunakan untuk menghasilkan koordinat hiposenter secara acak pada rentang area penelitian. Untuk setiap koordinat model, dihitung waktu tempuh teoritis gelombang menuju masing-masing stasiun menggunakan persamaan:
</p>

$$
t_{cal} = \frac{\sqrt{(x-x_s)^2 + (y-y_s)^2}}{v}
$$

<p>
dengan:
</p>

<ul>
<li>t<sub>cal</sub> = waktu tempuh teoritis</li>
<li>(x,y) = koordinat hiposenter model</li>
<li>(x<sub>s</sub>,y<sub>s</sub>) = koordinat stasiun</li>
<li>v = kecepatan rata-rata gelombang</li>
</ul>

<p>
Nilai waktu tempuh teoritis tersebut dibandingkan dengan waktu observasi untuk memperoleh nilai <i>misfit</i>. Model dengan nilai <i>misfit</i> terkecil dipilih sebagai estimasi lokasi hiposenter terbaik. Keunggulan metode <i>Monte Carlo</i> adalah kemampuannya mengeksplorasi ruang model secara luas dan tidak terlalu sensitif terhadap model awal, sehingga lebih mampu menghindari jebakan solusi lokal (<i>local minimum</i>). Selain itu, metode ini relatif sederhana untuk diimplementasikan pada permasalahan inversi dengan parameter banyak atau hubungan model yang kompleks.
</p>

<p>
Namun demikian, metode <i>Monte Carlo</i> memiliki kelemahan berupa kebutuhan jumlah iterasi yang besar agar solusi optimum dapat ditemukan dengan baik. Semakin banyak jumlah sampel acak yang digunakan, maka peluang memperoleh solusi mendekati minimum global (<i>global minimum</i>) semakin tinggi, tetapi waktu komputasi juga meningkat. Dalam praktikum inversi non-linear, metode <i>Monte Carlo</i> digunakan sebagai pendekatan alternatif terhadap metode <i>grid search</i> untuk menentukan lokasi hiposenter gempa. Perbandingan kedua metode dilakukan berdasarkan nilai <i>misfit</i>, jumlah iterasi, serta efisiensi komputasi dalam menemukan model optimum.
</p>

<h1 style="text-align:center; font-size:20pt"><b>BAB III HASIL DAN PENGOLAHAN DATA</b></h1> 
<h3 style="text-align:left; font-size:13pt"><b>3.1 Grid Search</b></h3> </div>

```python
import numpy as np                              #import library numpy untuk operasi matematis
import matplotlib.pyplot as plt                 #import library matplotlib untuk memvisualisasikan grafik

Vavg = 4                                        #Velocity rata-rata gelombang seismik
tobs = [7.1, 1.8, 5.0, 7.9]                    #Waktu observasi pada tiap stasiun

Ksts = [20, 50, 40, 10]                        #Koordinat X stasiun
Ysts = [10, 25, 50, 40]                        #Koordinat Y stasiun

# Grid
L = 1                                           #Interval grid pencarian
X0 = np.arange(0, 50, L)                       #Membuat grid koordinat X dari 0–49
Y0 = np.arange(0, 50, L)                       #Membuat grid koordinat Y dari 0–49

# Grid search
P = 10                                          #Nilai awal misfit minimum
HypoX = []                                      #List untuk menyimpan koordinat X hiposenter
HypoY = []                                      #List untuk menyimpan koordinat Y hiposenter

for i in range(len(Y0)):                        #Perulangan untuk seluruh grid Y
    for j in range(len(X0)):                    #Perulangan untuk seluruh grid X
        Dt = []                                 #List untuk menyimpan selisih waktu kuadrat
        for k in range(len(Ksts)):              #Perulangan untuk setiap stasiun
            tcal = np.sqrt((X0[j]-Ksts[k])**2 + (Y0[i]-Ysts[k])**2) / Vavg #Menghitung waktu tempuh teoritis berdasarkan jarak dan velocity 
            Dt.append((tobs[k] - tcal)**2)      #Menghitung selisih kuadrat antara waktu observasi dan waktu teoritis
                                                 

        misfit = np.sqrt(np.mean(Dt))           #Menghitung RMS misfit
        
        if misfit < P:                          #Seleksi jika misfit lebih kecil
            HypoX.append(X0[j])                 #Menyimpan koordinat X terbaik
            HypoY.append(Y0[i])                 #Menyimpan koordinat Y terbaik
            P = misfit                          #Update nilai misfit minimum

# Plot
plt.figure(figsize=(10, 8))                    #Membuat ukuran figure grafik

# Grid
for x in X0:                                    #Perulangan membuat garis grid vertikal
    plt.plot([x, x], [Y0[0], Y0[-1]], 'k-', alpha=0.1)#Plot garis vertikal grid                          

for y in Y0:                                    #Perulangan membuat garis grid horizontal
    plt.plot([X0[0], X0[-1]], [y, y], 'k-', alpha=0.1) #Plot garis horizontal grid
                                                 
# Jalur pencarian
plt.plot(HypoX, HypoY, '-r', label='Jalur Pencarian')                   #Menampilkan jalur pencarian hiposenter
plt.plot(HypoX[0], HypoY[0], '*c', markersize=10, label='Awal')         #Menampilkan titik awal pencarian                                               
plt.plot(HypoX[-1], HypoY[-1], '*y', markersize=10, label='Akhir')      #Menampilkan titik akhir pencarian
                                                 
# Stasiun
plt.plot(Ksts, Ysts, 'v', markersize=8, color='blue', label='Stasiun')  #Menampilkan posisi stasiun pengamatan                 

# Label
plt.xlabel('Koordinat X')                       #Label sumbu X
plt.ylabel('Koordinat Y')                       #Label sumbu Y
plt.title('Pencarian Hiposenter dengan Grid Search_Afifah Aulia Rahma')#Judul grafik
plt.legend()                                    #Menampilkan legenda
plt.grid(True)                                  #Menampilkan grid utama
plt.show()                                      #Menampilkan grafik

# Output
print('Koordinat X Hiposenter :', HypoX[-1])    #Menampilkan koordinat X hiposenter
print('Koordinat Y Hiposenter :', HypoY[-1])    #Menampilkan koordinat Y hiposenter
print('Nilai Misfit Minimum   :', P)            #Menampilkan nilai misfit minimum
print('Jumlah Iterasi:', len(HypoX))            #Menampilkan jumlah iterasi yang dilakukan dalam pencarian hiposenter
```

<h3 style="text-align:left; font-size:13pt"><b>3.2 Monte Carlo</b></h3> 

```python
# Monte Carlo
import numpy as np                              # Import library numpy untuk operasi matematis
import matplotlib.pyplot as plt                 # Import library matplotlib untuk visualisasi grafik
from numpy.linalg import inv                    # Import fungsi invers matriks dari numpy.linalg

Vavg = 4                                        # Kecepatan rata-rata gelombang seismik
tobs = [7.1,1.8,5.0,7.9]                       # Data waktu observasi pada tiap stasiun

#tobs = np.reshape(tobs,(len(tobs),1))         # Mengubah array menjadi matriks kolom (tidak digunakan)

Ksts = [20,50,40,10]                           # Koordinat X stasiun
Ysts = [10,25,50,40]                           # Koordinat Y stasiun

# Tebakan awal
xo_initial = 12                                # Koordinat X awal hiposenter
yo_initial = 10                                # Koordinat Y awal hiposenter

# Fungsi untuk menghitung waktu kalkulasi (tcal)
def calculate_tcal(xo, yo, xsts, ysts, vavg):
    return np.sqrt((xsts - xo)**2 + (ysts - yo)**2) / vavg  # Menghitung waktu tempuh teoritis

# Fungsi untuk menghitung error (misfit)
def calculate_misfit(tobs, tcal):
    tobs = np.array(tobs)                      # Mengubah data observasi menjadi array numpy
    tcal = np.array(tcal)                      # Mengubah data kalkulasi menjadi array numpy
    
    return np.sqrt(np.mean((tobs - tcal)**2)) # Menghitung RMS misfit

# Inisialisasi variabel
best_xo = xo_initial                           # Menyimpan koordinat X terbaik sementara
best_yo = yo_initial                           # Menyimpan koordinat Y terbaik sementara
best_misfit = float('inf')                    # Nilai awal misfit terbaik = tak hingga
n_iterations = 666                             # Jumlah iterasi Monte Carlo

# Lakukan pencarian acak (Monte Carlo)
for _ in range(n_iterations):                  # Perulangan Monte Carlo
    xo_random = np.random.uniform(0, 50)      # Membuat koordinat X acak
    yo_random = np.random.uniform(0, 50)      # Membuat koordinat Y acak

    tcal_random = [calculate_tcal(xo_random, yo_random, xsts, ysts, Vavg) for xsts, ysts in zip(Ksts, Ysts)]  # Menghitung waktu teoritis

    misfit_random = calculate_misfit(tobs, tcal_random)  # Menghitung misfit model acak

    if misfit_random < best_misfit:            # Seleksi model dengan misfit terkecil
        best_xo = xo_random                    # Update koordinat X terbaik
        best_yo = yo_random                    # Update koordinat Y terbaik
        best_misfit = misfit_random            # Update nilai misfit terbaik

# Plot hasil akhir
plt.plot(Ksts, Ysts, 'v', label='Stasiun')     # Menampilkan posisi stasiun
plt.plot(best_xo, best_yo, '*r', label='Hiposenter Terbaik')  # Menampilkan hiposenter terbaik
plt.xlabel('Koordinat X')                      # Label sumbu X
plt.ylabel('Koordinat Y')                      # Label sumbu Y
plt.legend()                                   # Menampilkan legenda
plt.title(f'Lokasi Hiposenter Terbaik: ({best_xo:.2f}, {best_yo:.2f})')  # Judul grafik
plt.grid(True)                                 # Menampilkan grid
plt.show()                                     # Menampilkan grafik

print(f"Lokasi Hiposenter Terbaik: ({best_xo:.2f}, {best_yo:.2f})")  # Output koordinat terbaik
print(f"Misfit Terbaik: {best_misfit:.2f}")    # Output nilai misfit terbaik
```

<h1 style="text-align:center; font-size:20pt"><b>BAB IV ANALISIS</b></h1>
Berdasarkan hasil yang diperoleh, metode <i>Grid Search</i> menghasilkan lokasi hiposenter pada koordinat (42, 29) dengan nilai <i>misfit minimum</i> sebesar 0.3892, sedangkan metode <i>Monte Carlo</i> menghasilkan lokasi terbaik pada koordinat (42.03, 28.10) dengan nilai <i>misfit</i> sebesar 0.44. Hasil dari kedua metode ini terlihat sangat berdekatan, sehingga dapat dikatakan bahwa keduanya berhasil menemukan area solusi yang sama atau mendekati solusi optimum.

Pada grafik <i>Grid Search</i>, garis merah menunjukkan jalur pencarian model selama proses inversi. Metode ini bekerja dengan mengecek seluruh kombinasi koordinat dalam <i>grid</i> secara sistematis, kemudian menghitung waktu tempuh teoritis dari setiap titik model ke masing-masing stasiun. Setelah itu, nilai waktu teoritis dibandingkan dengan data observasi untuk memperoleh nilai <i>misfit</i>. Dari proses tersebut, model terbaik ditemukan pada koordinat (42,29) setelah melalui 176 iterasi. Nilai <i>misfit</i> yang cukup kecil menunjukkan bahwa selisih antara data observasi dan hasil perhitungan model sudah cukup rendah, sehingga model dapat dianggap cukup baik.

Jika dilihat dari posisi stasiun, hiposenter hasil inversi berada di bagian kanan grafik, dekat dengan stasiun pada koordinat sekitar (40,50) dan (50,25). Hal ini cukup masuk akal karena salah satu waktu observasi bernilai paling kecil, yaitu 1.8 s, yang menunjukkan sumber gempa kemungkinan berada dekat dengan salah satu stasiun tersebut. Dengan asumsi kecepatan gelombang konstan sebesar 4 km/s, model menghitung waktu rambat berdasarkan jarak dari hiposenter ke setiap stasiun.

Pada metode <i>Monte Carlo</i>, pencarian model dilakukan secara acak sebanyak 666 iterasi. Setiap iterasi menghasilkan koordinat acak, kemudian dihitung waktu teoritis dan nilai <i>misfit</i>-nya. Dari seluruh model acak tersebut, solusi terbaik diperoleh pada koordinat (42.03, 28.10). Hasil ini sangat dekat dengan hasil <i>Grid Search</i>, meskipun sedikit berbeda pada koordinat Y. Perbedaan kecil ini wajar karena metode <i>Monte Carlo</i> menggunakan <i>sampling</i> acak, sehingga hasilnya bergantung pada distribusi model yang terambil selama iterasi.

Jika dibandingkan, metode <i>Grid Search</i> menghasilkan nilai <i>misfit</i> yang lebih kecil dibandingkan <i>Monte Carlo</i> (0.3892 &lt; 0.44), sehingga pada kasus ini <i>Grid Search</i> memberikan hasil yang sedikit lebih baik. Hal ini sesuai teori, karena <i>Grid Search</i> melakukan pencarian secara menyeluruh pada ruang model sehingga lebih teliti untuk jumlah parameter yang sedikit. Sementara itu, <i>Monte Carlo</i> lebih fleksibel dan cocok untuk ruang model yang lebih kompleks, walaupun akurasinya dipengaruhi jumlah iterasi yang digunakan.

Secara keseluruhan, kedua metode berhasil memberikan estimasi lokasi hiposenter yang hampir sama. Ini menunjukkan bahwa proses inversi yang dilakukan sudah cukup stabil dan model yang diperoleh mampu merepresentasikan data observasi dengan baik. Dari praktikum ini dapat dipahami bahwa <i>Grid Search</i> lebih unggul dari segi akurasi pada model sederhana, sedangkan <i>Monte Carlo</i> memiliki kelebihan dalam eksplorasi model secara global melalui pencarian acak.
