"><b>1.1 Latar Belakang</b></h2>
<p>
Geofisika merupakan ilmu yang mempelajari kondisi bawah permukaan bumi berdasarkan pengukuran parameter fisis di permukaan. Salah satu metode yang umum digunakan dalam eksplorasi geofisika adalah metode geolistrik resistivitas, yaitu metode yang memanfaatkan perbedaan sifat tahanan jenis batuan untuk mengidentifikasi struktur dan litologi bawah permukaan. Data hasil pengukuran geolistrik berupa resistivitas semu (<i>apparent resistivity</i>) masih dipengaruhi oleh kondisi geometri pengukuran dan sifat lapisan di bawah permukaan, sehingga belum dapat langsung diinterpretasikan sebagai model resistivitas sebenarnya.
</p>

<p>
Untuk mendapatkan model yang representatif, data perlu proses inversi, yaitu proses memperkirakan parameter model berdasarkan data observasi. Dalam metode geolistrik, parameter model yang dicari umumnya berupa nilai resistivitas tiap lapisan dan ketebalan lapisan. Permasalahan inversi pada geofisika umumnya bersifat non-linear karena hubungan antara model dan data observasi tidak dapat dinyatakan dalam bentuk linear sederhana. Selain itu, inversi juga memiliki sifat non-unik, yaitu lebih dari satu model dapat menghasilkan respons data yang mirip.
</p>

<p>
Permasalahan non-linear dan non-unik tersebut menyebabkan pencarian solusi optimum menjadi lebih rumit. Metode optimisasi konvensional umumnya sensitif terhadap model awal dan berisiko terjebak pada solusi lokal (<i>local minimum</i>). Oleh karena itu, dibutuhkan metode optimisasi global yang mampu mengeksplorasi ruang model secara lebih luas untuk memperoleh solusi yang lebih optimum.
</p>

<p>
Salah satu algoritma optimisasi global yang banyak digunakan adalah <i>Particle Swarm Optimization</i> (PSO). Algoritma ini dikembangkan berdasarkan perilaku kolektif kawanan burung atau ikan dalam mencari sumber makanan. Pada PSO, setiap partikel mewakili kandidat solusi yang bergerak di dalam ruang model dengan mempertimbangkan pengalaman terbaik individu (<i>personal best</i>) dan pengalaman terbaik seluruh populasi (<i>global best</i>). Pendekatan ini memungkinkan algoritma melakukan eksplorasi dan eksploitasi ruang solusi secara efisien.
</p>

<p>
Pada praktikum ini, algoritma PSO diterapkan pada inversi data geolistrik 1D metode <i>Vertical Electrical Sounding</i> (VES) untuk memperoleh model resistivitas bawah permukaan. Proses inversi dilakukan dengan mencocokkan data observasi resistivitas semu dengan data hasil <i>forward modelling</i>, kemudian memilih model dengan nilai <i>misfit</i> minimum sebagai solusi optimum. Melalui praktikum ini diharapkan praktikan dapat memahami konsep optimisasi global serta penerapan algoritma PSO dalam menyelesaikan permasalahan inversi geofisika.
</p>

"><b>1.3 Teori Dasar</b></h2>

<h3 style="text-align:left; font-size:13pt"><b>1.3.1 Konsep Dasar Inversi dalam Geofisika</b></h3>

<p>
Geofisika merupakan ilmu yang mempelajari kondisi bawah permukaan bumi berdasarkan pengukuran parameter fisis di permukaan. Data yang diperoleh dari pengukuran geofisika merupakan respons dari kondisi geologi bawah permukaan yang dipengaruhi oleh variasi sifat fisis material, seperti densitas, resistivitas, suseptibilitas magnetik, maupun kecepatan rambat gelombang seismik yang menghasilkan anomali untuk diinterpretasikan menjadi model bawah permukaan. Dalam geofisika, interpretasi data dilakukan melalui proses pemodelan. Model merupakan representasi matematis dari kondisi bawah permukaan yang dinyatakan dalam parameter fisis dan geometri tertentu. Melalui model, respons teoritis dapat dihitung menggunakan persamaan fisika sesuai metode geofisika yang digunakan.
</p>

<p>
Pemodelan geofisika terdiri atas dua proses utama, yaitu <i>forward modeling</i> dan <i>inverse modeling</i>. <i>Forward modeling</i> merupakan proses menghitung respons data berdasarkan model bawah permukaan yang telah diketahui. Sebaliknya, <i>inverse modeling</i> atau inversi merupakan proses memperkirakan parameter model berdasarkan data observasi. Dalam inversi, parameter model dicari sedemikian rupa sehingga respons hasil perhitungan memiliki kesesuaian terbaik dengan data pengamatan. Oleh karena itu, inversi sering disebut sebagai proses <i>data fitting</i>. Secara matematis, proses inversi dilakukan dengan meminimalkan fungsi objektif atau fungsi <i>misfit</i> yang menyatakan selisih antara data observasi dan data hasil perhitungan. Model optimum diperoleh ketika nilai <i>misfit</i> mencapai minimum.
</p>

<h3 style="text-align:left; font-size:13pt"><b>1.3.2 Inversi Non Linear</b></h3>

<p>
Inversi non-linear merupakan metode inversi yang digunakan ketika hubungan antara data observasi dan parameter model tidak dapat dinyatakan dalam bentuk persamaan linear. Secara umum hubungan tersebut dinyatakan sebagai:
</p>

$$
d = g(m)
$$

<p>
dengan d merupakan data observasi, m adalah parameter model, dan g(m) merupakan fungsi <i>forward modeling</i> non-linear.
</p>

<p>
Pada inversi non-linear, perubahan kecil pada parameter model tidak selalu menghasilkan perubahan linear pada data. Hal ini menyebabkan solusi model tidak dapat diperoleh secara langsung menggunakan invers matriks sederhana seperti pada inversi linear, melainkan memerlukan proses iteratif atau pencarian dalam ruang model. Karakteristik utama inversi non-linear adalah kemungkinan adanya banyak solusi lokal (<i>local minimum</i>) maupun solusi global (<i>global minimum</i>). Kondisi ini terjadi karena fungsi objektif dapat memiliki lebih dari satu nilai minimum.
</p>

<h3 style="text-align:left; font-size:13pt"><b>1.3.3 Fungsi Objektif dan Misfit Error</b></h3>

<p>
Dalam pemodelan inversi, tujuan utama yang ingin dicapai adalah memperoleh model yang mampu menghasilkan respons teoritis yang sesuai dengan data observasi.
</p>

Residual dinyatakan sebagai:

$$
r = d - g(m)
$$

Fungsi objektif:

$$
E = ||d - g(m)||^2
$$

Nilai <i>misfit error</i> menggunakan <i>Root Mean Square Error</i> (<i>RMSE</i>):

$$
RMSE = \sqrt{\frac{1}{N}\sum_{i=1}^{N}(d_i - g_i(m))^2}
$$

dengan:

<ul>
<li>N = jumlah data pengamatan</li>
<li>dᵢ = data observasi ke-i</li>
<li>gᵢ(m) = respons model pada data ke-i</li>
</ul>

<p>
Nilai <i>misfit</i> yang kecil menunjukkan bahwa model mampu merepresentasikan data observasi dengan baik.
</p>

<h3 style="text-align:left; font-size:13pt"><b>1.3.4 Metode Geolistrik Resistivitas</b></h3>
<p>
Metode geolistrik resistivitas merupakan salah satu metode geofisika yang digunakan untuk mengidentifikasi kondisi bawah permukaan berdasarkan perbedaan nilai tahanan jenis (resistivitas) material batuan. Prinsip dasar metode ini adalah menginjeksikan arus listrik ke dalam tanah melalui dua elektroda arus, kemudian mengukur beda potensial yang dihasilkan menggunakan dua elektroda potensial. Variasi nilai beda potensial yang terukur dipengaruhi oleh distribusi resistivitas material di bawah permukaan, sehingga dapat digunakan untuk menginterpretasi struktur geologi, litologi, zona alterasi, keberadaan air tanah, maupun rongga bawah tanah (Telford et al., 1990). Resistivitas suatu material adalah kemampuan suatu material dalam menghambat aliran arus listrik. Secara matematis, resistivitas dinyatakan sebagai:

$$
ρ = RA/L
$$

<p>
dengan:
</p>

<ul>
<li>ρ = resistivitas (ohm.m)</li>
<li>R = hambatan listrik (ohm)</li>
<li>A = luas penampang penghantar (m²)</li>
<li>L = panjang penghantar (m)</li>
</ul>

<p>
Dalam survei lapangan, medium bawah permukaan dianggap sebagai half-space homogen isotropik sehingga nilai resistivitas semu dapat dihitung menggunakan:
</p>

$$
ρ = K (ΔV / I)
$$

<ul>
<li>ρ<sub>a</sub> = resistivitas semu (ohm.m)</li>
<li>K = faktor geometri array</li>
<li>ΔV = beda potensial (volt)</li>
<li>I = arus listrik (ampere)</li>
</ul>

<p>
Disebut resistivitas semu karena nilai yang diperoleh merupakan representasi gabungan dari beberapa lapisan bawah permukaan yang dipengaruhi konfigurasi elektroda. Oleh karena itu, diperlukan proses inversi untuk memperoleh model resistivitas sebenarnya (true resistivity) dari data lapangan (Loke, 2004).Beberapa konfigurasi elektroda yang umum digunakan pada metode resistivitas antara lain konfigurasi Wenner, Schlumberger, Dipole-Dipole, dan Pole-Dipole. Konfigurasi Wenner memiliki sensitivitas yang baik terhadap variasi lateral dan vertikal dengan susunan elektroda berjarak sama. Schlumberger lebih efisien untuk sounding vertikal karena elektroda potensial relatif tetap, sedangkan Dipole-Dipole memiliki resolusi lateral yang baik untuk pemetaan struktur seperti sesar atau rongga bawah permukaan (Reynolds, 2011). Interpretasi data resistivitas didasarkan pada variasi nilai resistivitas tiap material. Material yang mengandung air dengan kadar ion tinggi umumnya memiliki resistivitas rendah, sedangkan batuan kompak dan kering cenderung memiliki resistivitas tinggi. Faktor-faktor yang memengaruhi resistivitas batuan meliputi porositas, saturasi fluida, kandungan mineral lempung, temperatur, serta salinitas air pori (Keller dan Frischknecht, 1966).
</p>

<p>
Pada praktikum ini, metode resistivitas diterapkan dalam bentuk <i>Vertical Electrical Sounding</i> (VES) 1D. Metode <i>Vertical Electrical Sounding</i> merupakan teknik akuisisi geolistrik yang bertujuan untuk mengetahui variasi resistivitas terhadap kedalaman dengan cara memperbesar spasi elektroda arus secara bertahap, sementara titik tengah pengukuran dipertahankan relatif tetap. Dengan bertambahnya jarak antar elektroda, kedalaman penetrasi arus listrik juga meningkat sehingga informasi yang diperoleh merepresentasikan lapisan bawah permukaan yang semakin dalam (Telford et al., 1990).
</p>
<p>
Pada pendekatan VES 1D, bawah permukaan diasumsikan tersusun atas beberapa lapisan horizontal homogen yang masing-masing memiliki nilai resistivitas dan ketebalan tertentu, kecuali lapisan terakhir yang dianggap tak hingga. Asumsi ini sesuai untuk daerah dengan dominasi perubahan resistivitas secara vertikal (Reynolds, 2011). Hasil pengukuran VES berupa kurva hubungan antara jarak elektroda (AB/2) dan resistivitas semu. Kurva ini digunakan untuk mengidentifikasi perubahan litologi terhadap kedalaman berdasarkan pola variasi resistivitas. Karena data yang diperoleh masih berupa resistivitas semu, maka diperlukan inversi numerik untuk memperoleh model resistivitas sebenarnya (<i>true resistivity</i>) dan ketebalan lapisan. Proses inversi dilakukan dengan membandingkan data observasi dan hasil <i>forward modeling</i> hingga diperoleh nilai <i>misfit</i> minimum (Loke, 2004).
</p>

<p>
Pada praktikum ini, data VES 1D digunakan sebagai input untuk proses inversi menggunakan algoritma <i>Particle Swarm Optimization</i> (PSO). Parameter model yang dicari meliputi resistivitas beberapa lapisan dan ketebalan lapisan, sedangkan kualitas model dievaluasi berdasarkan nilai Root Mean Square Error (RMSE) antara data observasi dan data kalkulasi. Dengan demikian, metode VES 1D memungkinkan estimasi struktur vertikal bawah permukaan secara lebih kuantitatif melalui integrasi pengukuran resistivitas dan optimasi model.
</p>

<h3 style="text-align:left; font-size:13pt"><b>1.3.6 Forward Modelling Geolistrik 1D</b></h3>
<p>
Forward modelling pada geolistrik 1D merupakan proses perhitungan respons teoritis resistivitas semu berdasarkan model bawah permukaan yang telah diasumsikan sebelumnya. Pada pendekatan ini, bawah permukaan direpresentasikan sebagai beberapa lapisan horizontal homogen, di mana setiap lapisan memiliki parameter resistivitas dan ketebalan tertentu, kecuali lapisan terakhir yang dianggap sebagai half-space dengan ketebalan tak hingga. Tujuan forward modelling adalah menghasilkan data sintetik yang dapat dibandingkan dengan data observasi lapangan. Dalam metode Vertical Electrical Sounding (VES), forward modelling menghitung nilai resistivitas semu untuk berbagai bentangan elektroda AB/2 menggunakan parameter model tertentu. Perhitungan ini dilakukan dengan menyelesaikan persamaan potensial listrik pada medium berlapis, sehingga diperoleh respons teoritis yang menggambarkan hubungan antara bentangan elektroda dan resistivitas semu. Pada implementasi numerik, proses ini umumnya dibantu filter digital, seperti filter Guptasarma, untuk mempercepat transformasi integral pada perhitungan resistivitas semu. Hasil forward modelling sangat penting dalam proses inversi karena menjadi dasar evaluasi model. Data hasil kalkulasi dibandingkan dengan data observasi menggunakan fungsi objektif, misalnya Root Mean Square Error (RMSE), untuk mengetahui tingkat kesesuaian model terhadap kondisi lapangan. Semakin kecil selisih antara data observasi dan hasil forward modelling, maka model dianggap semakin representatif terhadap struktur bawah permukaan sebenarnya.</p>


<h3 style="text-align:left; font-size:13pt"><b>1.3.7 Global Optimization</b></h3>
<p>Global optimization merupakan metode optimisasi yang bertujuan mencari solusi terbaik secara global pada suatu ruang parameter, sehingga tidak terjebak pada solusi lokal (local minimum). Dalam permasalahan inversi geofisika, global optimization banyak digunakan karena model bawah permukaan umumnya bersifat non-linier, multimodal, dan memiliki banyak kemungkinan solusi. Berbeda dengan optimisasi lokal yang sangat bergantung pada model awal, global optimization melakukan eksplorasi ruang solusi secara lebih luas sehingga peluang menemukan model optimum menjadi lebih besar. Pendekatan ini sangat sesuai untuk inversi geofisika, karena data observasi sering kali dapat dijelaskan oleh beberapa kombinasi parameter model yang berbeda. Beberapa algoritma yang termasuk dalam global optimization antara lain Genetic Algorithm (GA), Simulated Annealing (SA), Differential Evolution (DE), dan Particle Swarm Optimization (PSO). Pada praktikum ini, global optimization diterapkan untuk menentukan parameter resistivitas dan ketebalan lapisan yang menghasilkan misfit minimum antara data observasi dan data sintetik hasil forward modelling.</p>

<h3 style="text-align:left; font-size:13pt"><b>1.3.8 Particle Swarm Optimization (PSO)</b></h3>
<p>Particle Swarm Optimization (PSO) merupakan algoritma optimisasi berbasis populasi yang terinspirasi dari perilaku kolektif kawanan burung atau ikan saat mencari makanan. Algoritma ini pertama kali diperkenalkan oleh Kennedy dan Eberhart pada tahun 1995 sebagai metode pencarian solusi optimum pada ruang parameter multidimensi. Dalam PSO, setiap kandidat solusi direpresentasikan sebagai partikel yang bergerak dalam ruang pencarian. Masing-masing partikel memiliki posisi dan kecepatan yang diperbarui secara iteratif berdasarkan pengalaman terbaik individu (personal best atau Pbest) dan pengalaman terbaik seluruh populasi (global best atau Gbest). Mekanisme ini memungkinkan partikel melakukan keseimbangan antara eksplorasi ruang solusi secara global dan eksploitasi solusi terbaik yang telah ditemukan.</p>
<p>
Pada inversi geolistrik 1D, setiap partikel merepresentasikan kombinasi parameter model, misalnya resistivitas tiap lapisan dan ketebalan lapisan. Untuk setiap iterasi, partikel dihitung respons forward modelling-nya, kemudian dibandingkan dengan data observasi menggunakan fungsi objektif berupa RMSE. Nilai error tersebut digunakan untuk memperbarui Pbest dan Gbest. Keunggulan PSO terletak pada implementasinya yang sederhana, jumlah parameter kontrol yang relatif sedikit, serta kemampuan konvergensi yang baik dalam menyelesaikan permasalahan non-linier. Oleh karena itu, PSO banyak digunakan dalam inversi geofisika untuk memperoleh model bawah permukaan optimum dengan tingkat misfit minimum.</p>

```python
import numpy as np                          #import library numpy untuk operasi matematis
import matplotlib.pyplot as plt             #import library matplotlib untuk memvisualisasikan grafik

data = np.loadtxt(r'D:\smt 6\prak inversi\minggu 7\Data_Obs_Tanpa_Noise.txt')   #membaca data yang di input
ab2 = data[:, 0]                            #menyimpan ke variabel ab2
rho_apparent = data[:, 1]                   #menyimpan semua baris pada kolom kedua ke rho app

plt.figure(figsize=(6, 4))                  #plot grafik size 6x4
plt.plot(ab2, rho_apparent, '*-', color='blue',label='obs data')    #plot grafik ab2 dan rho app dengan keterngan marker, warna dan label
plt.xscale('log')                           #plot sumbu x ke log
plt.yscale('log')                           #plot sumbu y ke log
plt.xlabel('AB/2')                          #plot sumbu x menjadi AB/2
plt.ylabel('Rho app[ohm.m]')                #plot sumbu y menjadi Rho app
plt.grid(True, which='both', linestyle='--', linewidth=0.5) #plot grid pada grafik
plt.legend()                                #plot legenda
plt.title('Data Observasi Rho App vs AB/2') #plot judul
plt.tight_layout()                          #plot layout 
plt.show()                                  #plot visualisasi grafik
```

```python
bounds = [(10, 200), (10, 200), (10, 200), (1, 50), (1, 50)]    #batas minimum dan maksimum parameter model (3 resistivitas, 2 ketebalan)
num_particles = 100                                             #jumlah partikel dalam PSO
dimensions = 5                                                  #jumlah parameter yang dicari (3 rho + 2 thickness)
iterations = 100                                                #jumlah iterasi PSO
c1 = 1.5                                                        #koefisien pengaruh personal best
c2 = 1.5                                                        #koefisien pengaruh global best

def fungsi_objektif(rhoap_obs, rhoap_cal):                      #fungsi untuk menghitung error antara data observasi dan kalkulasi
    return np.sqrt(np.mean((rhoap_obs - rhoap_cal) ** (2)))     #menghitung nilai RMSE

def filter_coefficent(filter_coeff):                            #fungsi memilih koefisien filter digital
    if filter_coeff == 'guptasarma_7':                          #jika memakai filter guptasarma 7 titik
        a = np.array([-0.17445, 0.09672, 0.36789, 0.63906, 0.91023, 1.1814, 1.45257])   #nilai koefisien a
        phi = np.array([0.1732, 0.2945, 2.147, -2.1733, 0.6646, -0.1215, 0.0155])       #nilai koefisien phi

    elif filter_coeff == 'guptasarma_11':                       #jika memakai filter guptasarma 11 titik
        a = np.array([-0.420625, -0.20265625, 0.0153125, 0.23328125, 0.45125, 0.66921875, 0.8871875, 1.10515625, 1.323125, 1.54109375, 1.7590625])
        phi = np.array([0.041873, -0.022258, 0.38766, 0.647103, 1.84873, - 2.96084, 1.358412, -0.37759, 0.097107, -0.024243, 0.004046])

    elif filter_coeff == 'guptasarma_22':                       #jika memakai filter guptasarma 22 titik
        a = np.array([-0.980685, -0.771995, -0.563305, -0.354615, -0.145925, 0.062765, 0.271455, 0.480145,  0.688835,  0.897525,  1.106215, 1.314905, 1.523595, 1.732285, 1.940975, 2.149665,  2.358355,  2.567045, 2.775735])
        phi = np.array([0.00097112, -0.00102152, 0.00906965, 0.01404316, 0.09012,  0.30171582, 0.99627084, 1.3690832, -2.99681171, 1.65463068, -0.59399277,  0.22329813 , -0.10119309,  0.05186135, -0.02748647, 0.01384932, -0.00599074, 0.00190463, -0.0003216 ])
    return a, phi                                               #mengembalikan nilai a dan phi

def forward(ab2, rhotr, thick, a, phi):                         #fungsi forward modelling geolistrik 1D
    rhoap_cal = []                                              #list kosong untuk menyimpan rho apparent hasil perhitungan
    for L in ab2:                                               #looping tiap jarak AB/2
        layer = len(rhotr) - 1                                  #menentukan jumlah layer terakhir
        T = rhotr[-1]                                           #nilai resistivitas layer paling bawah
        T_phi = []                                              #list kosong hasil perkalian T dan phi

        for i in range(len(a)):                                 #looping setiap koefisien filter
            lambda_r = 10 ** (a[i] - np.log10(L))               #menghitung lambda
            temp_layer = layer                                  #menyimpan layer sementara
            temp_T = T                                          #menyimpan T sementara

            while temp_layer > 0:                               #perulangan dari bawah ke atas layer
                temp_layer -= 1                                 #naik satu layer
                val = lambda_r * thick[temp_layer]              #menghitung lambda x ketebalan
                val = np.clip(val, -50, 50)                     #membatasi nilai agar stabil
                tanh = np.tanh(val)                             #menghitung tanh
                tanh = np.tanh(lambda_r * thick[temp_layer])    #menghitung ulang tanh
                temp_T = (temp_T + rhotr[temp_layer] * tanh) / (1 + temp_T * tanh / rhotr[temp_layer]) #rekurensi resistivitas

            T_phi.append(temp_T * phi[i])                       #menyimpan hasil kali T dan phi

        rho_app = np.sum(T_phi)                                 #menjumlahkan seluruh kontribusi filter
        rhoap_cal.append(rho_app)                               #menyimpan rho apparent

    return np.array(rhoap_cal)                                  #mengembalikan rho apparent kalkulasi

def pso(bounds, num_particles, dimensions, ab2, rhoap_obs, iterations, c1=1.5, c2=1.5):   #fungsi utama PSO
    particles = []                                              #menyimpan posisi partikel
    velocities = []                                             #menyimpan kecepatan partikel
    pbest_positions = []                                        #menyimpan posisi terbaik tiap partikel
    pbest_scores = []                                           #menyimpan nilai error terbaik tiap partikel
    gbest_score = float('inf')                                  #nilai error global terbaik awal
    gbest_position = []                                         #posisi global terbaik
    erms = []                                                   #menyimpan error tiap iterasi

    a, phi = filter_coefficent('guptasarma_7')                  #memilih filter guptasarma 7

    for _ in range(num_particles):                              #membuat populasi awal partikel
        position = [np.random.uniform(bounds[d][0], bounds[d][1]) for d in range(dimensions)] #posisi acak awal
        velocity = np.zeros(dimensions)                         #kecepatan awal nol

        rhotr = position[:3]                                    #3 parameter pertama = resistivitas
        thick = position[3:]                                    #2 parameter terakhir = ketebalan
        rhoap_cal = forward(ab2, rhotr, thick, a, phi)          #forward modelling
        score = fungsi_objektif(rhoap_obs, rhoap_cal)           #hitung error

        particles.append(position)                              #simpan posisi partikel
        velocities.append(velocity)                             #simpan kecepatan
        pbest_positions.append(position[:])                     #simpan posisi terbaik awal
        pbest_scores.append(score)                              #simpan score terbaik awal

        if score < gbest_score:                                 #jika lebih baik dari global best
            gbest_score = score                                 #update global score
            gbest_position = position[:]                        #update global posisi

    for _ in range(iterations):                                 #looping iterasi utama
        for i in range(num_particles):                          #evaluasi semua partikel
            rhotr = particles[i][:3]                            #ambil resistivitas
            thick = particles[i][3:]                            #ambil ketebalan
            rhoap_cal = forward(ab2, rhotr, thick, a, phi)      #forward modelling
            score = fungsi_objektif(rhoap_obs, rhoap_cal)       #hitung error

            if score < pbest_scores[i]:                         #jika lebih baik dari personal best
                pbest_scores[i] = score                         #update score terbaik
                pbest_positions[i] = particles[i][:]            #update posisi terbaik

            if score < gbest_score:                             #jika lebih baik dari global best
                gbest_score = score                             #update global score
                gbest_position = particles[i][:]                #update global posisi

        for i in range(num_particles):                          #update kecepatan dan posisi
            for d in range(dimensions):                         #loop tiap dimensi parameter
                r1 = np.random.random()                         #bilangan acak 0-1
                r2 = np.random.random()                         #bilangan acak 0-1
                velocities[i][d] = (velocities[i][d] + c1 * r1 * (pbest_positions[i][d] - particles[i][d]) + c2 * r2 * (gbest_position[d] - particles[i][d])) #rumus update velocity
                particles[i][d] += velocities[i][d]             #update posisi
                particles[i][d] = max(bounds[d][0], min(bounds[d][1], particles[i][d])) #membatasi posisi sesuai bounds

        erms.append(gbest_score)                                #menyimpan error terbaik tiap iterasi

    return gbest_position, gbest_score, erms                    #mengembalikan model terbaik, misfit, dan error history

best_model, best_misfit, erms_results = pso(bounds, num_particles, dimensions, ab2, rho_apparent, iterations=100) #menjalankan PSO

print("Model terbaik:", best_model)                             #menampilkan model parameter terbaik
print("Misfit terbaik:", best_misfit)                           #menampilkan error terkecil
print(best_model[:3])                                           #menampilkan resistivitas layer
print(best_model[3:])                                           #menampilkan ketebalan layer
print(erms_results)                                             #menampilkan riwayat error tiap iterasi
```

```python
import matplotlib.pyplot as plt                        #import library matplotlib untuk memvisualisasikan grafik

rho_best = best_model[:3]                              #mengambil 3 parameter pertama dari model terbaik sebagai resistivitas layer
thick_best = best_model[3:]                            #mengambil parameter sisanya sebagai ketebalan layer

depth = [0]                                            #membuat list kedalaman awal dimulai dari 0 meter
for t in thick_best:                                   #looping setiap nilai ketebalan
    depth.append(depth[-1] + t)                        #menjumlahkan ketebalan sebelumnya untuk mendapatkan kedalaman kumulatif
depth.append(depth[-1])                                #menambahkan kedalaman terakhir agar grafik step tertutup

rho_plot = list(rho_best) + [rho_best[-1]]             #menambahkan nilai resistivitas terakhir agar jumlah data sesuai grafik step

plt.figure(figsize=(6, 5))                             #membuat ukuran grafik 6x5
plt.step(rho_plot, depth, where='post', label='Model Optimum', color='orange')  #memplot model berlapis dengan grafik tangga(step)

plt.gca().invert_yaxis()                               #membalik sumbu y agar kedalaman bertambah ke bawah

plt.xlabel(r'$\rho_{true}[\Omega m]$')                 #memberi nama sumbu x resistivitas
plt.ylabel('Depth [m]')                                #memberi nama sumbu y kedalaman
plt.title('Model Bawah Permukaan Hasil PSO')           #memberi judul grafik
plt.grid(True)                                         #menampilkan grid
plt.legend()                                           #menampilkan legenda
plt.tight_layout()                                     #mengatur layout agar rapi
plt.show()                                             #menampilkan grafik
```

<h1 style="text-align:center; font-size:20pt"><b>BAB IV ANALISIS</b></h1>

<p>Pada kode pertama data observasi dari resistivitas semu dibaca dan diplot dalam grafik. Pada grafik terlihat kurva sounding resistivitas, dengan sumbu x adalah spasi elektroda dan sumbu y adalah nilai resistivitas semu. Terlihat hasil observasi menunjukkan tren penurunan nilai resistivitas semu dari sekitar 100 Ωm pada jarak elektroda kecil hingga mendekati 10 Ωm pada jarak elektroda yang lebih besar. Pola ini mengindikasikan resistivitas bawah permukaan cenderung menurun seiring bertambahnya kedalaman atau bisa juga diindikasikan adanya perubahan litologi atau peningkatan kandungan air pada lapisan yang lebih dalam.</p>
<p>Setelah data observasi diperoleh, dilakukan proses inversi menggunakan metode Particle Swarm Optimization (PSO) untuk mencari model bawah permukaan yang representatif. Dalam proses ini, setiap partikel pada PSO merepresentasikan satu kandidat model berdasarkan tiga parameter resistivitas lapisan (ρ₁, ρ₂, ρ₃) dan dua parameter ketebalan (h₁, h₂). Dilakukan forward modelling disetiap model yang diuji untuk menghitung nilai resistivitas semu sintetik, lalu akan dibandingkan dengan data observasi menggunakan fungsi objektif (RMSE). Proses iteratif PSO akan memperbarui posisi partikel berdasarkan pengalaman terbaik masing-masing partikel (personal best) dan solusi terbaik global (global best). Hasil inversi menunjukkan model optimum dengan nilai resistivitas masing-masing sebesar 99,16 Ωm, 43,40 Ωm, dan 10 Ωm, serta ketebalan lapisan pertama sebesar 6,07 m dan lapisan kedua sebesar 22,61 m. Nilai misfit terbaik sebesar 1,08 menunjukkan bahwa model hasil inversi memiliki kecocokan yang sangat baik terhadap data observasi, karena selisih yang didapat kecil. Lapisan pertama yang memiliki resistivitas tinggi dapat diinterpretasikan sebagai material yang relatif kering atau kompak, seperti pasir kering atau tanah padat. Lapisan kedua dengan resistivitas menengah kemungkinan merupakan zona transisi, misalnya pasir lembap atau batuan yang mulai terpengaruh air. Sementara itu, lapisan ketiga yang memiliki resistivitas rendah mengindikasikan material yang lebih konduktif, seperti lempung atau zona jenuh air.</p>
<p>Setelah diplot, terlihat distribusi resistivitas terhadap kedalamannya. Dengan asumsi setiap lapisan bersifat homogen dan isotropik, resistivitas menurun seiring kedalaman makin dalam. Hal ini sama dengan hasil kurva sebelumnya yang dapat menjadi validasi bahwa hasil yang didapat sebelumnya sesuai baik dari data maupun geologinya. Optimasi PSO dicek dengan konvergensi error RMS terhadap iterasi dan menunjukkan adanya penurunan error pada awal iterasi, dari 6,3 menjadi 1,5. Hal ini menunjukkan bahwa algoritma PSO bisa menemukan model terbaik dari estimasi awal. Dan iterasi selanjutnya error turun hingga stabil di 1,08. Yang mana proses ini menunjukkan algoritma mencapai kodisi konvergen.</p>

<h1 style="text-align:center; font-size:20pt"><b>BAB V KESIMPULAN</b></h1>

Hasil inversi menunjukkan nilai resistivitas berturut-turut sebesar 99,16 Ωm, 43,40 Ωm, dan 10 Ωm, dengan ketebalan lapisan pertama sekitar 6,07 m dan lapisan kedua 22,61 m. Nilai resistivitas yang semakin menurun terhadap kedalaman mengindikasikan adanya perubahan karakter material bawah permukaan dari lapisan yang lebih resistif di bagian atas menuju lapisan yang lebih konduktif pada kedalaman yang lebih besar.</p>
